Bu operatörler ileride tek/çift sayı kontrolü, paketleme problemleri, sayaçlar ve algoritmalar gibi birçok yerde karşımıza çıkacaktır.

# 12. Liste Dilimleme (Slicing)

Bir listenin belirli bir bölümünü almak için dilimleme kullanabiliriz.

Genel yapı:

```python

```

`bitis` indeksi sonuca dahil edilmez.


````python
sayilar = [10, 20, 30, 40, 50, 60]

print(sayilar[1:4])
print(sayilar[:3])
print(sayilar[3:])
print(sayilar[:])
````

# 14. Listeye Eleman Ekleme

## `append()`

Listenin **sonuna bir eleman** ekler.

````python
meyveler = ["Elma", "Armut"]

meyveler.append("Muz")
print(meyveler)

meyveler.append("Çilek")
print(meyveler)
````

## `insert()`

Belirli bir indekse eleman eklemek için `insert()` kullanılır.

````python
meyveler = ["Elma", "Armut", "Muz"]

meyveler.insert(1, "Çilek")
print(meyveler)
````

Burada `1` indeksine `"Çilek"` ekledik. Mevcut elemanlar silinmedi; sağa doğru kaydırıldı.

## `extend()`

Başka bir listenin elemanlarını mevcut listenin sonuna eklemek için kullanılır.

````python
birinci_liste = [1, 2, 3]
ikinci_liste = [4, 5, 6]

birinci_liste.extend(ikinci_liste)
print(birinci_liste)
````

`append()` ile `extend()` arasındaki fark önemlidir:

```python
liste.append([4, 5])
```

listeye tek bir eleman olarak `[4, 5]` ekler.

```python
liste.extend([4, 5])
```

ise `4` ve `5` değerlerini ayrı elemanlar olarak ekler.


````python
liste1 = [1, 2, 3]
liste1.append([4, 5])
print("append:", liste1)

liste2 = [1, 2, 3]
liste2.extend([4, 5])
print("extend:", liste2)
````

# 15. Listeden Eleman Silme

## `remove()`

Belirli bir **değeri** silmek için kullanılır.

````python
meyveler = ["Elma", "Armut", "Muz", "Çilek"]

meyveler.remove("Muz")
print(meyveler)
````

## `pop()`

`pop()` indeks kullanarak eleman siler ve silinen elemanı bize geri döndürür.

## `clear()`

````python
sayilar = [10, 20, 30]
sayilar.clear()
print(sayilar)
````

# 16. Liste Elemanı Arama

Bir değerin listedeki ilk bulunduğu konumu verir.

````python
meyveler = ["Elma", "Armut", "Muz", "Çilek"]

print(meyveler.index("Muz"))
````

# 17. Liste Sıralama

## `sort()`

Listenin elemanlarını küçükten büyüğe veya alfabetik sıraya göre sıralayabiliriz.

````python
sayilar = [50, 10, 80, 30, 20]

sayilar.sort()
print(sayilar)
````

Büyükten küçüğe sıralamak için `reverse=True` kullanabiliriz.

````python
sayilar = [50, 10, 80, 30, 20]

sayilar.sort(reverse=True)
print(sayilar)
````

Metinlerde de alfabetik sıralama yapılabilir.

````python
isimler = ["Zeynep", "Ali", "Mehmet", "Ayşe"]
isimler.sort()
print(isimler)
````

## `sorted()`

`sort()` listenin kendisini değiştirir.

`sorted()` ise sıralanmış yeni bir sonuç üretir ve orijinal listeyi değiştirmeden kullanabiliriz.

````python
sayilar = [50, 10, 80, 30, 20]

sirali = sorted(sayilar)

print("Orijinal:", sayilar)
print("Sıralı:", sirali)
````

# 18. Listeyi Ters Çevirme: `reverse()`

`reverse()` listenin mevcut sırasını tersine çevirir.

````python
sayilar = [10, 20, 30, 40, 50]
sayilar.reverse()
print(sayilar)
````

Burada önemli bir ayrım var:

- `sort()` → sıralama yapar.
- `reverse()` → mevcut sırayı ters çevirir.
- `sorted()` → sıralanmış yeni bir liste döndürür.

Bu fonksiyonların birbirinden farklı amaçları vardır.


# 19. Liste Kopyalama

Şimdi önemli bir konuya geliyoruz.

Bir listeyi:

```python
liste2 = liste1
```

şeklinde yazarsak bağımsız bir kopya oluşturmuş olmayız. İki değişken aynı liste nesnesini gösterebilir.

Kaynak materyalde bu durum örneklenmiş ve gerçek bir kopya için `copy()` kullanılması gösterilmiştir. fileciteturn7file0L951-L999


````python
liste1 = [10, 20, 30]
liste2 = liste1

liste2.append(40)

print("liste1:", liste1)
print("liste2:", liste2)
````

Bağımsız bir kopya için:

````python
liste1 = [10, 20, 30]
liste2 = liste1.copy()

liste2.append(40)

print("liste1:", liste1)
print("liste2:", liste2)
````

# 20. Sayısal Listelerle Hazır Fonksiyonlar

Python'da bazı hazır fonksiyonlar sayısal listeler üzerinde çok kullanışlıdır.

````python
notlar = [70, 85, 90, 65, 80]

print("Toplam:", sum(notlar))
print("Eleman sayısı:", len(notlar))
print("En yüksek:", max(notlar))
print("En düşük:", min(notlar))
print("Ortalama:", sum(notlar) / len(notlar))
````

Bu fonksiyonları veri analizi derslerinde çok daha fazla kullanacağız.

Örneğin bir öğrencinin sınav notları:

```python
notlar = [70, 85, 90, 65, 80]
```

ise ortalama:

```python
sum(notlar) / len(notlar)
```

ile hesaplanabilir.

Daha sonra Pandas ve NumPy'ye geçtiğimizde aynı fikirleri çok daha büyük veri kümeleri üzerinde uygulayacağız. NumPy'nin sayısal hesaplama ve çok boyutlu veri için temel araçlardan biri olması da bu geçişin neden önemli olduğunu gösterir. fileciteturn7file1L1088-L1123


# 21. Listelerde Döngü Mantığına Hazırlık

Döngüleri henüz ayrıntılı olarak öğrenmedik. Ancak listelerin neden önemli olduğunu göstermek için basit bir örnek görelim.

Bir listenin her elemanını tek tek ele almak istediğimizde `for` döngüsünü kullanacağız.



````python
meyveler = ["Elma", "Armut", "Muz", "Çilek"]

for meyve in meyveler:
    print(meyve)
````

Bu kodu şimdilik şöyle okuyabiliriz:

> `meyveler` listesindeki her elemanı sırayla al ve o elemana `meyve` adını vererek ekrana yazdır.

`for` konusunu bir sonraki derslerde ayrıntılı şekilde inceleyeceğiz.

# 23. Mini Proje — Öğrenci Bilgi ve Not Sistemi

Şimdi öğrendiğimiz kavramları bir araya getirelim.

Bir öğrencinin:

- adı,
- soyadı,
- sınıfı,
- yaşı,
- ders notları

bilgilerini program içinde tutalım.

Notları liste olarak saklayalım ve ortalamasını hesaplayalım.


````python
ad = "Ayşe"
soyad = "Kaya"
sinif = "8-B"
yas = 13

notlar = [85, 90, 78, 95]

ortalama = sum(notlar) / len(notlar)

print("==============================")
print("       ÖĞRENCİ BİLGİLERİ")
print("==============================")
print(f"Ad Soyad : {ad} {soyad}")
print(f"Yaş      : {yas}")
print(f"Sınıf    : {sinif}")
print(f"Notlar   : {notlar}")
print(f"Ortalama : {ortalama:.2f}")
print("==============================")
````

Buradaki `:.2f` ifadesi ortalamayı virgülden sonra iki basamak göstermek için kullanılır.
