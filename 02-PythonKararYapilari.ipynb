## 1. Karar Yapıları Neden Gereklidir?

Bir programın bazı durumlarda farklı davranması gerekir.

Örneğin:

- Bir öğrencinin notu 50 veya üzerindeyse başarılı kabul etmek
- Yaş 18 veya üzerindeyse yetişkin kabul etmek
- Bir sayının pozitif, negatif veya sıfır olduğunu belirlemek
- Kullanıcının girdiği parolanın doğru olup olmadığını kontrol etmek
- Hava sıcaklığına göre bir mesaj göstermek
- Bir ürünün stokta olup olmadığını kontrol etmek

Bu problemlerin tamamında programın bir koşulu değerlendirmesi gerekir.

Örneğin:

> Eğer not 50 veya daha büyükse başarılı yaz.

Python'da bu düşünce `if` yapısı ile ifade edilir.


## 2. `if` Yapısı

En temel karar yapısı `if` yapısıdır.

Genel yazım:

```python
if koşul:
    yapılacak_işlem
```

Burada iki önemli nokta vardır:

1. `if` sonrasında bir koşul bulunur.
2. Koşulun sonunda `:` bulunur.

Koşul doğruysa alt satırlardaki girintili kod çalışır. Koşul yanlışsa bu kod çalıştırılmaz.

Örnek:


````python
sayi = 10

if sayi > 0:
    print("Sayı pozitiftir.")
````

Burada `sayi > 0` ifadesi değerlendirilir.

`10 > 0` doğru olduğu için ekrana:

```text
Sayı pozitiftir.
```

yazılır.

Şimdi farklı bir değer deneyelim.


````python
sayi = -5

if sayi > 0:
    print("Sayı pozitiftir.")

print("Program devam ediyor.")
````

Bu örnekte koşul yanlış olduğu için `print("Sayı pozitiftir.")` çalışmaz.

Ancak programın devamındaki:

```python
print("Program devam ediyor.")
```

çalışır.

Bu ayrım önemlidir: `if` bloğunun çalışmaması programın tamamının durduğu anlamına gelmez.


## 3. Karşılaştırma Operatörleri

Karar yapılarında çoğunlukla karşılaştırma operatörleri kullanılır.

| Operatör | Anlamı |
|---|---|
| `==` | eşittir |
| `!=` | eşit değildir |
| `>` | büyüktür |
| `<` | küçüktür |
| `>=` | büyük veya eşittir |
| `<=` | küçük veya eşittir |

Önemli:

`=` atama operatörüdür.

`==` karşılaştırma operatörüdür.

Örneğin:

```python
yas = 15
```

değişkene değer atar.

```python
yas == 15
```

ise `yas` değişkeninin 15'e eşit olup olmadığını kontrol eder.


````python
yas = 15

print(yas == 15)
print(yas != 15)
print(yas > 10)
print(yas < 10)
print(yas >= 15)
print(yas <= 15)
````

Karşılaştırmaların sonucu `True` veya `False` olur.

Bunlara boolean değerler denir.

```python
True
False
```

Karar yapıları bu sonuçları kullanır.


## 4. `if - else` Yapısı

Bazen koşul doğru olduğunda bir işlem, yanlış olduğunda başka bir işlem yapmak isteriz.

Bu durumda `else` kullanılır.

Genel yapı:

```python
if koşul:
    koşul_doğruysa
else:
    koşul_yanlışsa
```

Örnek:


````python
notu = 70

if notu >= 50:
    print("Öğrenci başarılı.")
else:
    print("Öğrenci başarısız.")
````

Şimdi aynı programı farklı notlarla inceleyelim.


````python
notu = 35

if notu >= 50:
    print("Öğrenci başarılı.")
else:
    print("Öğrenci başarısız.")
````

Burada iki seçenekten yalnızca biri çalışır.

Koşul doğruysa `if` bloğu, yanlışsa `else` bloğu çalışır.


## 5. Kullanıcıdan Veri Alarak Karar Vermek

Karar yapılarını kullanıcıdan alınan bilgilerle kullanmak programları daha etkileşimli hale getirir.

Örneğin kullanıcıdan yaş bilgisini alalım.

`input()` fonksiyonu metin döndürür. Sayısal karşılaştırma yapabilmek için `int()` ile tam sayıya dönüştürmemiz gerekir.


````python
yas = int(input("Yaşınızı giriniz: "))

if yas >= 18:
    print("18 yaş ve üzerindesiniz.")
else:
    print("18 yaşından küçüksünüz.")
````

Burada programın çalışma sırası şöyledir:

1. Kullanıcıdan yaş bilgisi alınır.
2. Girilen metin `int()` ile sayıya dönüştürülür.
3. `yas >= 18` koşulu kontrol edilir.
4. Sonuca göre iki bloktan biri çalıştırılır.


## 6. `if - elif - else`

Bir problemde ikiden fazla seçenek varsa `elif` kullanılır.

Örneğin bir notu üç farklı seviyede değerlendirebiliriz:

- 85 ve üzeri: Çok iyi
- 50-84 arası: Başarılı
- 50'nin altı: Geliştirilmeli

Genel yapı:

```python
if koşul_1:
    işlem_1
elif koşul_2:
    işlem_2
else:
    işlem_3
```


````python
notu = 78

if notu >= 85:
    print("Çok iyi")
elif notu >= 50:
    print("Başarılı")
else:
    print("Geliştirilmeli")
````

Python koşulları yukarıdan aşağıya kontrol eder.

`notu = 78` olduğunda:

- `78 >= 85` yanlış
- `78 >= 50` doğru

olduğu için ikinci blok çalışır ve program diğer koşulları değerlendirmeye devam etmez.


## 7. Birden Fazla `elif` Kullanmak

İhtiyaca göre birden fazla `elif` kullanılabilir.

Örneğin sınav notunu daha ayrıntılı değerlendirelim:


````python
notu = 92

if notu >= 90:
    print("AA")
elif notu >= 80:
    print("BA")
elif notu >= 70:
    print("BB")
elif notu >= 60:
    print("CB")
elif notu >= 50:
    print("CC")
else:
    print("FF")
````

Burada koşulların sırası çok önemlidir.

En yüksek aralıktan başlayarak aşağı doğru ilerlemek, bu örneğin okunmasını kolaylaştırır.

Örneğin `notu = 85` ise:

- `85 >= 90` yanlış
- `85 >= 80` doğru

olduğu için sonuç `BA` olur.


````python
notu = 85

if notu >= 90:
    print("AA")
elif notu >= 80:
    print("BA")
elif notu >= 70:
    print("BB")
elif notu >= 60:
    print("CB")
elif notu >= 50:
    print("CC")
else:
    print("FF")
````

## 8. Koşulların Sırası Neden Önemlidir?

Aşağıdaki kodu inceleyelim:

```python
notu = 95

if notu >= 50:
    print("Başarılı")
elif notu >= 90:
    print("Çok başarılı")
```

Bu kod teknik olarak çalışır. Ancak mantıksal olarak istediğimiz sonucu vermez.

Çünkü `95 >= 50` zaten doğrudur. Python ilk doğru koşulda ilgili bloğu çalıştırır ve `elif` kısmına geçmez.

Bu nedenle daha özel koşullar önce, daha genel koşullar sonra yazılmalıdır.


````python
notu = 95

if notu >= 90:
    print("Çok başarılı")
elif notu >= 50:
    print("Başarılı")
else:
    print("Başarısız")
````

## 9. Mantıksal Operatörler

Birden fazla koşulu birleştirmek için mantıksal operatörler kullanılır.

### `and`

İki koşulun da doğru olması gerekir.

```python
yas >= 18 and yas <= 30
```

### `or`

Koşullardan en az birinin doğru olması yeterlidir.

```python
gun == "Cumartesi" or gun == "Pazar"
```

### `not`

Bir koşulun sonucunu tersine çevirir.

```python
not aktif
```


````python
yas = 25

if yas >= 18 and yas <= 30:
    print("Yaş 18 ile 30 arasındadır.")
````

````python
gun = "Pazar"

if gun == "Cumartesi" or gun == "Pazar":
    print("Hafta sonu")
else:
    print("Hafta içi")
````

````python
aktif = False

if not aktif:
    print("Sistem aktif değil.")
````

## 10. Birden Fazla Koşulu Birlikte Kullanmak

Gerçek programlarda daha karmaşık koşullar kurmamız gerekebilir.

Örneğin bir öğrencinin bir etkinliğe katılabilmesi için:

- yaşının en az 12 olması
- kayıt durumunun aktif olması

gerektiğini düşünelim.


````python
yas = 14
kayit_aktif = True

if yas >= 12 and kayit_aktif:
    print("Etkinliğe katılabilirsiniz.")
else:
    print("Etkinliğe katılamazsınız.")
````

Burada iki koşulun da doğru olması gerekir.

Yaş yeterli olsa bile kayıt aktif değilse sonuç olumsuz olur.


````python
yas = 14
kayit_aktif = False

if yas >= 12 and kayit_aktif:
    print("Etkinliğe katılabilirsiniz.")
else:
    print("Etkinliğe katılamazsınız.")
````

## 11. Metinlerle Karar Yapıları

Karşılaştırmalar yalnızca sayılarla yapılmaz.

Metinleri de karşılaştırabiliriz.


````python
kullanici_adi = "ersin"

if kullanici_adi == "ersin":
    print("Kullanıcı adı doğru.")
else:
    print("Kullanıcı adı yanlış.")
````

Metin karşılaştırmalarında büyük-küçük harf farkı vardır.

Örneğin:

```python
"Python" == "python"
```

sonucu `False` olur.


````python
print("Python" == "python")
print("Python" == "Python")
````

Kullanıcıdan gelen metinlerde `lower()` veya `upper()` gibi string metotları kullanılarak karşılaştırma daha kontrollü yapılabilir.


````python
cevap = input("Python öğrenmek istiyor musunuz? ")

if cevap.lower() == "evet":
    print("Derse devam ediyoruz.")
else:
    print("Yine de Python öğrenmeye devam edeceğiz.")
````

## 12. Sayının Pozitif, Negatif veya Sıfır Olduğunu Bulmak

Karar yapılarını kullanabileceğimiz klasik problemlerden biri bir sayının durumunu belirlemektir.


````python
sayi = -8

if sayi > 0:
    print("Pozitif")
elif sayi < 0:
    print("Negatif")
else:
    print("Sıfır")
````

Burada üç farklı durum vardır ve `if - elif - else` yapısı bu problem için uygundur.


## 13. İç İçe Karar Yapıları

Bir `if` bloğunun içinde başka bir `if` kullanılabilir.

Buna iç içe karar yapısı denir.

Örneğin önce kullanıcının yaşını, sonra öğrenci olup olmadığını kontrol edebiliriz.


````python
yas = 16
ogrenci = True

if yas >= 18:
    print("Yetişkin kullanıcı.")
else:
    if ogrenci:
        print("18 yaşından küçük öğrenci.")
    else:
        print("18 yaşından küçük ve öğrenci değil.")
````

İç içe karar yapıları bazı problemlerde gerekli olabilir. Ancak çok fazla iç içe yapı kodun okunmasını zorlaştırabilir.

Bu nedenle aynı problem daha sade bir koşulla ifade edilebiliyorsa sade olan çözüm tercih edilmelidir.


## 14. Boolean Değerleri Doğrudan Kullanmak

Bazı değişkenler zaten `True` veya `False` değerleri taşır.

Bu durumda aşağıdaki kullanım mümkündür:


````python
giris_yapildi = True

if giris_yapildi:
    print("Ana sayfaya yönlendiriliyor.")
else:
    print("Lütfen giriş yapınız.")
````

Şu kullanım da mümkündür:

```python
if giris_yapildi == True:
```

Ancak boolean değişkenlerde çoğu zaman doğrudan değişkeni kullanmak daha okunaklıdır.

## 18. Bir Problemi Koda Dönüştürme

Program yazarken doğrudan kod yazmaya başlamak yerine problemi önce parçalara ayırmak faydalıdır.

Örnek problem:

> Kullanıcının girdiği sayının 100'den büyük, 100'e eşit veya 100'den küçük olduğunu ekrana yazdır.

Önce seçenekleri belirleriz:

1. 100'den büyük
2. 100'e eşit
3. 100'den küçük

Sonra koşulları oluştururuz.


````python
sayi = 135

if sayi > 100:
    print("Sayı 100'den büyüktür.")
elif sayi == 100:
    print("Sayı 100'e eşittir.")
else:
    print("Sayı 100'den küçüktür.")
````

Bu yaklaşımı daha sonra döngüler, fonksiyonlar ve veri yapılarıyla birleştireceğiz.

Amaç yalnızca `if` yazmayı öğrenmek değil, bir problemi analiz edip uygun karar yapısını seçebilmektir.


## 19. Daha Fazla Örnek

### Örnek: Sıcaklık kontrolü


````python
sicaklik = 32

if sicaklik >= 30:
    print("Hava sıcak.")
elif sicaklik >= 20:
    print("Hava ılık.")
else:
    print("Hava serin.")
````

### Örnek: Çift veya tek sayı


````python
sayi = 17

if sayi % 2 == 0:
    print("Çift sayı")
else:
    print("Tek sayı")
````

Burada `%` operatörü bölümden kalanı verir.

Bir sayının 2 ile bölümünden kalan `0` ise sayı çifttir.


### Örnek: Üç sayıdan en büyüğünü bulmak


````python
a = 18
b = 25
c = 12

if a >= b and a >= c:
    print("En büyük sayı:", a)
elif b >= a and b >= c:
    print("En büyük sayı:", b)
else:
    print("En büyük sayı:", c)
````

### Örnek: Kargo ücreti


````python
siparis_tutari = 750

if siparis_tutari >= 1000:
    print("Kargo ücretsiz.")
elif siparis_tutari >= 500:
    print("Kargo ücreti 50 TL.")
else:
    print("Kargo ücreti 100 TL.")
````

### Örnek: Kullanıcı rolü


````python
rol = "ogretmen"

if rol == "yonetici":
    print("Tüm yönetim işlemlerine erişebilirsiniz.")
elif rol == "ogretmen":
    print("Ders ve öğrenci işlemlerine erişebilirsiniz.")
elif rol == "ogrenci":
    print("Ders içeriklerine erişebilirsiniz.")
else:
    print("Tanımsız kullanıcı rolü.")
````
