Bu tür durumlarda programın tamamen kontrolsüz biçimde durması yerine hatayı yakalayıp uygun bir işlem yapmasını isteyebiliriz.

Python'da hata yönetiminin temel araçları:

- `try`

## 1. Hata Nedir?

Python kodu çalıştırıldığında programın normal akışını durduran beklenmeyen durumlara hata diyebiliriz.

Örneğin:


````python
sayi = 10
sonuc = 10 / 0
print(sonuc)
````

Bu kod çalıştırıldığında `ZeroDivisionError` oluşur.

Python bize bir hata mesajı gösterir ve programın o noktadaki çalışması durur.

Şimdi başka bir örneğe bakalım:


````python
sayi = int("merhaba")
print(sayi)
````

Burada `"merhaba"` metni tam sayıya dönüştürülemediği için `ValueError` oluşur.

Bu örnekler bize önemli bir şey gösterir:

Bir programın yalnızca doğru verilerle çalışacağını varsaymak güvenli değildir.

Özellikle kullanıcıdan veri alan programlarda hataları kontrol etmek gerekir.


## 2. `try` ve `except`

Hata oluşturabilecek kodu `try` bloğuna, hatanın nasıl ele alınacağını `except` bloğuna yazabiliriz.

Genel yapı:

```python
try:
    hata oluşabilecek kod
except:
    hata oluştuğunda çalışacak kod
```

Örnek:


````python
try:
    sayi = int(input("Bir sayı girin: "))
    print("Girdiğiniz sayı:", sayi)
except:
    print("Geçerli bir sayı girmelisiniz.")
````

Kullanıcı `25` girerse program normal şekilde devam eder.

Kullanıcı `merhaba` girerse `int()` dönüşümü başarısız olur ve `except` bloğu çalışır.

Bu yapı programın kullanıcı hatası nedeniyle kontrolsüz biçimde sonlanmasını engelleyebilir.


## 3. Belirli Hata Türlerini Yakalamak

`except` bölümünü genel bırakmak yerine hangi hatayı beklediğimizi belirtmek daha doğru bir yaklaşımdır.

Örneğin sayı dönüşümünde oluşabilecek hata `ValueError` olabilir.


````python
try:
    sayi = int(input("Bir sayı girin: "))
    print("Sayı:", sayi)
except ValueError:
    print("Lütfen geçerli bir tam sayı girin.")
````

Böylece yalnızca `ValueError` durumunu ele almış oluruz.

Belirli hata türlerini kullanmak, programdaki farklı hata durumlarını birbirinden ayırmamızı sağlar.


## 4. `ZeroDivisionError`

Bir sayıyı sıfıra bölmek Python'da `ZeroDivisionError` oluşturur.


````python
try:
    sayi = int(input("Bir sayı girin: "))
    sonuc = 10 / sayi
    print("Sonuç:", sonuc)
except ZeroDivisionError:
    print("Bir sayı sıfıra bölünemez!")
````

Kullanıcı `0` girerse bölme işlemi gerçekleştirilemez ve `ZeroDivisionError` yakalanır.

Kullanıcı `2` girerse:

```text
Sonuç: 5.0
```

elde edilir.


## 5. Birden Fazla `except`

Aynı programda farklı hata türleri için farklı mesajlar gösterebiliriz.

Şimdi derste mutlaka bilmemiz gereken temel örneği inceleyelim:


````python
try:
    sayi = int(input("Bir sayı girin: "))
    sonuc = 10 / sayi
    print("Sonuç:", sonuc)

except ZeroDivisionError:
    print("Bir sayı sıfıra bölünemez!")

except ValueError:
    print("Lütfen sadece geçerli bir sayı girin!")

finally:
    print("İşlem bitti.")
````

Bu örnekte üç farklı bölüm vardır.

### `try`

Normal olarak çalışmasını istediğimiz ve hata oluşabilecek kod burada bulunur.

### `except ZeroDivisionError`

Kullanıcı `0` girerse oluşan sıfıra bölme hatasını yakalar.

### `except ValueError`

Kullanıcı `10` yerine `merhaba` gibi sayı olmayan bir değer girerse `int()` dönüşümü sırasında oluşan `ValueError` hatasını yakalar.

### `finally`

Hata oluşsa da oluşmasa da çalışmasını istediğimiz kod burada bulunur.

Bu nedenle:

```python
print("İşlem bitti.")
```

mesajı her iki durumda da gösterilir.


## 6. Örneği Farklı Girdilerle İnceleyelim

### Kullanıcı 2 girerse

`10 / 2` işlemi yapılır ve sonuç gösterilir.

Ardından `finally` çalışır.

### Kullanıcı 0 girerse

`10 / 0` işlemi yapılamaz.

`ZeroDivisionError` çalışır.

Ardından `finally` çalışır.

### Kullanıcı `abc` girerse

`int("abc")` işlemi yapılamaz.

`ValueError` çalışır.

Ardından `finally` çalışır.

Bu örnek, hata yönetiminin mantığını anlamak için çok önemlidir.


````python
# Aynı örneği farklı girişlerle deneyebilirsiniz.

try:
    sayi = int(input("Bir sayı girin: "))
    sonuc = 10 / sayi
    print("Sonuç:", sonuc)

except ZeroDivisionError:
    print("Bir sayı sıfıra bölünemez!")

except ValueError:
    print("Lütfen sadece geçerli bir sayı girin!")

finally:
    print("İşlem bitti.")
````

## 7. `finally` Ne Zaman Kullanılır?

`finally`, işlemin sonucu ne olursa olsun çalışması gereken kodlar için kullanılır.

Örneğin:

- dosya kapatma
- bağlantı kapatma
- geçici işlemleri sonlandırma
- kullanıcıya işlem tamamlandı mesajı verme

gibi durumlarda kullanılabilir.

Basit örnek:


````python
try:
    print("İşlem başladı.")
    sayi = int("25")
    print("Sayı:", sayi)
finally:
    print("İşlem tamamlandı.")
````

Burada hata oluşmasa da `finally` çalışır.

Hata oluşan bir durumda da `finally` bölümünün çalıştırıldığını görebiliriz.


````python
try:
    print("İşlem başladı.")
    sayi = int("abc")
    print("Sayı:", sayi)
except ValueError:
    print("Sayıya dönüştürme başarısız.")
finally:
    print("İşlem tamamlandı.")
````

## 8. `else` ile `try`

`try` yapısında `else` de kullanılabilir.

`else`, `try` bloğunda hata oluşmadığında çalışır.

Genel yapı:

```python
try:
    işlem
except HataTürü:
    hata işlemi
else:
    hata oluşmadığında yapılacak işlem
finally:
    her durumda yapılacak işlem
```

Örnek:


````python
try:
    sayi = int(input("Bir sayı girin: "))
except ValueError:
    print("Geçersiz sayı.")
else:
    print("Başarıyla sayı girdiniz:", sayi)
finally:
    print("İşlem tamamlandı.")
````

Burada:

- `try`: sayıyı okumayı dener.
- `except`: dönüşüm başarısızsa çalışır.
- `else`: dönüşüm başarılıysa çalışır.
- `finally`: her durumda çalışır.

`else` özellikle hata oluşmayan durumdaki normal akışı hata yönetiminden ayırmak için kullanılabilir.


## 9. Hata Türlerini Tanımak

Python'da karşılaşacağımız bazı yaygın hata türleri:

| Hata | Açıklama |
|---|---|
| `ValueError` | Uygun olmayan değer |
| `TypeError` | Uygun olmayan veri türüyle işlem |
| `ZeroDivisionError` | Sıfıra bölme |
| `IndexError` | Listenin olmayan indeksine erişme |
| `KeyError` | Sözlükte olmayan anahtara erişme |
| `FileNotFoundError` | Bulunmayan dosyaya erişme |
| `NameError` | Tanımlanmamış isim/değişken kullanma |

Hangi hatanın oluştuğunu anlamak, doğru `except` bloğunu yazmamızı sağlar.


## 10. `TypeError` Örneği

Farklı veri türleriyle uygun olmayan bir işlem yapıldığında `TypeError` oluşabilir.


````python
try:
    sonuc = "10" + 5
    print(sonuc)
except TypeError:
    print("Bu işlem için veri türleri uygun değil.")
````

Burada bir metin ile bir tam sayı doğrudan `+` operatörüyle birleştirilmeye çalışılmıştır.

Doğru yaklaşım, yapmak istediğimiz işleme göre veri türlerini uygun hale getirmektir.


## 11. `IndexError` Örneği

Listelerde olmayan bir indekse erişmeye çalışırsak `IndexError` oluşur.


````python
sayilar = [10, 20, 30]

try:
    print(sayilar[5])
except IndexError:
    print("Listede böyle bir indeks bulunmuyor.")
````

Liste indeksleri `0` değerinden başlar.

Bu listede:

- `sayilar[0]` -> 10
- `sayilar[1]` -> 20
- `sayilar[2]` -> 30

vardır.

`5` numaralı indeks olmadığı için hata oluşur.


## 12. `KeyError` Örneği

Sözlüklerde bulunmayan bir anahtara doğrudan erişmeye çalışmak `KeyError` oluşturabilir.


````python
ogrenci = {
    "ad": "Ali",
    "sinif": 8
}

try:
    print(ogrenci["yas"])
except KeyError:
    print("Sözlükte bu anahtar bulunmuyor.")
````

Daha ileride sözlüklerle çalışırken `get()` metodunun da bu tür durumlarda nasıl kullanılabileceğini göreceğiz.


## 13. Hata Yönetimi ve Kullanıcı Girdileri

Özellikle `input()` kullanılan programlarda hata yönetimi çok önemlidir.

Örneğin kullanıcıdan yaş alırken:

```python
yas = int(input("Yaşınızı girin: "))
```

kullanıcı sayı yerine metin girerse program hata verir.

Bunu güvenli hale getirebiliriz.


````python
try:
    yas = int(input("Yaşınızı girin: "))

    if yas >= 18:
        print("Yetişkinsiniz.")
    else:
        print("18 yaşından küçüksünüz.")

except ValueError:
    print("Yaş bilgisi sayı olarak girilmelidir.")
````

Burada hata yönetimi ile karar yapısını birlikte kullandık.

Bu, önceki derste öğrendiğimiz `if` yapılarının gerçek programlarda nasıl başka konularla birleştiğini gösterir.


## 14. Hata Yönetimini Döngü ile Birleştirmek

Bir kullanıcıdan doğru veri gelene kadar tekrar giriş istemek için hata yönetimini ileride döngülerle birlikte kullanacağız.

Örneğin:


````python
while True:
    try:
        sayi = int(input("Bir sayı girin: "))
        print("Girdiğiniz sayı:", sayi)
        break
    except ValueError:
        print("Geçersiz giriş. Lütfen tekrar deneyin.")
````

Bu örnekte:

1. Program kullanıcıdan sayı ister.
2. Kullanıcı geçerli sayı girerse `break` ile döngüden çıkar.
3. Geçersiz veri girerse `ValueError` yakalanır.
4. Kullanıcıdan tekrar veri istenir.

Bu yapı daha sonra kullanıcıdan veri alan uygulamalarımızda çok sık kullanılacaktır.


## 15. Hataları Gizlemek Neden Doğru Değildir?

Aşağıdaki gibi her hatayı yakalamak mümkün olsa da:

```python
try:
    ...
except:
    print("Bir hata oldu.")
```

hangi hatanın oluştuğunu anlamak zorlaşabilir.

Daha kontrollü bir yaklaşım:

```python
except ValueError:
```

veya:

```python
except ZeroDivisionError:
```

gibi beklenen hata türlerini açıkça belirtmektir.

Böylece programın hangi problemi nasıl ele aldığı daha anlaşılır olur.


## 16. Birden Fazla Hata Türünü Birlikte Yakalamak

Bazı durumlarda iki farklı hata için aynı işlemi yapmak isteyebiliriz.

Bunun için hata türlerini bir demet içinde belirtebiliriz.


````python
try:
    sayi = int(input("Bir sayı girin: "))
    sonuc = 100 / sayi
    print("Sonuç:", sonuc)
except (ValueError, ZeroDivisionError):
    print("Giriş veya işlem sırasında bir hata oluştu.")
````

Burada hem:

- sayı yerine geçersiz bir değer girilmesi
- sıfıra bölme

durumları aynı `except` bloğunda ele alınmaktadır.

Ancak kullanıcıya farklı hata mesajları vermek istiyorsak ayrı `except` blokları daha uygundur.


## 17. Gerçek Uygulama Örneği: Basit Hesap Makinesi

Hata yönetimini basit bir hesap makinesinde kullanalım.


````python
try:
    sayi1 = float(input("Birinci sayıyı girin: "))
    sayi2 = float(input("İkinci sayıyı girin: "))
    islem = input("İşlem seçin (+, -, *, /): ")

    if islem == "+":
        sonuc = sayi1 + sayi2
    elif islem == "-":
        sonuc = sayi1 - sayi2
    elif islem == "*":
        sonuc = sayi1 * sayi2
    elif islem == "/":
        sonuc = sayi1 / sayi2
    else:
        print("Geçersiz işlem.")
        sonuc = None

    if sonuc is not None:
        print("Sonuç:", sonuc)

except ValueError:
    print("Lütfen sayıları doğru biçimde girin.")
except ZeroDivisionError:
    print("Bir sayı sıfıra bölünemez.")
finally:
    print("Hesaplama işlemi sona erdi.")
````

Bu örnekte önceki derslerde öğrendiğimiz karar yapıları ile hata yönetimini bir araya getirdik.

Program:

- kullanıcıdan iki sayı alıyor,
- işlem seçimini kontrol ediyor,
- seçilen işleme göre hesaplama yapıyor,
- yanlış sayı girişini yakalıyor,
- sıfıra bölme hatasını yakalıyor,
- işlemin sonunda `finally` bölümünü çalıştırıyor.

Gerçek uygulamalarda farklı Python konularını bu şekilde bir araya getireceğiz.


## 18. Sık Yapılan Hatalar

### `try` içine gereğinden fazla kod yazmak

Hata oluşabilecek kodu mümkün olduğunca ilgili `try` bloğunda tutmak okunabilirliği artırır.

### Her şeyi `except:` ile yakalamak

Beklenen hata türlerini mümkün olduğunca açıkça belirtmek daha sağlıklıdır.

### Hata mesajını anlamadan değiştirmek

Python'ın verdiği hata türünü ve mesajını okumak, problemi çözmenin önemli bir parçasıdır.

### `finally` ile `else` arasındaki farkı karıştırmak

- `else`: hata olmadığında çalışır.
- `finally`: hata olsa da olmasa da çalışır.


## 19. Ders Özeti

Bu derste programlarda oluşabilecek hataları kontrol etmeyi öğrendik.

Temel yapı:

```python
try:
    ...
except:
    ...
```

Daha kontrollü kullanım:

```python
try:
    ...
except ValueError:
    ...
except ZeroDivisionError:
    ...
```

İşlemin her durumda sonunda çalışacak bölüm:

```python
finally:
    ...
```

Hata oluşmadığında çalışacak bölüm:

```python
else:
    ...
```

Özellikle şu hata türlerini tanıdık:

- `ValueError`
- `ZeroDivisionError`
- `TypeError`
- `IndexError`
- `KeyError`
- `FileNotFoundError`
- `NameError`

Hata yönetiminin amacı hataları gizlemek değil, programın beklenmeyen durumları kontrollü şekilde ele almasını sağlamaktır.
