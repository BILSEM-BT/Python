## 1. Döngü Mantığını Anlamak

Döngü, belirli bir kod bölümünün birden fazla kez çalıştırılmasıdır.

Örneğin aynı mesajı beş kez yazdırmak istersek bunu şöyle yazabiliriz:


````python
print("Merhaba")
print("Merhaba")
print("Merhaba")
print("Merhaba")
print("Merhaba")
````

Kod çalışır ancak tekrar sayısı arttıkça bu yaklaşım kullanışsız hale gelir.

Döngü kullanarak aynı işlemi daha kısa ve kontrollü şekilde yapabiliriz.


## 2. `for` Döngüsüne Giriş

`for` döngüsü bir koleksiyonun veya belirli bir aralığın elemanları üzerinde sırayla işlem yapmak için çok kullanılır.

En basit örnek:


````python
for i in range(5):
    print("Merhaba")
````

Burada `range(5)` bize 0'dan başlayarak 5'ten küçük değerlere kadar bir aralık oluşturur.

Döngü sırasıyla:

```text
0
1
2
3
4
```

değerlerini alır.

`i` değişkeni her turda o anki değeri temsil eder.


## 3. `range()` Fonksiyonu

`range()` döngülerde çok sık kullanılan bir yapıdır.

### `range(5)`

0'dan 4'e kadar üretir.

### `range(2, 7)`

2'den 6'ya kadar üretir.

### `range(2, 10, 2)`

2'den başlayıp 2'şer artırarak ilerler.

Örnek:


````python
for i in range(5):
    print(i)
````

````python
for i in range(2, 7):
    print(i)
````

````python
for i in range(2, 10, 2):
    print(i)
````

Burada önemli nokta, `range()` yapısında bitiş değerinin genellikle dahil edilmemesidir.

```python
range(1, 6)
```

değerleri:

```text
1, 2, 3, 4, 5
```

şeklindedir.


## 4. `for` ile Sayıları İşlemek

Döngüleri yalnızca ekrana yazdırmak için değil, hesaplama yapmak için de kullanabiliriz.

Örneğin 1'den 10'a kadar olan sayıların toplamını bulalım.


````python
toplam = 0

for i in range(1, 11):
    toplam = toplam + i

print("Toplam:", toplam)
````

Burada `toplam` değişkeni biriktirici olarak kullanılır.

Her turda:

```python
toplam = toplam + i
```

işlemi yapılır.

Bu yaklaşımı veri analizi yaparken çok sık kullanacağız.


## 5. `for` ve Karar Yapıları

Döngüler ile `if` yapıları birlikte kullanıldığında daha güçlü programlar oluşturabiliriz.

Örneğin 1 ile 20 arasındaki çift sayıları bulalım.


````python
for sayi in range(1, 21):
    if sayi % 2 == 0:
        print(sayi)
````

Burada:

- `for` sayıları sırayla dolaşıyor.
- `if` sayının çift olup olmadığını kontrol ediyor.
- `% 2` işleminin sonucu 0 ise sayı çift kabul ediliyor.

Bu, önceki derste öğrendiğimiz karar yapısının döngü içerisinde kullanımına güzel bir örnektir.


## 6. Tek Sayıları Bulmak

````python
for sayi in range(1, 21):
    if sayi % 2 != 0:
        print(sayi)
````

## 7. Bir Listenin Elemanlarını Dolaşmak

Örneğin:


````python
ogrenciler = ["Ali", "Ayşe", "Mehmet", "Zeynep"]

for ogrenci in ogrenciler:
    print(ogrenci)
````

Burada `ogrenci` değişkeni her turda listenin bir elemanını temsil eder.

İlk turda:

```text
Ali

```

Bu yapı daha sonra veri analizi çalışmalarımızın temel araçlarından biri olacaktır.


## 8. Liste Elemanları Üzerinde İşlem Yapmak

````python
notlar = [65, 80, 92, 45, 73]

for notu in notlar:
    if notu >= 50:
        print(notu, "Başarılı")
    else:
        print(notu, "Başarısız")
````

Burada aynı anda üç Python konusunu kullanıyoruz:

- liste
- `for`
- `if`

Programlama konularını ayrı ayrı öğrenmek önemlidir; ancak gerçek programlar bu yapıların birlikte kullanılmasından oluşur.


## 9. `while` Döngüsü

`while` döngüsü bir koşul doğru olduğu sürece çalışır.

Genel yapı:

```python
while koşul:
    yapılacak işlem
```

Örneğin:


````python
sayac = 1

while sayac <= 5:
    print(sayac)
    sayac = sayac + 1
````

Burada döngü:

1. `sayac <= 5` koşulunu kontrol eder.
2. Koşul doğruysa kodu çalıştırır.
3. `sayac` değerini artırır.
4. Koşulu tekrar kontrol eder.
5. Koşul yanlış olduğunda döngü sona erer.

`while` döngülerinde koşulu değiştiren işlemin unutulmaması çok önemlidir.


## 10. Sonsuz Döngü

Aşağıdaki kodu inceleyelim:

```python
sayac = 1

while sayac <= 5:
    print(sayac)
```

Burada `sayac` hiç değişmediği için koşul sürekli doğru kalır.

Bu nedenle döngü sonsuza kadar devam eder.

Doğru kullanım:


````python
sayac = 1

while sayac <= 5:
    print(sayac)
    sayac += 1
````

Sonsuz döngüler bazı uygulamalarda bilinçli olarak kullanılabilir. Ancak kontrolsüz sonsuz döngüler programların kilitlenmesine neden olabilir.


## 11. `while` ile Kullanıcıdan Veri Almak

`while` özellikle kullanıcıdan doğru veri gelene kadar tekrar istemek için kullanışlıdır.

Örneğin:


````python
sayi = -1

while sayi < 0:
    sayi = int(input("Pozitif bir sayı girin: "))

print("Girdiğiniz sayı:", sayi)
````

Kullanıcı negatif bir sayı girerse program tekrar sayı ister.

Pozitif bir sayı girildiğinde koşul yanlış olur ve döngü sona erer.

Bu yapı, önceki hata yönetimi dersimizde gördüğümüz `try-except` ile de birleştirilebilir.


## 12. `while` ve `try-except` Birlikte

Kullanıcıdan doğru bir sayı alana kadar hem geçersiz metinleri hem de yanlış değerleri kontrol edebiliriz.


````python
while True:
    try:
        sayi = int(input("Pozitif bir sayı girin: "))

        if sayi > 0:
            print("Geçerli sayı:", sayi)
            break
        else:
            print("Sayı 0'dan büyük olmalıdır.")

    except ValueError:
        print("Lütfen geçerli bir tam sayı girin.")
````

Bu örnekte daha önce öğrendiğimiz konuları birleştirdik:

- `while`
- `try`
- `except`
- `if`
- `break`

İleride daha karmaşık programlar oluştururken konuları bu şekilde bir araya getireceğiz.


## 13. `break`

`break`, döngüyü hemen sonlandırmak için kullanılır.

Örneğin:


````python
for sayi in range(1, 11):
    if sayi == 6:
        break
    print(sayi)
````

Döngü 6 değerine geldiğinde `break` çalışır ve döngü sona erer.

Çıktı:

```text
1
2
3
4
5
```


## 14. `continue`

`continue`, döngünün o turundaki işlemlerin geri kalanını atlayarak bir sonraki tura geçer.

Örneğin çift sayıları atlayalım:


````python
for sayi in range(1, 11):
    if sayi % 2 == 0:
        continue

    print(sayi)
````

Çift sayılarda `continue` çalışır ve `print()` satırına gelinmeden bir sonraki tur başlar.

Bu nedenle yalnızca tek sayılar yazdırılır.


## 15. `break` ve `continue` Arasındaki Fark

`break`:

> Döngünün tamamını bitirir.

`continue`:

> Sadece mevcut turu atlar ve döngünün sonraki turuna geçer.

Bu farkı anlamak önemlidir.


## 16. İç İçe Döngüler

Bir döngünün içinde başka bir döngü bulunabilir.

Örneğin:


````python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
````

İç döngü, dış döngünün her turunda yeniden çalışır.

Bu yapı özellikle:

- tablolar
- matrisler
- koordinatlar
- oyun tahtaları
- veri karşılaştırmaları

gibi problemlerde kullanılabilir.


## 17. Çarpım Tablosu

İç içe döngüler için klasik bir örnek çarpım tablosudur.


````python
for i in range(1, 6):
    for j in range(1, 6):
        print(i, "x", j, "=", i * j)
    print()
````

Burada dış döngü satırları, iç döngü ise her satırdaki çarpma işlemlerini oluşturur.


## 18. Liste İçinde Koşullu Sayım

Bir listedeki kaç öğrencinin başarılı olduğunu bulalım.


````python
notlar = [45, 70, 82, 35, 90, 55, 48]

basarili = 0

for notu in notlar:
    if notu >= 50:
        basarili += 1

print("Başarılı öğrenci sayısı:", basarili)
````

Burada `basarili` değişkeni sayaç olarak kullanılıyor.

Bu sayaç mantığı daha sonra veri analizi yaparken çok işimize yarayacak.


## 19. Liste Elemanlarının Toplamı ve Ortalaması

Bir veri grubunun toplamını ve ortalamasını döngü kullanarak hesaplayabiliriz.


````python
notlar = [70, 85, 60, 90, 75]

toplam = 0

for notu in notlar:
    toplam += notu

ortalama = toplam / len(notlar)

print("Toplam:", toplam)
print("Ortalama:", ortalama)
````

Burada `len()` listenin eleman sayısını verir.

Bu örnekte döngü ile toplamı bulduk, `len()` ile eleman sayısını bulduk ve ardından ortalamayı hesapladık.

İleride Python'ın hazır fonksiyonlarıyla bu işlemleri daha kısa yollarla da yapacağız.


## 20. Metinleri Döngüyle İncelemek

Python'da stringler de karakterlerden oluşan bir yapı gibi dolaşılabilir.


````python
metin = "Python"

for karakter in metin:
    print(karakter)
````

Bu özellik metin analizi çalışmalarında önemlidir.

## 22. Döngülerde `else`

Python'da döngülerin de `else` bölümü olabilir.

Örneğin:


````python
for sayi in range(5):
    print(sayi)
else:
    print("Döngü tamamlandı.")
````

Döngü normal şekilde tamamlandığında `else` çalışır.

Ancak `break` ile döngü kesilirse döngünün `else` bölümü çalışmaz.


````python
for sayi in range(10):
    if sayi == 5:
        break
    print(sayi)
else:
    print("Döngü tamamlandı.")
````

Bu kullanım başlangıç seviyesinde çok sık gerekli değildir; ancak Python'un döngü mantığını tam anlamak açısından bilinmesi faydalıdır.


## 23. Asal Sayı Kontrolü

Döngü ve karar yapısını birlikte kullanarak bir sayının asal olup olmadığını kontrol edebiliriz.


````python
sayi = 29
asal = True

if sayi < 2:
    asal = False
else:
    for i in range(2, sayi):
        if sayi % i == 0:
            asal = False
            break

if asal:
    print("Asal sayı")
else:
    print("Asal değil")
````

Bu örnekte program sayının 2 ile kendisinden küçük sayılar arasında bir böleni olup olmadığını kontrol eder.

Bir bölen bulunduğunda `break` ile gereksiz kontroller durdurulur.

Bu, algoritma düşüncesinin gelişmesi açısından güzel bir örnektir.

## 25. Gerçek Hayat Örneği: Öğrenci Not Analizi

Bir grup öğrencinin notlarını inceleyelim.

Program:

- toplamı bulacak,
- ortalamayı hesaplayacak,
- başarılı öğrenci sayısını bulacak,
- en yüksek notu belirleyecek.


````python
notlar = [65, 80, 92, 45, 73, 88, 54]

toplam = 0
basarili = 0
en_yuksek = notlar[0]

for notu in notlar:
    toplam += notu

    if notu >= 50:
        basarili += 1

    if notu > en_yuksek:
        en_yuksek = notu

ortalama = toplam / len(notlar)

print("Toplam:", toplam)
print("Ortalama:", ortalama)
print("Başarılı öğrenci sayısı:", basarili)
print("En yüksek not:", en_yuksek)
````

Bu örnek, ileride yapacağımız veri analizi çalışmalarının küçük bir başlangıcıdır.

Gerçek veri setlerinde binlerce hatta milyonlarca kayıt olabilir. Döngüler, verileri tek tek inceleme mantığının temelini oluşturur.

Daha ileride `pandas` gibi kütüphaneler kullanarak bu işlemleri çok daha güçlü yöntemlerle gerçekleştireceğiz.


## 26. Döngüler ve Veri Analizi

Döngüleri veri analizi açısından düşünelim.

Elimizde:

```python
sicakliklar = [18, 21, 25, 30, 27, 16, 22]
```

gibi bir veri grubu olduğunu düşünelim.

Ortalama sıcaklığı hesaplayabiliriz:


````python
sicakliklar = [18, 21, 25, 30, 27, 16, 22]

toplam = 0

for sicaklik in sicakliklar:
    toplam += sicaklik

ortalama = toplam / len(sicakliklar)

print("Ortalama sıcaklık:", ortalama)
````

Aynı veri üzerinde belirli bir değerin üzerindeki ölçümleri sayabiliriz.


````python
sicakliklar = [18, 21, 25, 30, 27, 16, 22]

sicak_gun = 0

for sicaklik in sicakliklar:
    if sicaklik >= 25:
        sicak_gun += 1

print("25 derece ve üzerindeki gün sayısı:", sicak_gun)
````

Bu düşünce yapısı ileride veri analizi derslerinde daha büyük veri kümelerine uygulanacaktır.
