# BÖLÜM 1 — TUPLE (DEMETLER)

## 1. Tuple Nedir?

Tuple, birden fazla değeri bir arada tutan ve **sıralı** çalışan bir Python veri yapısıdır.

Listeye benzer:

```python
liste = [10, 20, 30]
```

Tuple ise genellikle parantez ile oluşturulur:

```python
demet = (10, 20, 30)
```

En önemli farklardan biri şudur:

**Tuple oluşturulduktan sonra elemanları değiştirilemez.**

Bu özelliğe **değiştirilemezlik (immutable)** denir.


````python
sayilar = (10, 20, 30, 40)

print(sayilar)
print(type(sayilar))
````

## 2. Tuple Oluşturma

Parantez kullanmak en yaygın yöntemdir.


````python
meyveler = ("elma", "armut", "muz")

print(meyveler)
````

Tek elemanlı tuple oluştururken virgül önemlidir.

Aşağıdaki yapı tuple'dır:


````python
tek = (10,)

print(type(tek))
````

Şu yapı ise tuple değildir:


````python
tek = (10)

print(type(tek))
````

Burada `(10)` sadece parantez içine alınmış bir sayıdır. Tuple olduğunu belirleyen önemli unsur virgüldür.

```python
(10,)
```


## 3. Tuple İndeksleme

Tuple'lar sıralı olduğu için indeks kullanabiliriz.


````python
renkler = ("kırmızı", "yeşil", "mavi", "sarı")

print(renkler[0])
print(renkler[2])
print(renkler[-1])
````

Listelerde olduğu gibi indeksler `0`dan başlar.

Negatif indeksler de kullanılabilir:

- `-1` → son eleman
- `-2` → sondan ikinci eleman


## 4. Tuple Dilimleme

Tuple'larda dilimleme de yapılabilir.


````python
sayilar = (10, 20, 30, 40, 50, 60)

print(sayilar[1:4])
print(sayilar[:3])
print(sayilar[3:])
print(sayilar[::2])
````

Bu yapı listelerdeki dilimleme mantığıyla aynıdır.

## 5. Tuple Elemanı Değiştirilebilir mi?

Hayır.

Aşağıdaki kod hata verir:


````python
renkler = ("kırmızı", "yeşil", "mavi")

# renkler[0] = "sarı"
````

Tuple'ın amacı bazı verileri oluşturulduktan sonra değiştirilmeden saklamaktır.

Örneğin:

- koordinatlar,
- sabit seçenekler,
- tarih bilgileri,
- bir kayda ait sabit bilgiler

gibi veriler tuple ile tutulabilir.


## 6. Tuple İçindeki Verinin Türleri

Tuple içinde farklı veri türleri bulunabilir.


````python
ogrenci = ("Ali", 14, 8, 85.5, True)

print(ogrenci)
````

Tuple'ın elemanları kendi içinde farklı türlerde olabilir.

## 7. `len()` ile Tuple Uzunluğu

````python
dersler = ("Python", "Matematik", "Fizik", "Biyoloji")

print(len(dersler))
````

## 8. `in` ile Tuple İçinde Arama

````python
dersler = ("Python", "Matematik", "Fizik", "Biyoloji")

print("Python" in dersler)
print("Kimya" in dersler)
````

Bir değerin bulunmadığını kontrol etmek için `not in` kullanabiliriz.

````python
print("Kimya" not in dersler)
````

## 9. Tuple Metodu: `count()`

Bir değerin kaç kez bulunduğunu sayar.


````python
sayilar = (10, 20, 10, 30, 10, 40)

print(sayilar.count(10))
print(sayilar.count(50))
````

## 10. Tuple Metodu: `index()`

Bir değerin ilk bulunduğu indeksini verir.


````python
meyveler = ("elma", "armut", "muz", "elma")

print(meyveler.index("muz"))
print(meyveler.index("elma"))
````

Aranan değer tuple içinde yoksa `ValueError` oluşur. Bu nedenle kullanıcıdan gelen verilerle çalışırken kontrol yapmak gerekebilir.

## 11. Tuple ile Döngü

Tuple'lar üzerinde `for` döngüsü kullanabiliriz.


````python
renkler = ("kırmızı", "yeşil", "mavi")

for renk in renkler:
    print(renk)
````

## 12. Tuple Unpacking

Tuple içindeki değerleri ayrı değişkenlere aktarabiliriz.


````python
ogrenci = ("Ali", 14, 8)

isim, yas, sinif = ogrenci

print("İsim:", isim)
print("Yaş:", yas)
print("Sınıf:", sinif)
````

Bu işleme **unpacking** denir.

Değişken sayısının tuple'daki eleman sayısıyla uyumlu olması gerekir.

## 13. Değişken Sayıda Unpacking

`*` kullanarak kalan değerleri bir liste halinde alabiliriz.


````python
sayilar = (10, 20, 30, 40, 50)

ilk, *orta, son = sayilar

print("İlk:", ilk)
print("Orta:", orta)
print("Son:", son)
````

Bu teknik ileride fonksiyonlarda ve veri işlemede oldukça kullanışlı olacaktır.

## 14. Tuple ve Liste Arasındaki Fark

| Özellik | Liste | Tuple |
|---|---|---|
| Sıralı | Evet | Evet |
| İndeks | Var | Var |
| Dilimleme | Var | Var |
| Eleman değiştirilebilir | Evet | Hayır |
| `append()` | Var | Yok |
| `remove()` | Var | Yok |
| `count()` | Var | Var |
| `index()` | Var | Var |

Kısaca:

**Liste → değişebilen veri grubu**

**Tuple → değişmemesi gereken veri grubu**


## 15. Listeyi Tuple'a Dönüştürmek

````python
liste = [10, 20, 30, 40]

demet = tuple(liste)

print(demet)
print(type(demet))
````

## 16. Tuple'ı Listeye Dönüştürmek

````python
demet = (10, 20, 30, 40)

liste = list(demet)

print(liste)
print(type(liste))
````

Tuple doğrudan değiştirilemediği için bazen listeye dönüştürüp değişiklik yaptıktan sonra tekrar tuple haline getirebiliriz.

````python
demet = (10, 20, 30)

liste = list(demet)
liste.append(40)

demet = tuple(liste)

print(demet)
````

# BÖLÜM 2 — SET (KÜMELER)

## 17. Set Nedir?

Set, birden fazla veriyi bir arada tutan ancak **tekrarlanan değerleri saklamayan** bir Python veri yapısıdır.

Set oluşturmak için süslü parantez kullanabiliriz:

```python
sayilar = {10, 20, 30}
```

Set'in en önemli özelliklerinden biri:

**Aynı değer birden fazla kez bulunamaz.**


````python
sayilar = {10, 20, 20, 30, 30, 30}

print(sayilar)
````

Tekrarlı değerlerin otomatik olarak tek değere indirildiğini görüyoruz.

## 18. Set ile Listeden Tekrarları Temizlemek

Bu kullanım veri analizi açısından çok önemlidir.


````python
isimler = ["Ali", "Ayşe", "Ali", "Mehmet", "Ayşe", "Zeynep"]

benzersiz_isimler = set(isimler)

print(benzersiz_isimler)
````

Örneğin bir veri dosyasında aynı şehir birçok kez geçiyorsa, benzersiz şehirleri bulmak için set kullanabiliriz.

````python
sehirler = ["İstanbul", "Ankara", "İzmir", "İstanbul", "Bursa", "Ankara"]

benzersiz_sehirler = set(sehirler)

print(benzersiz_sehirler)
````

## 19. Set Elemanlarına İndeks ile Ulaşabilir miyiz?

Hayır.

Set'ler indeks mantığıyla kullanılmaz.

Örneğin:

```python
sayilar[0]
```

şeklinde erişim yapılamaz.

Çünkü set'in temel amacı sıralı bir veri dizisi oluşturmak değil, benzersiz değerleri tutmak ve küme işlemlerini hızlı biçimde yapabilmektir.


## 20. Boş Set Oluşturmak

Dikkat:

```python
bos = {}
```

ifadesi boş sözlük oluşturur.

Boş set için:

```python
bos = set()
```

kullanılır.


````python
bos_set = set()

print(type(bos_set))
````

## 21. `add()`

Set'e tek bir eleman eklemek için `add()` kullanılır.


````python
sayilar = {10, 20, 30}

sayilar.add(40)

print(sayilar)
````

Aynı elemanı tekrar eklemeye çalışırsak set ikinci kez saklamaz.

````python
sayilar = {10, 20, 30}

sayilar.add(20)
sayilar.add(20)

print(sayilar)
````

## 22. `update()`

Birden fazla eleman eklemek için `update()` kullanabiliriz.


````python
sayilar = {10, 20, 30}

sayilar.update([40, 50, 60])

print(sayilar)
````

Liste, tuple veya başka bir iterable yapıdan eleman ekleyebiliriz.

````python
renkler = {"kırmızı", "mavi"}

renkler.update(("yeşil", "sarı"))

print(renkler)
````

## 23. `remove()` ve `discard()`

`remove()` belirtilen elemanı siler.

Eleman yoksa hata oluşur.


````python
sayilar = {10, 20, 30, 40}

sayilar.remove(30)

print(sayilar)
````

`discard()` ise eleman yoksa hata vermez.

````python
sayilar = {10, 20, 30}

sayilar.discard(50)

print(sayilar)
````

Bu farkı bilmek özellikle kullanıcıdan gelen verilerle çalışırken önemlidir.

## 24. `pop()`

Set'te `pop()` bir elemanı çıkarır.

Set'lerde sıralı erişim olmadığı için hangi elemanın çıkarılacağını indeksle belirleyemeyiz.


````python
sayilar = {10, 20, 30, 40}

cikarilan = sayilar.pop()

print("Çıkarılan:", cikarilan)
print("Set:", sayilar)
````

## 25. `clear()`

Set'in bütün elemanlarını siler.


````python
sayilar = {10, 20, 30}

sayilar.clear()

print(sayilar)
````

# BÖLÜM 3 — KÜME İŞLEMLERİ

Set'lerin en güçlü tarafı matematikteki küme işlemlerine benzer işlemler yapabilmemizdir.

Örneğin iki öğrencinin aldığı dersleri karşılaştırabiliriz.


## 26. Birleşim — `union()`

İki kümedeki tüm farklı elemanları bir araya getirir.


````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

sonuc = a.union(b)

print(sonuc)
````

Aynı işlem `|` operatörüyle de yapılabilir.

````python
a = {1, 2, 3}
b = {3, 4, 5}

print(a | b)
````

## 27. Kesişim — `intersection()`

İki kümede ortak bulunan elemanları verir.


````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a.intersection(b))
````

`&` operatörü de kesişim için kullanılabilir.

````python
a = {1, 2, 3}
b = {2, 3, 4}

print(a & b)
````

## 28. Fark — `difference()`

Birinci kümede olup ikinci kümede olmayan elemanları verir.


````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a.difference(b))
print(b.difference(a))
````

Operatör olarak `-` kullanılabilir.

````python
a = {1, 2, 3, 4}
b = {3, 4, 5}

print(a - b)
````

## 29. Simetrik Fark — `symmetric_difference()`

İki kümede ortak olmayan elemanları verir.

Yani yalnızca bir kümede bulunan elemanlar alınır.


````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a.symmetric_difference(b))
````

`^` operatörü de kullanılabilir.

````python
a = {1, 2, 3}
b = {2, 3, 4}

print(a ^ b)
````

## 30. Alt Küme — `issubset()`

Bir kümedeki bütün elemanların başka bir kümede bulunup bulunmadığını kontrol eder.


````python
a = {1, 2}
b = {1, 2, 3, 4}

print(a.issubset(b))
````

`a`, `b` kümesinin alt kümesidir.

## 31. Üst Küme — `issuperset()`

````python
a = {1, 2, 3, 4}
b = {1, 2}

print(a.issuperset(b))
````

## 32. Kümelerin Ayrık Olup Olmadığını Kontrol Etmek

`isdisjoint()` iki kümenin ortak elemanı olup olmadığını kontrol eder.


````python
a = {1, 2, 3}
b = {4, 5, 6}

print(a.isdisjoint(b))
````

Ortak eleman varsa sonuç `False` olur.

````python
a = {1, 2, 3}
b = {3, 4, 5}

print(a.isdisjoint(b))
````

# BÖLÜM 4 — VERİ ANALİZİ ÖRNEKLERİ

## 33. Bir Sınıftaki Benzersiz Dersler

Öğrencilerin seçtiği dersleri düşünelim.


````python
dersler = [
    "Python",
    "Python",
    "Veri Analizi",
    "Grafik",
    "Python",
    "Yapay Zeka",
    "Veri Analizi"
]

benzersiz_dersler = set(dersler)

print("Benzersiz dersler:", benzersiz_dersler)
print("Ders sayısı:", len(benzersiz_dersler))
````

Burada set kullanarak tekrar eden dersleri temizledik ve kaç farklı ders olduğunu bulduk.

## 34. İki Öğrencinin Ortak Dersleri

````python
ali_dersleri = {"Python", "Matematik", "Fizik", "Robotik"}
ayse_dersleri = {"Python", "Biyoloji", "Fizik", "Grafik"}

ortak = ali_dersleri.intersection(ayse_dersleri)

print("Ortak dersler:", ortak)
````

## 35. Sadece Bir Öğrencinin Aldığı Dersler

````python
ali_dersleri = {"Python", "Matematik", "Fizik", "Robotik"}
ayse_dersleri = {"Python", "Biyoloji", "Fizik", "Grafik"}

print("Ali'nin alıp Ayşe'nin almadığı:", ali_dersleri - ayse_dersleri)
print("Ayşe'nin alıp Ali'nin almadığı:", ayse_dersleri - ali_dersleri)
````

## 36. İki Sınıftaki Ortak Öğrenciler

Öğrenci isimlerini set olarak düşünelim.


````python
sinif_a = {"Ali", "Ayşe", "Mehmet", "Zeynep"}
sinif_b = {"Mehmet", "Zeynep", "Can", "Elif"}

print("Her iki sınıfta bulunanlar:", sinif_a & sinif_b)
print("En az bir sınıfta bulunanlar:", sinif_a | sinif_b)
````

## 37. Benzersiz Ürün Kategorileri

````python
urunler = [
    "Laptop",
    "Telefon",
    "Laptop",
    "Tablet",
    "Telefon",
    "Kulaklık",
    "Tablet"
]

kategoriler = set(urunler)

print("Ürün türleri:", kategoriler)
print("Farklı ürün sayısı:", len(kategoriler))
````

## 38. Tuple ile Sabit Kayıt

Bir öğrencinin temel bilgilerini tuple içinde tutabiliriz.


````python
ogrenci = ("Ali", 14, "Niyazi Sayın BİLSEM", "Python")

isim, yas, okul, ders = ogrenci

print("İsim:", isim)
print("Yaş:", yas)
print("Okul:", okul)
print("Ders:", ders)
````

Burada öğrenci kaydının yapısını değiştirmemek istediğimiz için tuple uygun bir yapı olabilir.
