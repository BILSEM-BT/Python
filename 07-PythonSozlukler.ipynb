# 1. Dictionary Nedir?

Dictionary, verileri **anahtar-değer (key-value)** çiftleri şeklinde saklayan bir veri yapısıdır.

Temel yapı:

```python
sozluk = {
    "anahtar": "değer"
}
```

Örnek:


````python
ogrenci = {
    "isim": "Ali",
    "yas": 14,
    "sinif": 8,
    "puan": 85
}

print(ogrenci)
print(type(ogrenci))
````

Burada:

- `"isim"` → anahtar
- `"Ali"` → değer
- `"yas"` → anahtar
- `14` → değer

şeklindedir.

Bir sözlükte her bilgiye anlamlı bir anahtar vererek ulaşabiliriz.

# 2. Dictionary Elemanlarına Ulaşmak

Listelerde indeks kullanıyorduk:

```python
liste[0]
```

Dictionary'de ise anahtarı kullanırız.


````python
ogrenci = {
    "isim": "Ali",
    "yas": 14,
    "sinif": 8,
    "puan": 85
}

print(ogrenci["isim"])
print(ogrenci["puan"])
````

Bu yapı gerçek hayattaki verileri anlamlandırmak için çok kullanışlıdır.

# 3. `get()` Metodu

Anahtar mevcut değilse doğrudan `sozluk["anahtar"]` kullanmak hata oluşturabilir.

`get()` daha güvenli bir yöntemdir.

````python
ogrenci = {
    "isim": "Ali",
    "yas": 14
}

print(ogrenci.get("isim"))
print(ogrenci.get("puan"))
````

İkinci sorguda `"puan"` olmadığı için `None` döndü.

İstersek varsayılan bir değer de belirleyebiliriz.

````python
ogrenci = {
    "isim": "Ali",
    "yas": 14
}

print(ogrenci.get("puan", 0))
print(ogrenci.get("sinif", "Bilgi yok"))
````

# 4. Dictionary'ye Yeni Veri Eklemek

Yeni bir anahtar kullanarak veri ekleyebiliriz.

````python
ogrenci = {
    "isim": "Ali",
    "yas": 14
}

ogrenci["sinif"] = 8
ogrenci["puan"] = 85

print(ogrenci)
````

# 5. Var Olan Veriyi Değiştirmek

Aynı anahtara yeni değer verirsek mevcut değer değişir.

````python
ogrenci = {
    "isim": "Ali",
    "yas": 14,
    "puan": 70
}

ogrenci["puan"] = 85

print(ogrenci)
````

Bu nedenle dictionary'lerde anahtarın var olup olmaması önemlidir:

- Anahtar yoksa → yeni veri eklenir.
- Anahtar varsa → mevcut değer değiştirilir.


# 6. Dictionary'den Veri Silmek

`del` ile belirli bir anahtar silinebilir.

````python
ogrenci = {
    "isim": "Ali",
    "yas": 14,
    "sinif": 8
}

del ogrenci["yas"]

print(ogrenci)
````

`pop()` ile de bir anahtarı silip silinen değeri alabiliriz.

````python
ogrenci = {
    "isim": "Ali",
    "yas": 14,
    "sinif": 8
}

yas = ogrenci.pop("yas")

print("Silinen yaş:", yas)
print(ogrenci)
````

`pop()` ile olmayan bir anahtar üzerinde çalışırken varsayılan değer verebiliriz.

````python
ogrenci = {
    "isim": "Ali"
}

sonuc = ogrenci.pop("puan", None)

print(sonuc)
````

# 7. `clear()`

Sözlüğün bütün elemanlarını siler.

````python
ogrenci = {
    "isim": "Ali",
    "yas": 14,
    "sinif": 8
}

ogrenci.clear()

print(ogrenci)
````

# 8. Dictionary Uzunluğu

`len()` ile kaç anahtar-değer çifti olduğunu bulabiliriz.

````python
ogrenci = {
    "isim": "Ali",
    "yas": 14,
    "sinif": 8,
    "puan": 85
}

print(len(ogrenci))
````

# 9. `keys()`

Sözlükteki bütün anahtarları verir.

````python
ogrenci = {
    "isim": "Ali",
    "yas": 14,
    "sinif": 8,
    "puan": 85
}

print(ogrenci.keys())
````

Anahtarları döngüyle de kullanabiliriz.

````python
ogrenci = {
    "isim": "Ali",
    "yas": 14,
    "sinif": 8,
    "puan": 85
}

for anahtar in ogrenci.keys():
    print(anahtar)
````

# 10. `values()`

Sözlükteki bütün değerleri verir.

````python
ogrenci = {
    "isim": "Ali",
    "yas": 14,
    "sinif": 8,
    "puan": 85
}

print(ogrenci.values())
````

Örneğin sayısal değerler üzerinde işlem yapabiliriz.

````python
notlar = {
    "matematik": 80,
    "fizik": 75,
    "python": 95,
    "biyoloji": 70
}

print("Not toplamı:", sum(notlar.values()))
print("Not sayısı:", len(notlar))
````

# 11. `items()`

Anahtar ve değerleri birlikte verir.

````python
ogrenci = {
    "isim": "Ali",
    "yas": 14,
    "sinif": 8
}

print(ogrenci.items())
````

`items()` özellikle döngülerde çok kullanılır.

````python
ogrenci = {
    "isim": "Ali",
    "yas": 14,
    "sinif": 8
}

for anahtar, deger in ogrenci.items():
    print(anahtar, ":", deger)
````

# 12. Dictionary'de `in`

Anahtarın sözlükte bulunup bulunmadığını kontrol edebiliriz.

````python
ogrenci = {
    "isim": "Ali",
    "yas": 14,
    "sinif": 8
}

print("isim" in ogrenci)
print("puan" in ogrenci)
````

`in` doğrudan değerleri değil, anahtarları kontrol eder.

# 13. İç İçe Dictionary

Bir dictionary'nin değeri başka bir dictionary olabilir.

Bu yapı daha karmaşık verileri modellemek için çok önemlidir.

````python
ogrenci = {
    "bilgiler": {
        "isim": "Ali",
        "yas": 14,
        "sinif": 8
    },
    "notlar": {
        "python": 90,
        "matematik": 80,
        "fizik": 75
    }
}

print(ogrenci["bilgiler"]["isim"])
print(ogrenci["notlar"]["python"])
````

# 14. Dictionary İçinde Liste

Bir anahtarın değeri liste olabilir.

````python
ogrenci = {
    "isim": "Ali",
    "dersler": ["Python", "Matematik", "Fizik"]
}

print(ogrenci["dersler"])
print(ogrenci["dersler"][0])
````

Bu yapı gerçek uygulamalarda çok sık karşımıza çıkar.

# 15. Liste İçinde Dictionary

Tersini de yapabiliriz.

Birden fazla öğrenciyi liste içinde dictionary olarak saklayabiliriz.

````python
ogrenciler = [
    {"isim": "Ali", "puan": 85},
    {"isim": "Ayşe", "puan": 92},
    {"isim": "Mehmet", "puan": 76}
]

for ogrenci in ogrenciler:
    print(ogrenci["isim"], ogrenci["puan"])
````

Bu veri yapısı ileride JSON, web API'leri ve veri analizi çalışmalarında karşımıza çok çıkacaktır.

# 16. Öğrencilerin Ortalama Puanını Hesaplamak

````python
ogrenciler = [
    {"isim": "Ali", "puan": 85},
    {"isim": "Ayşe", "puan": 92},
    {"isim": "Mehmet", "puan": 76},
    {"isim": "Zeynep", "puan": 88}
]

toplam = 0

for ogrenci in ogrenciler:
    toplam += ogrenci["puan"]

ortalama = toplam / len(ogrenciler)

print("Sınıf ortalaması:", ortalama)
````

# 17. Başarılı Öğrencileri Bulmak

````python
ogrenciler = [
    {"isim": "Ali", "puan": 45},
    {"isim": "Ayşe", "puan": 92},
    {"isim": "Mehmet", "puan": 38},
    {"isim": "Zeynep", "puan": 88}
]

for ogrenci in ogrenciler:
    if ogrenci["puan"] >= 50:
        print(ogrenci["isim"], "başarılı")
````

# 18. Dictionary Kopyalama

Dictionary'lerde `copy()` metodu kullanılabilir.

````python
ogrenci = {
    "isim": "Ali",
    "puan": 85
}

yeni_ogrenci = ogrenci.copy()
yeni_ogrenci["puan"] = 95

print("Orijinal:", ogrenci)
print("Yeni:", yeni_ogrenci)
````

# 19. `update()`

Bir dictionary'ye birden fazla veri eklemek veya mevcut değerleri güncellemek için `update()` kullanılabilir.

````python
ogrenci = {
    "isim": "Ali",
    "puan": 85
}

ogrenci.update({
    "yas": 14,
    "sinif": 8,
    "puan": 90
})

print(ogrenci)
````

# 20. Dictionary Birleştirme

İki dictionary'yi birleştirebiliriz.

````python
bilgiler = {
    "isim": "Ali",
    "yas": 14
}

notlar = {
    "python": 90,
    "matematik": 80
}

ogrenci = bilgiler | notlar

print(ogrenci)
````

# 21. Dictionary İçinde Sayaç Oluşturmak

Dictionary'ler sayma işlemlerinde çok kullanışlıdır.

Örneğin bir listedeki kelimelerin kaç kez geçtiğini hesaplayabiliriz.

````python
kelimeler = ["python", "veri", "python", "kod", "veri", "python"]

sayac = {}

for kelime in kelimeler:
    if kelime in sayac:
        sayac[kelime] += 1
    else:
        sayac[kelime] = 1

print(sayac)
````

Sonuç yaklaşık olarak şu mantıkta olacaktır:

```python
{
    "python": 3,
    "veri": 2,
    "kod": 1
}
```

Bu teknik veri analizinde oldukça değerlidir.

# 22. Daha Kısa Sayaç Mantığı

`get()` kullanarak aynı işlemi daha kısa yazabiliriz.

````python
kelimeler = ["python", "veri", "python", "kod", "veri", "python"]

sayac = {}

for kelime in kelimeler:
    sayac[kelime] = sayac.get(kelime, 0) + 1

print(sayac)
````

# 23. Harf Frekansı

Bir metindeki harflerin kaç kez geçtiğini bulabiliriz.

````python
metin = "python"

frekans = {}

for harf in metin:
    frekans[harf] = frekans.get(harf, 0) + 1

print(frekans)
````

Bu mantık daha sonra metin analizi ve yapay zeka çalışmalarında daha gelişmiş biçimde kullanılabilir.

# 24. Ürün Fiyatları

Dictionary, anahtar-değer yapısı nedeniyle ürün-fiyat eşleştirmelerinde çok uygundur.

````python
urunler = {
    "Laptop": 35000,
    "Tablet": 18000,
    "Telefon": 25000,
    "Kulaklık": 2500
}

print("Laptop fiyatı:", urunler["Laptop"])
````

Belirli fiyatın üzerindeki ürünleri bulabiliriz.

````python
urunler = {
    "Laptop": 35000,
    "Tablet": 18000,
    "Telefon": 25000,
    "Kulaklık": 2500
}

for urun, fiyat in urunler.items():
    if fiyat >= 20000:
        print(urun, fiyat)
````

# 25. En Pahalı Ürünü Bulmak

````python
urunler = {
    "Laptop": 35000,
    "Tablet": 18000,
    "Telefon": 25000,
    "Kulaklık": 2500
}

en_pahali_urun = None
en_yuksek_fiyat = 0

for urun, fiyat in urunler.items():
    if fiyat > en_yuksek_fiyat:
        en_yuksek_fiyat = fiyat
        en_pahali_urun = urun

print("En pahalı ürün:", en_pahali_urun)
print("Fiyat:", en_yuksek_fiyat)
````

# 26. Sözlüklerde Fonksiyon Kullanımı

Bir dictionary'yi fonksiyona gönderebiliriz.

````python
def ogrenci_goster(ogrenci):
    print("İsim:", ogrenci["isim"])
    print("Yaş:", ogrenci["yas"])
    print("Puan:", ogrenci["puan"])

ogrenci = {
    "isim": "Ali",
    "yas": 14,
    "puan": 85
}

ogrenci_goster(ogrenci)
````

Bu konu bir önceki dersteki fonksiyonlarla doğrudan bağlantılıdır.

Büyük programlarda veriyi dictionary ile, işlemleri ise fonksiyonlarla düzenlemek oldukça yaygın bir yaklaşımdır.

# 27. Dictionary ve Fonksiyon ile Not Analizi

````python
def ortalama_hesapla(notlar):
    return sum(notlar.values()) / len(notlar)

notlar = {
    "Python": 90,
    "Matematik": 80,
    "Fizik": 75,
    "Biyoloji": 85
}

print("Ortalama:", ortalama_hesapla(notlar))
````

# 28. Dictionary'den En Yüksek Notu Bulmak

````python
def en_yuksek_not(notlar):
    ders = None
    puan = 0

    for isim, not_degeri in notlar.items():
        if not_degeri > puan:
            ders = isim
            puan = not_degeri

    return ders, puan

notlar = {
    "Python": 90,
    "Matematik": 80,
    "Fizik": 75,
    "Biyoloji": 95
}

ders, puan = en_yuksek_not(notlar)

print("En yüksek ders:", ders)
print("Not:", puan)
````

# 34. Mini Uygulama: Not Defteri Analizi

````python
notlar = {
    "Python": 95,
    "Matematik": 80,
    "Fizik": 45,
    "Biyoloji": 90
}

toplam = sum(notlar.values())
ortalama = toplam / len(notlar)

print("Ortalama:", ortalama)

for ders, puan in notlar.items():
    if puan >= 50:
        print(ders, "başarılı")
    else:
        print(ders, "başarısız")
````

# 35. Mini Uygulama: Kelime Sayacı

````python
metin = "python veri python kod veri python analiz"

kelimeler = metin.split()
sayac = {}

for kelime in kelimeler:
    sayac[kelime] = sayac.get(kelime, 0) + 1

for kelime, adet in sayac.items():
    print(kelime, ":", adet)
````

# 37. Mini Uygulama: Sınıf Analizi

````python
ogrenciler = [
    {"isim": "Ali", "puan": 85},
    {"isim": "Ayşe", "puan": 92},
    {"isim": "Mehmet", "puan": 45},
    {"isim": "Zeynep", "puan": 78},
    {"isim": "Can", "puan": 60}
]

basarili = {}
basarisiz = {}

for ogrenci in ogrenciler:
    if ogrenci["puan"] >= 50:
        basarili[ogrenci["isim"]] = ogrenci["puan"]
    else:
        basarisiz[ogrenci["isim"]] = ogrenci["puan"]

print("Başarılı:", basarili)
print("Başarısız:", basarisiz)
````

# 38. Dictionary ve Set Birlikte

Dictionary'nin anahtarları benzersizdir. Bu nedenle bazı durumlarda set ile benzer bir mantık görürüz.

Ancak aralarında önemli fark vardır:

**Set:** Benzersiz değerleri tutar.

**Dictionary:** Benzersiz anahtarları ve bu anahtarlara karşılık gelen değerleri tutar.

Örneğin:

```python
sehirler = {"İstanbul", "Ankara", "İzmir"}
```

sadece şehirleri tutar.

```python
plakalar = {
    "İstanbul": 34,
    "Ankara": 06,
    "İzmir": 35
}
```

ise şehir ile plaka arasında ilişki kurar.


# 39. Dictionary ile Eşleştirme

````python
plakalar = {
    "İstanbul": 34,
    "Ankara": 6,
    "İzmir": 35,
    "Bursa": 16
}

sehir = "Ankara"

print("Plaka:", plakalar[sehir])
````

Bu kullanım, sözlüklerin neden gerçek hayattaki veriler için güçlü olduğunu gösterir: Bir bilgiyi başka bir bilgiyle eşleştirebiliriz.
