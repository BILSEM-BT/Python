Şimdiye kadar değişkenleri, veri türlerini, koşulları, hata yönetimini, döngüleri ve listeleri öğrendik. Artık bu yapıların üzerine fonksiyonları ekleyeceğiz.

Fonksiyon; belirli bir işi yapan, gerektiğinde tekrar çağrılabilen kod bloğudur.

Fonksiyonlar sayesinde kodu küçük parçalara ayırabilir, tekrarları azaltabilir ve daha büyük projeleri daha düzenli geliştirebiliriz.

İleride yapacağımız veri analizi, grafik, masaüstü uygulama, web uygulaması ve yapay zeka çalışmalarında fonksiyonları sürekli kullanacağız.


## 1. İlk Fonksiyon

Bir fonksiyon `def` anahtar kelimesiyle tanımlanır.

```python
def selamla():
    print("Merhaba!")
    print("Python dersine hoş geldiniz.")

selamla()
```

Fonksiyonu tanımlamak ile çalıştırmak farklıdır. `selamla()` çağrısı fonksiyonu çalıştırır.

## 2. Parametreler

Fonksiyona dışarıdan bilgi gönderebiliriz.

```python
def selamla(isim):
    print("Merhaba", isim)

selamla("Ali")
selamla("Ayşe")
selamla("Mehmet")
```

Burada `isim` parametredir. Çağrı sırasında gönderilen değerler argümandır.

- Parametre: fonksiyon tanımındaki değişken
- Argüman: fonksiyon çağrısında gönderilen değer


## 3. Birden Fazla Parametre

```python
def ogrenci_bilgisi(isim, yas, sinif):
    print("İsim:", isim)
    print("Yaş:", yas)
    print("Sınıf:", sinif)

ogrenci_bilgisi("Ali", 14, 8)
```

## 4. `return` ile Sonuç Döndürmek

Fonksiyon yalnızca ekrana yazdırmak zorunda değildir. Sonucu programa geri gönderebilir.

```python
def topla(a, b):
    return a + b

sonuc = topla(10, 20)
print("Sonuç:", sonuc)
```

`print()` sonucu ekrana gösterir; `return` sonucu fonksiyonun dışına gönderir. Dönen değer başka hesaplamalarda kullanılabilir.

```python
def topla(a, b):
    return a + b

sonuc = topla(10, 20)
print("İki katı:", sonuc * 2)
```

## 5. Fonksiyon İçinde Hesaplama

```python
def ortalama_hesapla(not1, not2, not3):
    toplam = not1 + not2 + not3
    ortalama = toplam / 3
    return ortalama

print(ortalama_hesapla(70, 80, 90))
```

## 6. Listeleri Fonksiyonlara Göndermek

```python
def liste_toplami(sayilar):
    return sum(sayilar)

sayilar = [10, 20, 30, 40]
print(liste_toplami(sayilar))
```

```python
def ortalama(sayilar):
    return sum(sayilar) / len(sayilar)

notlar = [70, 80, 90, 60, 75]
print("Ortalama:", ortalama(notlar))
```

## 7. Varsayılan Parametreler

```python
def selamla(isim="Öğrenci"):
    print("Merhaba", isim)

selamla("Ali")
selamla()
```

## 8. Keyword Argümanları

```python
def ogrenci_bilgisi(isim, yas, sinif):
    print(isim, yas, sinif)

ogrenci_bilgisi(sinif=8, isim="Zeynep", yas=14)
```

## 9. Birden Fazla Değer Döndürmek

```python
def hesapla(a, b):
    toplam = a + b
    fark = a - b
    carpim = a * b
    return toplam, fark, carpim

toplam, fark, carpim = hesapla(10, 5)

print("Toplam:", toplam)
print("Fark:", fark)
print("Çarpım:", carpim)
```

## 10. Fonksiyon + `if`

```python
def not_durumu(not_degeri):
    if not_degeri >= 50:
        return "Başarılı"
    else:
        return "Başarısız"

print(not_durumu(75))
print(not_durumu(40))
```

## 11. Fonksiyon + Döngü + Liste

```python
def cift_sayilar(sayilar):
    sonuc = []

    for sayi in sayilar:
        if sayi % 2 == 0:
            sonuc.append(sayi)

    return sonuc

sayilar = [1, 2, 3, 4, 5, 6, 7, 8]
print(cift_sayilar(sayilar))
```

## 12. Fonksiyon + `try-except`

Fonksiyonların içinde hata yönetimi yapabiliriz.

```python
def bol(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Bir sayı sıfıra bölünemez."

print(bol(10, 2))
print(bol(10, 0))
```

Kullanıcıdan güvenli sayı alan bir fonksiyon da yazabiliriz.

```python
def sayi_al():
    while True:
        try:
            return int(input("Bir tam sayı girin: "))
        except ValueError:
            print("Lütfen geçerli bir tam sayı girin.")

sayi = sayi_al()
print("Girilen sayı:", sayi)
```

## 13. `None`

`return` kullanılmadan biten bir fonksiyon `None` döndürür.

```python
def merhaba():
    print("Merhaba")

sonuc = merhaba()
print("Fonksiyon sonucu:", sonuc)
```

## 14. Yerel ve Global Değişkenler

Fonksiyon içinde oluşturulan değişkenler yereldir.

```python
def hesapla():
    sayi = 10
    print("Fonksiyon içinde:", sayi)

hesapla()
```

Fonksiyon dışında oluşturulan değişkenler global kapsamda bulunabilir. Büyük programlarda veriyi mümkün olduğunca parametrelerle geçirmek kodun daha anlaşılır olmasını sağlar.

```python
mesaj = "Merhaba"

def yazdir():
    print(mesaj)

yazdir()
```

## 15. Fonksiyonların Birbirini Çağırması

```python
def kare(sayi):
    return sayi ** 2

def kup(sayi):
    return sayi ** 3

def hesapla(sayi):
    return kare(sayi), kup(sayi)

print(hesapla(4))
```

## 16. Liste Analizi Fonksiyonu

```python
def liste_analizi(sayilar):
    en_kucuk = min(sayilar)
    en_buyuk = max(sayilar)
    toplam = sum(sayilar)
    ortalama = toplam / len(sayilar)

    return en_kucuk, en_buyuk, toplam, ortalama

veriler = [15, 20, 35, 40, 50]

kucuk, buyuk, toplam, ortalama = liste_analizi(veriler)

print("En küçük:", kucuk)
print("En büyük:", buyuk)
print("Toplam:", toplam)
print("Ortalama:", ortalama)
```

## 17. Başarılı Notları Bulan Fonksiyon

```python
def basarili_notlar(notlar):
    sonuc = []

    for notu in notlar:
        if notu >= 50:
            sonuc.append(notu)

    return sonuc

notlar = [35, 60, 75, 40, 90, 55]
print(basarili_notlar(notlar))
```

Fonksiyonları mümkün olduğunca tek ve belirgin bir görev etrafında tasarlamak iyi bir programlama alışkanlığıdır.

## 18. Fonksiyonları Küçük Görevlere Bölmek

```python
def toplam_hesapla(sayilar):
    return sum(sayilar)

def ortalama_hesapla(sayilar):
    return sum(sayilar) / len(sayilar)

def en_buyuk_bul(sayilar):
    return max(sayilar)

sayilar = [10, 20, 30, 40, 50]

print("Toplam:", toplam_hesapla(sayilar))
print("Ortalama:", ortalama_hesapla(sayilar))
print("En büyük:", en_buyuk_bul(sayilar))
```

## 19. Docstring

Fonksiyonun amacını açıklamak için docstring kullanabiliriz.

```python
def kare(sayi):
    """Gelen sayının karesini döndürür."""
    return sayi ** 2

print(kare(5))
```

## 20. Tip İpuçları

```python
def topla(a: int, b: int) -> int:
    return a + b

print(topla(10, 20))
```

Tip ipuçları beklenen veri türlerini ifade eder; Python tarafından zorunlu bir tür kontrolü olarak uygulanmaz. Kodun okunmasını ve geliştirme araçlarının yardımını kolaylaştırır.

## 21. Fonksiyon ve Liste Kopyalama

```python
def yeni_liste_olustur(liste):
    yeni_liste = liste.copy()
    yeni_liste.append(100)
    return yeni_liste

sayilar = [10, 20, 30]
sonuc = yeni_liste_olustur(sayilar)

print("Orijinal:", sayilar)
print("Yeni:", sonuc)
```

## 22. `*args`

Değişken sayıda konumsal argüman almak için `*args` kullanılabilir.

```python
def toplam(*sayilar):
    return sum(sayilar)

print(toplam(10, 20))
print(toplam(10, 20, 30, 40))
```

`*args` ile gelen değerler fonksiyon içinde tuple olarak bulunur.

```python
def yazdir(*degerler):
    print(degerler)

yazdir("Ali", "Ayşe", "Mehmet")
```

## 23. `**kwargs`

Değişken sayıda isimlendirilmiş argüman için `**kwargs` kullanılabilir.

```python
def bilgiler(**veriler):
    print(veriler)

bilgiler(isim="Ali", yas=14, sinif=8)
```

## 24. Mini Proje: Hesap Makinesi

```python
def topla(a, b):
    return a + b

def cikar(a, b):
    return a - b

def carp(a, b):
    return a * b

def bol(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Sıfıra bölme yapılamaz."

print("Toplam:", topla(10, 5))
print("Fark:", cikar(10, 5))
print("Çarpım:", carp(10, 5))
print("Bölüm:", bol(10, 5))
print("Hatalı bölüm:", bol(10, 0))
```

## 25. Mini Proje: Öğrenci Not Analizi

```python
def ortalama_hesapla(notlar):
    return sum(notlar) / len(notlar)

def basarili_sayisi(notlar):
    sayac = 0

    for notu in notlar:
        if notu >= 50:
            sayac += 1

    return sayac

def en_yuksek_not(notlar):
    return max(notlar)

notlar = [45, 70, 82, 35, 91, 64]

ortalama = ortalama_hesapla(notlar)
basarili = basarili_sayisi(notlar)
en_yuksek = en_yuksek_not(notlar)

print("Ortalama:", ortalama)
print("Başarılı sayısı:", basarili)
print("En yüksek not:", en_yuksek)
```

## 26. Mini Proje: Sıcaklık Analizi

```python
def ortalama_sicaklik(sicakliklar):
    return sum(sicakliklar) / len(sicakliklar)

def sicak_gun_sayisi(sicakliklar, esik=25):
    sayac = 0

    for sicaklik in sicakliklar:
        if sicaklik >= esik:
            sayac += 1

    return sayac

sicakliklar = [18, 21, 25, 29, 27, 16, 22]

print("Ortalama:", ortalama_sicaklik(sicakliklar))
print("25 derece ve üzeri:", sicak_gun_sayisi(sicakliklar))
print("20 derece ve üzeri:", sicak_gun_sayisi(sicakliklar, 20))
```

## 27. Fonksiyon Tasarlama Mantığı

1. Hangi işi yapacak?
2. Hangi verilere ihtiyaç duyuyor?
3. Bu veriler parametre olmalı mı?
4. Sonuç üretecek mi?
5. Üretecekse ne döndürecek?
6. Hatalı veri gelirse ne olacak?
7. Aynı işlem başka yerlerde de kullanılacak mı?

Örneğin "listedeki başarılı notları bul" problemi:

- Girdi: not listesi
- İşlem: notları dolaş
- Koşul: not >= 50
- Çıktı: başarılı notların listesi

şeklinde tasarlanabilir.
