## 1. Dosya İşlemleri Neden Gereklidir?

Bir programın öğrenci kayıtlarını, ürünleri, notları, günlükleri veya ayarları program kapandıktan sonra da hatırlamasını istiyorsak verileri kalıcı bir yere kaydetmeliyiz.

Örneğin:

```python
ogrenci = {
    "isim": "Ali",
    "puan": 90
}
```

Bu bilgi bellekte tutulur. Program kapandığında değişkenin kendisi kaybolur. Dosyaya kaydettiğimizde ise program daha sonra bu bilgiyi tekrar okuyabilir.

Temel yaklaşımımız:

**Dosya → Python → Veri yapısı → İşlem → Dosya**


## 2. `open()` Fonksiyonu

Python'da dosyayla çalışmanın temel fonksiyonu `open()`'dır.

```python
open("dosya.txt", "r")
```

İlk parametre dosyanın adı veya yoludur. İkinci parametre dosyanın hangi amaçla açılacağını belirtir.


````python
dosya = open("deneme.txt", "w", encoding="utf-8")
dosya.write("Merhaba Python!")
dosya.close()
````

Burada dosyayı açtık, veri yazdık ve kapattık. `close()` dosyayı kapatır. Fakat modern Python kodlarında çoğu zaman `with open()` kullanacağız; böylece dosyanın kapatılmasını Python'a bırakabiliriz.


## 3. Dosya Açma Modları

| Mod | Görevi |
|---|---|
| `r` | Okuma |
| `w` | Yazma; varsa eski içeriğin üzerine yazar |
| `a` | Dosyanın sonuna ekleme |
| `x` | Yeni dosya oluşturma; dosya varsa hata |

`w` ile `a` arasındaki fark özellikle önemlidir.


## 4. `w` Modu

`w` dosya yoksa oluşturur. Dosya varsa mevcut içeriğin üzerine yazar. Bu nedenle mevcut veriyi korumak istediğimiz durumlarda dikkatli kullanılmalıdır.


````python
with open("mesaj.txt", "w", encoding="utf-8") as dosya:
    dosya.write("İlk içerik")

with open("mesaj.txt", "w", encoding="utf-8") as dosya:
    dosya.write("Yeni içerik")

with open("mesaj.txt", "r", encoding="utf-8") as dosya:
    print(dosya.read())
````

## 5. `a` Modu

`a` mevcut içeriği silmeden dosyanın sonuna veri ekler. Günlük kayıtları, işlem geçmişi ve benzeri uygulamalarda kullanışlıdır.


````python
with open("gunluk.txt", "a", encoding="utf-8") as dosya:
    dosya.write("Birinci kayıt\n")
    dosya.write("İkinci kayıt\n")

with open("gunluk.txt", "r", encoding="utf-8") as dosya:
    print(dosya.read())
````

## 6. `x` Modu

`x` yeni dosya oluşturmak için kullanılır. Dosya zaten varsa `FileExistsError` oluşur.


````python
try:
    with open("yeni_dosya.txt", "x", encoding="utf-8") as dosya:
        dosya.write("Yeni dosya oluşturuldu.")
except FileExistsError:
    print("Bu dosya zaten mevcut.")
````

## 7. Dosya Okuma: `read()`

`read()` dosyanın tamamını tek bir metin olarak okuyabilir.


````python
with open("mesaj.txt", "r", encoding="utf-8") as dosya:
    icerik = dosya.read()

print(icerik)
````

`read()` içine bir sayı verirsek belirtilen sayıda karakter okumaya çalışır.


````python
with open("mesaj.txt", "r", encoding="utf-8") as dosya:
    print(dosya.read(5))
````

## 8. `readline()`

`readline()` dosyadan bir satır okur. Her çağrıda bir sonraki satıra geçilir.


````python
with open("gunluk.txt", "r", encoding="utf-8") as dosya:
    print(dosya.readline().strip())
    print(dosya.readline().strip())
````

## 9. `readlines()`

`readlines()` satırları bir liste olarak döndürür.


````python
with open("gunluk.txt", "r", encoding="utf-8") as dosya:
    satirlar = dosya.readlines()

print(satirlar)
````

## 10. `for` Döngüsü ile Dosya Okuma

Dosyayı satır satır işlemek için `for` döngüsü çok kullanışlıdır.


````python
with open("gunluk.txt", "r", encoding="utf-8") as dosya:
    for satir in dosya:
        print(satir.strip())
````

## 11. `strip()` ve `split()`

`strip()` satır sonundaki `\n` gibi karakterleri ve gereksiz boşlukları temizlemek için kullanılabilir. `split()` ise bir metni belirlediğimiz ayırıcıdan parçalara böler.


````python
metin = "Ali,85"
isim, puan = metin.split(",")
print(isim)
print(puan)
print(int(puan) + 5)
````

## 12. Listeyi Dosyaya Kaydetmek

Listelerdeki verileri satır satır dosyaya yazabiliriz.


````python
meyveler = ["Elma", "Armut", "Muz", "Portakal"]

with open("meyveler.txt", "w", encoding="utf-8") as dosya:
    for meyve in meyveler:
        dosya.write(meyve + "\n")

with open("meyveler.txt", "r", encoding="utf-8") as dosya:
    meyveler = [satir.strip() for satir in dosya]

print(meyveler)
````

## 13. Dosya Verisini Dictionary'ye Aktarmak

Şimdi daha gerçekçi bir örnek yapalım. Dosyada şu biçimde öğrenci kayıtlarımız olduğunu düşünelim:

```text
Ali,85
Ayşe,92
Mehmet,76
Zeynep,88
```

Bu verileri okuyup dictionary içinde saklayabiliriz.


````python
with open("ogrenciler.txt", "w", encoding="utf-8") as dosya:
    dosya.write("Ali,85\n")
    dosya.write("Ayşe,92\n")
    dosya.write("Mehmet,76\n")
    dosya.write("Zeynep,88\n")

ogrenciler = {}

with open("ogrenciler.txt", "r", encoding="utf-8") as dosya:
    for satir in dosya:
        isim, puan = satir.strip().split(",")
        ogrenciler[isim] = int(puan)

print(ogrenciler)
````

Artık dosya verisini Python veri yapısına dönüştürdük. Bu noktadan sonra normal Python işlemlerini uygulayabiliriz.


````python
ortalama = sum(ogrenciler.values()) / len(ogrenciler)
print("Sınıf ortalaması:", ortalama)

for isim, puan in ogrenciler.items():
    if puan >= 90:
        print(isim, "yüksek başarı gösterdi.")
````

## 14. `with open()` Kullanımı

Dosyayı `open()` ile açıp `close()` ile kapatmak mümkündür. Ancak dosya işlemlerinde `with open()` kullanmak daha güvenli ve okunabilir bir yaklaşımdır.

```python
with open("dosya.txt", "r", encoding="utf-8") as dosya:
    ...
```

`with` bloğu sona erdiğinde dosya otomatik olarak kapatılır. Derslerimizde mümkün olduğunca bu yöntemi kullanacağız.


## 15. Türkçe Karakterler ve UTF-8

Türkçe karakterlerle çalışırken `encoding="utf-8"` kullanmak iyi bir alışkanlıktır.


````python
with open("turkce.txt", "w", encoding="utf-8") as dosya:
    dosya.write("Çalışıyorum, öğreniyorum ve geliştiriyorum.")

with open("turkce.txt", "r", encoding="utf-8") as dosya:
    print(dosya.read())
````

## 16. Dosya Hataları ve `try-except`

Programlarımızda dosyanın bulunmaması, erişim izninin olmaması veya dosyanın zaten var olması gibi durumlar oluşabilir. Bu durumları hata yönetimi ile kontrol edebiliriz.

Özellikle şu hataları tanımamız gerekir:

- `FileNotFoundError`
- `PermissionError`
- `FileExistsError`


````python
try:
    with open("olmayan_dosya.txt", "r", encoding="utf-8") as dosya:
        print(dosya.read())
except FileNotFoundError:
    print("Dosya bulunamadı!")
except PermissionError:
    print("Dosyaya erişim izniniz yok!")
finally:
    print("Dosya işlemi tamamlandı.")
````

`try` riskli işlemi dener. Uygun `except` bloğu oluşan hatayı yakalar. `finally` ise hata oluşsa da oluşmasa da çalışır. Bu yapı gerçek uygulamalarda kullanıcıya anlaşılır mesajlar göstermek için önemlidir.

## 19. Klasör Oluşturmak

`os.mkdir()` ile klasör oluşturabiliriz. Önce klasörün mevcut olup olmadığını kontrol etmek iyi bir uygulamadır.

## 20. Dosya Silmek

`os.remove()` dosya silmek için kullanılabilir. Silme işlemi geri alınamayabileceği için önce varlık kontrolü yapmak güvenlidir.

# 21. JSON Nedir?

JSON, yapılandırılmış verileri saklamak ve uygulamalar arasında taşımak için kullanılan çok yaygın bir veri formatıdır. Python dictionary ve listelerine benzer bir yapısı vardır.

Örneğin:

```json
{
    "isim": "Ali",
    "yas": 14,
    "puan": 90
}
```

JSON; web uygulamalarında, API'lerde, veri aktarımında ve yapay zeka uygulamalarında sıkça karşımıza çıkar.


## 22. Dictionary'yi JSON Dosyasına Kaydetmek

Python'da JSON işlemleri için `json` modülünü kullanırız.


````python
import json

ogrenci = {
    "isim": "Ali",
    "yas": 14,
    "puan": 90
}

with open("ogrenci.json", "w", encoding="utf-8") as dosya:
    json.dump(ogrenci, dosya, ensure_ascii=False, indent=4)

print("JSON dosyası oluşturuldu.")
````

`json.dump()` Python verisini JSON dosyasına yazar. `ensure_ascii=False` Türkçe karakterlerin korunmasına, `indent=4` ise dosyanın okunabilir biçimde düzenlenmesine yardımcı olur.


## 23. JSON Dosyasını Okumak

`json.load()` JSON dosyasındaki veriyi tekrar Python veri yapısına dönüştürür.


````python
import json

with open("ogrenci.json", "r", encoding="utf-8") as dosya:
    ogrenci = json.load(dosya)

print(ogrenci)
print(ogrenci["isim"])
print(ogrenci["puan"])
````

## 24. Birden Fazla Öğrenciyi JSON'a Kaydetmek

Liste içinde dictionary kullanımı gerçek uygulamalarda çok yaygındır. Bu yapıyı JSON dosyasına doğrudan kaydedebiliriz.


````python
import json

ogrenciler = [
    {"isim": "Ali", "puan": 85},
    {"isim": "Ayşe", "puan": 92},
    {"isim": "Mehmet", "puan": 76},
    {"isim": "Zeynep", "puan": 88}
]

with open("ogrenciler.json", "w", encoding="utf-8") as dosya:
    json.dump(ogrenciler, dosya, ensure_ascii=False, indent=4)

with open("ogrenciler.json", "r", encoding="utf-8") as dosya:
    veriler = json.load(dosya)

for ogrenci in veriler:
    print(ogrenci["isim"], "->", ogrenci["puan"])
````

## 25. JSON Verisini Güncellemek

JSON dosyasındaki veriyi okuyup Python içinde değiştirebilir ve tekrar dosyaya yazabiliriz.

Bu mantık ileride oluşturacağımız kayıt sistemlerinin temelidir:

1. Veriyi oku.
2. Python içinde değiştir.
3. Güncel veriyi tekrar kaydet.


````python
import json

with open("ogrenciler.json", "r", encoding="utf-8") as dosya:
    ogrenciler = json.load(dosya)

ogrenciler[0]["puan"] = 95

with open("ogrenciler.json", "w", encoding="utf-8") as dosya:
    json.dump(ogrenciler, dosya, ensure_ascii=False, indent=4)

print(ogrenciler)
````

## 26. JSON Hatalarını Yönetmek

JSON dosyası bozuksa `json.JSONDecodeError` oluşabilir. Dosyanın bulunmaması ise `FileNotFoundError` oluşturur.


````python
import json

try:
    with open("ogrenciler.json", "r", encoding="utf-8") as dosya:
        ogrenciler = json.load(dosya)
    print(ogrenciler)
except FileNotFoundError:
    print("JSON dosyası bulunamadı!")
except json.JSONDecodeError:
    print("JSON dosyasının biçimi hatalı!")
````

# 27. Fonksiyon + Dosya İşlemleri

Önceki derslerde öğrendiğimiz fonksiyonları dosya işlemleriyle birleştirebiliriz. Böylece programımız daha düzenli ve tekrar kullanılabilir hale gelir.


````python
def ogrenci_kaydet(isim, puan):
    with open("kayitlar.txt", "a", encoding="utf-8") as dosya:
        dosya.write(f"{isim},{puan}\n")


def ogrencileri_goster():
    try:
        with open("kayitlar.txt", "r", encoding="utf-8") as dosya:
            for satir in dosya:
                print(satir.strip())
    except FileNotFoundError:
        print("Henüz kayıt bulunmuyor.")

ogrenci_kaydet("Ali", 90)
ogrenci_kaydet("Ayşe", 95)
ogrenci_kaydet("Mehmet", 80)
ogrencileri_goster()
````

# 28. Mini Uygulama: JSON Tabanlı Öğrenci Kayıt Sistemi

Şimdi dosya işlemleri, dictionary, listeler, fonksiyonlar ve hata yönetimini aynı uygulamada birleştirelim.


````python
import json

def ogrencileri_yukle():
    try:
        with open("ogrenciler.json", "r", encoding="utf-8") as dosya:
            return json.load(dosya)
    except FileNotFoundError:
        return []
    except json.JSONDecodeError:
        print("Veri dosyası bozuk.")
        return []


def ogrencileri_kaydet(ogrenciler):
    with open("ogrenciler.json", "w", encoding="utf-8") as dosya:
        json.dump(ogrenciler, dosya, ensure_ascii=False, indent=4)


ogrenciler = ogrencileri_yukle()

ogrenciler.append({
    "isim": "Can",
    "puan": 87
})

ogrencileri_kaydet(ogrenciler)

print(ogrenciler)
````

# 29. Mini Uygulama: Ürün ve Stok Kaydı

Aynı mantığı ürünler için de kullanabiliriz.


````python
import json

urunler = [
    {"urun": "Defter", "stok": 50, "fiyat": 35},
    {"urun": "Kalem", "stok": 120, "fiyat": 15},
    {"urun": "Silgi", "stok": 80, "fiyat": 10}
]

with open("urunler.json", "w", encoding="utf-8") as dosya:
    json.dump(urunler, dosya, ensure_ascii=False, indent=4)

with open("urunler.json", "r", encoding="utf-8") as dosya:
    urunler = json.load(dosya)

for urun in urunler:
    toplam_deger = urun["stok"] * urun["fiyat"]
    print(urun["urun"], "stok değeri:", toplam_deger)
````

# 30. Dosyadan Veri Analizine

Artık dosyadan veri okuyup Python veri yapılarına aktarabiliyoruz. Bu, sonraki aşamadaki veri analizi çalışmalarımız için önemli bir adımdır.

Örneğin bir dosyada binlerce ölçüm olduğunu düşünün:

```text
23.4
24.1
22.8
25.6
...
```

Bu verileri Python'a alıp:

- toplamını,
- ortalamasını,
- en yüksek değerini,
- en düşük değerini

hesaplayabiliriz. Daha sonra bu verileri grafiklerle gösterebiliriz.


````python
with open("sicaklik.txt", "w", encoding="utf-8") as dosya:
    for sicaklik in [23.4, 24.1, 22.8, 25.6, 26.2, 24.9]:
        dosya.write(f"{sicaklik}\n")

sicakliklar = []

with open("sicaklik.txt", "r", encoding="utf-8") as dosya:
    for satir in dosya:
        sicakliklar.append(float(satir.strip()))

print("Veriler:", sicakliklar)
print("Ortalama:", sum(sicakliklar) / len(sicakliklar))
print("En düşük:", min(sicakliklar))
print("En yüksek:", max(sicakliklar))
````
