# 1. Modül Nedir?

Programlarımız büyüdükçe bütün kodları tek dosyada tutmak zorlaşır. Python'da kodlarımızı farklı dosyalara ayırabilir ve gerektiğinde başka programlarda kullanabiliriz.

Bu şekilde kullanılan Python dosyalarına **modül** denir.

Modüller sayesinde:
- kodlarımızı düzenleriz,
- aynı kodu tekrar tekrar yazmayız,
- projeleri parçalara ayırırız,
- hazır araçlardan yararlanırız.


# 2. `import` Kullanımı

Bir modülü kullanmak için `import` komutunu kullanırız.

````python
import math

print(math.pi)
print(math.sqrt(25))
print(math.pow(2, 3))
````

# 3. `math` Modülü

`math`, matematiksel işlemler için birçok hazır fonksiyon ve sabit içerir.

````python
import math

sayi = 25

print("Karekök:", math.sqrt(sayi))
print("Pi:", math.pi)
print("2 üzeri 5:", math.pow(2, 5))
print("Yukarı yuvarlama:", math.ceil(4.2))
print("Aşağı yuvarlama:", math.floor(4.8))
print("Faktöriyel:", math.factorial(5))
````

Sık kullanılanlar:

- `math.sqrt()` → karekök
- `math.pow()` → üs
- `math.ceil()` → yukarı yuvarlama
- `math.floor()` → aşağı yuvarlama
- `math.factorial()` → faktöriyel
- `math.pi` → pi sayısı

Her fonksiyonu ezberlemek yerine ihtiyaç olduğunda dokümantasyondan bulabilmek daha önemlidir.


# 4. `random` Modülü

Rastgele sayılar ve seçimler için `random` modülünü kullanabiliriz.

````python
import random

print(random.randint(1, 100))

for i in range(5):
    print(random.randint(1, 100))
````

# 7. `datetime` Modülü

Tarih ve saat işlemleri için `datetime` kullanılır.

````python
from datetime import datetime

simdi = datetime.now()

print(simdi)
print("Yıl:", simdi.year)
print("Ay:", simdi.month)
print("Gün:", simdi.day)
print("Saat:", simdi.hour)
````

# 8. Tarihi Biçimlendirmek

````python
from datetime import datetime

simdi = datetime.now()

print(simdi.strftime("%d.%m.%Y"))
print(simdi.strftime("%d.%m.%Y %H:%M:%S"))
````

`strftime()` içinde sık kullanılan biçimler:

- `%d` → gün
- `%m` → ay
- `%Y` → yıl
- `%H` → saat
- `%M` → dakika
- `%S` → saniye


# 9. Kendi Modülümüzü Oluşturmak

Visual Studio'da aynı klasörde `matematik.py` ve `ana_program.py` dosyalarını oluşturduğumuzu düşünelim.

`matematik.py`:

````python
def topla(a, b):
    return a + b

def cikar(a, b):
    return a - b

def carp(a, b):
    return a * b

def bol(a, b):
    if b == 0:
        return "Sıfıra bölme yapılamaz."
    return a / b
````

`ana_program.py`:

````python
import matematik

print(matematik.topla(10, 5))
print(matematik.cikar(10, 5))
print(matematik.carp(10, 5))
print(matematik.bol(10, 5))
````

Böylece matematik işlemleri ayrı bir dosyada, ana program ayrı bir dosyada tutulmuş olur.

Bu yaklaşım büyük projelerde kodun düzenlenmesini ve tekrar kullanılmasını kolaylaştırır.

# 13. `statistics` Modülü

````python
import statistics

puanlar = [70, 85, 90, 95, 80]

print("Ortalama:", statistics.mean(puanlar))
print("Ortanca:", statistics.median(puanlar))
print("En küçük:", min(puanlar))
print("En büyük:", max(puanlar))
````

# 18. Matplotlib'e İlk Bakış

````python
import matplotlib.pyplot as plt

isimler = ["Ali", "Ayşe", "Mehmet", "Zeynep"]
puanlar = [85, 92, 76, 88]

plt.bar(isimler, puanlar)
plt.title("Öğrenci Puanları")
plt.xlabel("Öğrenciler")
plt.ylabel("Puan")
plt.show()
````

# 21. Modüller ile Dosya İşlemlerini Birleştirmek

9. derste öğrendiğimiz dosya işlemlerini ayrı bir modüle taşıyabiliriz.

`dosya_islemleri.py`:

````python
def kaydet(veri):
    with open("veriler.txt", "a", encoding="utf-8") as dosya:
        dosya.write(veri + "\n")

def oku():
    try:
        with open("veriler.txt", "r", encoding="utf-8") as dosya:
            return dosya.readlines()
    except FileNotFoundError:
        return []
````

Ana programda:

````python
import dosya_islemleri

dosya_islemleri.kaydet("Python öğreniyorum")

veriler = dosya_islemleri.oku()

for veri in veriler:
    print(veri.strip())
````

Bu örnek, önceki ders ile bugünkü dersin doğrudan birleştiği önemli bir noktadır.

Artık dosya işlemleri de ana programdan ayrılmış durumdadır.


# 22. Hata Yönetimi ile Modüller

Fonksiyonların içinde hata yönetimi kullanabiliriz.

````python
def sayiyi_oku():
    try:
        sayi = int(input("Bir sayı girin: "))
        return sayi
    except ValueError:
        print("Lütfen geçerli bir sayı girin.")
        return None

sonuc = sayiyi_oku()
print("Sonuç:", sonuc)
````

Böylece ana programın kodunu sade tutabilir, belirli görevleri fonksiyon ve modüllere dağıtabiliriz.
