# 1. NumPy Nedir?

Python listeleriyle birçok işlem yapabiliriz. Ancak çok sayıda sayısal veriyle çalışmaya başladığımızda NumPy bize daha uygun araçlar sunar.

NumPy özellikle:

- büyük sayı kümeleri,
- matematiksel işlemler,
- çok boyutlu diziler,
- istatistiksel hesaplamalar,
- veri analizi,
- makine öğrenmesi,
- yapay zeka

çalışmalarında kullanılır.

NumPy'nin temel veri yapısına **ndarray** denir.

# 35. Koşullu Filtreleme

Koşulu sağlayan değerleri doğrudan seçebiliriz.

````python
puanlar = np.array([45, 72, 88, 55, 91, 63, 100])

print(puanlar[puanlar >= 70])
````

**Veri → Koşul → Filtrelenmiş veri**


# 36. Birden Fazla Koşul

NumPy dizilerinde birden fazla koşulu `&` ve `|` ile birleştirebiliriz.

````python
puanlar = np.array([45, 72, 88, 55, 91, 63, 100])

print(puanlar[(puanlar >= 70) & (puanlar < 90)])
print(puanlar[(puanlar < 50) | (puanlar > 95)])
````

# 41. `axis` Kavramına Giriş

İki boyutlu veri üzerinde satır veya sütun yönünde işlem yapmak için `axis` kullanılır.

````python
notlar = np.array([
    [80, 90, 70],
    [75, 85, 95],
    [90, 88, 92]
])

print("Genel ortalama:", np.mean(notlar))
print("Satır ortalamaları:", np.mean(notlar, axis=1))
print("Sütun ortalamaları:", np.mean(notlar, axis=0))
````

Genel olarak:

- `axis=0` → sütunlar üzerinde sonuç üretir
- `axis=1` → satırlar üzerinde sonuç üretir

Bu konu Pandas derslerinde de karşımıza çıkacaktır.


# 42. Gerçek Problem: Öğrenci Puanları

Dört öğrencinin üç sınav sonucunu analiz edelim.

````python
ogrenciler = np.array(["Ali", "Ayşe", "Mehmet", "Zeynep"])

notlar = np.array([
    [80, 90, 70],
    [75, 85, 95],
    [90, 88, 92],
    [60, 72, 68]
])

print(notlar)
````

Her öğrencinin ortalamasını hesaplayalım.

````python
ortalamalar = np.mean(notlar, axis=1)

for isim, ortalama in zip(ogrenciler, ortalamalar):
    print(isim, "->", round(ortalama, 2))
````

# 43. Her Sınavın Sınıf Ortalaması

````python
sinav_ortalamalari = np.mean(notlar, axis=0)

print("1. sınav:", round(sinav_ortalamalari[0], 2))
print("2. sınav:", round(sinav_ortalamalari[1], 2))
print("3. sınav:", round(sinav_ortalamalari[2], 2))
````

# 44. Başarılı Öğrencileri Filtrelemek

Ortalaması 80 ve üzerinde olan öğrencileri seçelim.

````python
ortalamalar = np.mean(notlar, axis=1)

basarili_ogrenciler = ogrenciler[ortalamalar >= 80]

print(basarili_ogrenciler)
````

# 45. En Yüksek Ortalamaya Sahip Öğrenci

````python
ortalamalar = np.mean(notlar, axis=1)

index = np.argmax(ortalamalar)

print("Öğrenci:", ogrenciler[index])
print("Ortalama:", round(ortalamalar[index], 2))
````

Bu örnekte `argmax()` ile en yüksek ortalamanın indeksini bulup aynı indeks üzerinden öğrenci adına ulaştık.

# 52. Veri Analizi Örneği: Sıcaklıklar

````python
sicakliklar = np.array([18, 20, 21, 19, 23, 25, 22])

print("Ortalama:", round(np.mean(sicakliklar), 2))
print("En yüksek:", np.max(sicakliklar))
print("En düşük:", np.min(sicakliklar))

print("22 derece ve üzeri:")
print(sicakliklar[sicakliklar >= 22])
````

# 53. Veri Analizi Örneği: Stok Miktarları

````python
stoklar = np.array([120, 95, 80, 65, 72, 50, 42])

print("Ortalama stok:", round(np.mean(stoklar), 2))
print("En yüksek stok:", np.max(stoklar))
print("En düşük stok:", np.min(stoklar))
print("50 ve altındaki stoklar:", stoklar[stoklar <= 50])
````

# 54. Veri Analizi Örneği: 50 Öğrenci Puanı

Test amacıyla 50 öğrencinin puanlarını üretelim.

````python
np.random.seed(10)

puanlar = np.random.randint(0, 101, 50)

print(puanlar)

print("Ortalama:", round(np.mean(puanlar), 2))
print("En yüksek:", np.max(puanlar))
print("En düşük:", np.min(puanlar))
print("Ortanca:", np.median(puanlar))
print("Standart sapma:", round(np.std(puanlar), 2))
print("70 ve üzeri öğrenci:", np.sum(puanlar >= 70))
print("50'nin altındaki öğrenci:", np.sum(puanlar < 50))
````

# 55. NumPy ile Fonksiyon Yazmak

Daha önce öğrendiğimiz fonksiyonları NumPy ile birleştirebiliriz.

````python
def veri_ozeti(veriler):
    return {
        "toplam": np.sum(veriler),
        "ortalama": np.mean(veriler),
        "minimum": np.min(veriler),
        "maksimum": np.max(veriler),
        "standart_sapma": np.std(veriler)
    }

sayilar = np.array([10, 15, 20, 25, 30])

sonuc = veri_ozeti(sayilar)

print(sonuc)
````

Burada NumPy, fonksiyon ve dictionary konularını birlikte kullandık.
