# 3. Series Nedir?

`Series`, tek sütunlu bir veri yapısı gibi düşünülebilir.

````python
import pandas as pd

puanlar = pd.Series([80, 90, 75, 95, 88])

print(puanlar)
````

```text

```

# 4. Series İçin Özel Index Kullanmak

````python
puanlar = pd.Series(
    [80, 90, 75, 95],
    index=["Ali", "Ayşe", "Mehmet", "Zeynep"]
)

print(puanlar)
````

````python
print("Ali:", puanlar["Ali"])
print("Zeynep:", puanlar["Zeynep"])
````

# 5. Series Üzerinde Temel Hesaplamalar

````python
puanlar = pd.Series([80, 90, 75, 95, 88])

print("Toplam:", puanlar.sum())
print("Ortalama:", puanlar.mean())
print("En düşük:", puanlar.min())
print("En yüksek:", puanlar.max())
print("Ortanca:", puanlar.median())
````

# 21. Koşullu Filtreleme

Pandas'ın en güçlü özelliklerinden biri koşula göre satır seçebilmektir.

````python
sonuc = df[df["Python"] >= 80]

print(sonuc)
````

# 22. Birden Fazla Koşul

Python notu 80 ve üzeri, Matematik notu 85 ve üzeri olan öğrenciler:

````python
sonuc = df[
    (df["Python"] >= 80) &
    (df["Matematik"] >= 85)
]

print(sonuc)
````

Pandas koşullarında:

- `&` → ve
- `|` → veya
- `~` → değil

kullanılabilir.

Her koşulu parantez içine almak önemlidir.

# 43. `str.contains()`

````python
isimler = pd.DataFrame({
    "İsim": ["Ali", "Ayşe", "Mehmet", "Zeynep", "Ahmet"]
})

sonuc = isimler[isimler["İsim"].str.contains("A", case=False)]

print(sonuc)
````

# 45. `apply()`

Bir fonksiyonu sütundaki değerlere uygulayabiliriz.

````python
def kategori(puan):
    if puan >= 90:
        return "Çok İyi"
    elif puan >= 75:
        return "İyi"
    else:
        return "Geliştirilmeli"

df["Python Kategorisi"] = df["Python"].apply(kategori)

print(df)
````

`apply()` özellikle kendi yazdığımız fonksiyonları veri tablolarına uygulamak için kullanışlıdır.


# 46. Gruplama: `groupby()`

Verileri kategorilere göre gruplandırıp analiz edebiliriz.

Sınıflara göre Python ortalamasını hesaplayalım.

````python
sonuc = df.groupby("Sınıf")["Python"].mean()

print(sonuc)
````

# 47. Birden Fazla İstatistik

`agg()` ile birden fazla hesaplamayı birlikte yapabiliriz.

````python
sonuc = df.groupby("Sınıf")["Python"].agg([
    "count",
    "mean",
    "min",
    "max"
])

print(sonuc)
````

# 48. İki Sütunu Birlikte Gruplamak

Daha büyük veri setlerinde birden fazla kategoriye göre gruplama yapılabilir.

````python
ornek = pd.DataFrame({
    "Sınıf": [7, 7, 8, 8, 8, 7],
    "Cinsiyet": ["K", "E", "K", "E", "K", "E"],
    "Puan": [80, 75, 90, 85, 95, 70]
})

print(ornek.groupby(["Sınıf", "Cinsiyet"])["Puan"].mean())
````

# 61. İki DataFrame'i Alt Alta Birleştirmek: `concat()`

````python
df1 = pd.DataFrame({
    "İsim": ["Ali", "Ayşe"],
    "Puan": [85, 92]
})

df2 = pd.DataFrame({
    "İsim": ["Mehmet", "Zeynep"],
    "Puan": [76, 88]
})

birlesik = pd.concat([df1, df2], ignore_index=True)

print(birlesik)
````

# 62. İki Tabloyu Ortak Sütun Üzerinden Birleştirmek: `merge()`

````python
ogrenci_bilgileri = pd.DataFrame({
    "No": [1, 2, 3],
    "İsim": ["Ali", "Ayşe", "Mehmet"]
})

not_bilgileri = pd.DataFrame({
    "No": [1, 2, 3],
    "Python": [90, 95, 70]
})

sonuc = pd.merge(
    ogrenci_bilgileri,
    not_bilgileri,
    on="No"
)

print(sonuc)
````

`merge()` mantığı daha sonra veri tabanı çalışmalarında göreceğimiz ilişkilendirme işlemlerine benzer.


# 63. Tarih Verileriyle Çalışmak

Pandas tarih verilerini özel bir veri türüne dönüştürebilir.

````python
tarih_df = pd.DataFrame({
    "Tarih": ["2026-01-10", "2026-02-15", "2026-03-20"],
    "Değer": [10, 20, 30]
})

tarih_df["Tarih"] = pd.to_datetime(tarih_df["Tarih"])

print(tarih_df)
print(tarih_df.dtypes)
````

# 64. Tarihten Yıl, Ay ve Gün Çıkarmak

# 68. Veri Analizi Problemi: Satış Verileri

Bir mağazanın satış kayıtlarını inceleyelim.

````python
satislar = pd.DataFrame({
    "Ürün": ["Laptop", "Tablet", "Laptop", "Telefon", "Tablet", "Telefon", "Laptop"],
    "Kategori": ["Bilgisayar", "Mobil", "Bilgisayar", "Mobil", "Mobil", "Mobil", "Bilgisayar"],
    "Adet": [2, 3, 1, 4, 2, 1, 3],
    "BirimFiyat": [35000, 18000, 35000, 25000, 18000, 25000, 35000]
})

satislar["Toplam"] = satislar["Adet"] * satislar["BirimFiyat"]

print(satislar)
````

# 69. Toplam Ciroyu Hesaplamak

````python
print("Toplam ciro:", satislar["Toplam"].sum())
````

# 70. Ürün Bazında Satış Analizi

````python
urun_analizi = satislar.groupby("Ürün").agg({
    "Adet": "sum",
    "Toplam": "sum"
})

print(urun_analizi)
````

# 71. En Çok Ciro Üreten Ürün

````python
urun_analizi = satislar.groupby("Ürün")["Toplam"].sum()

print(urun_analizi.idxmax())
print(urun_analizi.max())
````

# 72. Kategori Bazında Analiz

````python
kategori_analizi = satislar.groupby("Kategori")["Toplam"].sum()

print(kategori_analizi)
````

# 73. Veri Analizi Problemi: Sıcaklık Verileri

````python
sicakliklar = pd.DataFrame({
    "Gün": ["Pazartesi", "Salı", "Çarşamba", "Perşembe", "Cuma", "Cumartesi", "Pazar"],
    "Sıcaklık": [18, 21, 25, 23, 27, 29, 22],
    "Nem": [60, 55, 48, 52, 45, 43, 58]
})

print(sicakliklar)
````

# 74. Sıcaklık Analizi

````python
print("Ortalama sıcaklık:", sicakliklar["Sıcaklık"].mean())
print("En yüksek sıcaklık:", sicakliklar["Sıcaklık"].max())
print("En düşük sıcaklık:", sicakliklar["Sıcaklık"].min())

print("\n25 derece ve üzerindeki günler:")
print(sicakliklar[sicakliklar["Sıcaklık"] >= 25])
````

# 75. Veri Analizi Problemi: Öğrenci Verileri

Daha büyük bir örnek oluşturalım.

````python
ogrenci_verisi = pd.DataFrame({
    "İsim": ["Ali", "Ayşe", "Mehmet", "Zeynep", "Can", "Elif", "Mert", "Ece"],
    "Sınıf": [7, 7, 7, 8, 8, 8, 8, 7],
    "Python": [90, 95, 65, 88, 72, 98, 81, 76],
    "Matematik": [80, 90, 70, 92, 75, 95, 83, 88],
    "Fen": [85, 93, 68, 90, 79, 97, 86, 82]
})

ogrenci_verisi["Ortalama"] = ogrenci_verisi[
    ["Python", "Matematik", "Fen"]
].mean(axis=1)

print(ogrenci_verisi)
````

# 76. En Yüksek Ortalama

````python
index = ogrenci_verisi["Ortalama"].idxmax()

print(ogrenci_verisi.loc[index])
````

# 77. Sınıflara Göre Ortalama

````python
print(
    ogrenci_verisi.groupby("Sınıf")[
        ["Python", "Matematik", "Fen", "Ortalama"]
    ].mean()
)
````

# 78. 85 Üzeri Ortalamaya Sahip Öğrenciler

````python
print(
    ogrenci_verisi[
        ogrenci_verisi["Ortalama"] >= 85
    ][["İsim", "Ortalama"]]
)
````

# 79. Ders Bazında En Yüksek Puanlar

````python
print("Python:", ogrenci_verisi["Python"].max())
print("Matematik:", ogrenci_verisi["Matematik"].max())
print("Fen:", ogrenci_verisi["Fen"].max())
````

# 80. Korelasyona İlk Bakış

Sayısal sütunların birbirleriyle ilişkisini incelemek için `corr()` kullanılabilir.

````python
korelasyon = ogrenci_verisi[
    ["Python", "Matematik", "Fen", "Ortalama"]
].corr()

print(korelasyon)
````

Korelasyon konusu ileride veri analizi ve yapay zeka çalışmalarında daha ayrıntılı ele alınacaktır.

Burada amaç yalnızca Pandas'ın sayısal sütunlar arasındaki ilişkileri hesaplayabildiğini görmektir.


# 81. Pandas ile Grafik Oluşturmaya İlk Bakış

Pandas, Matplotlib'i kullanarak hızlı grafikler oluşturabilir.

````python
ogrenci_verisi.plot(
    x="İsim",
    y="Ortalama",
    kind="bar",
    title="Öğrenci Ortalamaları"
)
````

Bu grafiği bir sonraki derste Matplotlib ile çok daha ayrıntılı biçimde özelleştireceğiz.

df[(df["Puan"] >= 50) & (df["Puan"] <= 80)]

`isna().sum()` ile kontrol etmek önemlidir.

### Orijinal veriyi istemeden değiştirmek

Gerektiğinde `copy()` kullanabiliriz.

### `loc` ve `iloc` farkını karıştırmak

- `iloc` → konum
