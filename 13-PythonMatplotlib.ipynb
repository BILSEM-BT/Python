# 4. İlk Çizgi Grafiğimiz

Bir haftanın sıcaklık değerlerini çizgi grafikle gösterelim.

````python
import matplotlib.pyplot as plt

gunler = ["Pzt", "Sal", "Çar", "Per", "Cum", "Cmt", "Paz"]
sicakliklar = [18, 20, 21, 19, 23, 25, 22]

plt.plot(gunler, sicakliklar)
plt.show()
````

Temel yapı:

```python
plt.plot(x_degerleri, y_degerleri)
plt.show()
```

şeklindedir.

`plt.show()` oluşturduğumuz grafiği gösterir.


# 5. Grafiğe Başlık Eklemek

Grafiğin ne anlattığını göstermek için `title()` kullanabiliriz.

````python
plt.plot(gunler, sicakliklar)
plt.title("Haftalık Sıcaklık Değişimi")
plt.show()
````

# 6. Eksen İsimleri

X ve Y eksenlerine açıklama eklemek için:

- `xlabel()`
- `ylabel()`

kullanılır.


````python
plt.plot(gunler, sicakliklar)

plt.title("Haftalık Sıcaklık Değişimi")
plt.xlabel("Günler")
plt.ylabel("Sıcaklık")

plt.show()
````

# 7. İşaretleyiciler

Çizgi üzerindeki veri noktalarını daha belirgin hale getirmek için `marker` kullanılabilir.

````python
plt.plot(gunler, sicakliklar, marker="o")

plt.title("Haftalık Sıcaklık Değişimi")
plt.xlabel("Günler")
plt.ylabel("Sıcaklık")

plt.show()
````

İşaretleyiciler, ölçüm yapılan gerçek veri noktalarının daha kolay görülmesini sağlar.


# 8. Izgara Eklemek

Grafikte değerleri okumayı kolaylaştırmak için `grid()` kullanılabilir.

````python
plt.plot(gunler, sicakliklar, marker="o")

plt.title("Haftalık Sıcaklık Değişimi")
plt.xlabel("Günler")
plt.ylabel("Sıcaklık")
plt.grid()

plt.show()
````

# 9. Grafik Boyutunu Ayarlamak

`figure()` ile grafik boyutunu değiştirebiliriz.

````python
plt.figure(figsize=(10, 5))

plt.plot(gunler, sicakliklar, marker="o")
plt.title("Haftalık Sıcaklık Değişimi")
plt.xlabel("Günler")
plt.ylabel("Sıcaklık")
plt.grid()

plt.show()
````

`figsize=(genişlik, yükseklik)` biçiminde kullanılır.


# 10. İki Veri Serisini Aynı Grafikte Göstermek

Bir haftanın gündüz ve gece sıcaklıklarını karşılaştıralım.

````python
gunler = ["Pzt", "Sal", "Çar", "Per", "Cum", "Cmt", "Paz"]

gunduz = [18, 20, 21, 19, 23, 25, 22]
gece = [10, 11, 12, 11, 14, 15, 13]

plt.plot(gunler, gunduz, marker="o", label="Gündüz")
plt.plot(gunler, gece, marker="o", label="Gece")

plt.title("Gündüz ve Gece Sıcaklıkları")
plt.xlabel("Günler")
plt.ylabel("Sıcaklık")
plt.legend()
plt.grid()

plt.show()
````

# 11. `legend()`

# 15. Yatay Sütun Grafik

`barh()` ile yatay sütun grafik oluşturabiliriz.

````python
urunler = ["Laptop", "Tablet", "Telefon", "Kulaklık"]
satis = [45, 62, 78, 35]

plt.barh(urunler, satis)

plt.title("Ürün Satışları")
plt.xlabel("Satış Adedi")
plt.ylabel("Ürünler")

plt.show()
````

Uzun kategori isimleri olduğunda yatay sütun grafik daha okunabilir olabilir.


# 16. Sütunlara Değer Yazdırmak

Grafikte her sütunun üzerinde değer gösterebiliriz.

````python
ogrenciler = ["Ali", "Ayşe", "Mehmet", "Zeynep"]
puanlar = [85, 92, 76, 88]

plt.bar(ogrenciler, puanlar)

for i, puan in enumerate(puanlar):
    plt.text(i, puan + 1, str(puan), ha="center")

plt.title("Öğrenci Puanları")
plt.xlabel("Öğrenciler")
plt.ylabel("Puan")

plt.show()
````

Bu yöntem rapor ve sunum grafiklerinde değerlerin doğrudan görülmesini sağlar.


# 17. Pasta Grafik

Bir bütünün parçalarını göstermek için pasta grafik kullanılabilir.

````python
kategoriler = ["Python", "Matematik", "Fen", "Robotik"]
saatler = [8, 6, 4, 6]

plt.pie(saatler, labels=kategoriler, autopct="%1.1f%%")
plt.title("Haftalık Ders Dağılımı")

plt.show()
````

# 20. Histogram Nedir?

Histogram, sayısal verilerin hangi aralıklarda ne kadar yoğunlaştığını gösterir.

Örneğin bir sınıftaki 50 öğrencinin puanlarını düşünelim.

Tek tek bütün puanları görmek yerine, kaç öğrencinin:

- 0-20,
- 20-40,
- 40-60,
- 60-80,
- 80-100

aralığında olduğunu inceleyebiliriz.


````python
import numpy as np

np.random.seed(10)

puanlar = np.random.randint(0, 101, 50)

plt.hist(puanlar, bins=5)

plt.title("Öğrenci Puan Dağılımı")
plt.xlabel("Puan Aralıkları")
plt.ylabel("Öğrenci Sayısı")

plt.show()
````

# 21. `bins`

Farklı `bins` değerleri grafiğin yorumunu değiştirebilir.

# 23. Scatter Plot: Dağılım Grafiği

İki sayısal değişken arasındaki ilişkiyi incelemek için scatter plot kullanılır.

Örneğin öğrencilerin çalışma süresi ile sınav puanı arasındaki ilişkiyi inceleyelim.


````python
calisma_saati = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]
puan = [45, 50, 55, 60, 68, 70, 78, 82, 90, 95]

plt.scatter(calisma_saati, puan)

plt.title("Çalışma Süresi ve Puan")
plt.xlabel("Çalışma Saati")
plt.ylabel("Puan")

plt.show()
````

# 27. Belirli Noktaya Açıklama Eklemek

`annotate()` ile grafikte önemli bir noktaya açıklama ekleyebiliriz.

````python
aylar = ["Ocak", "Şubat", "Mart", "Nisan", "Mayıs"]
satis = [120, 145, 170, 160, 210]

plt.plot(aylar, satis, marker="o")

plt.annotate(
    "En yüksek satış",
    xy=("Mayıs", 210),
    xytext=("Nisan", 220),
    arrowprops={"arrowstyle": "->"}
)

plt.title("Aylık Satışlar")
plt.ylabel("Satış")

plt.show()
````

# 29. NumPy ile Grafik

NumPy kullanarak düzenli sayısal veriler oluşturabilir ve grafik çizebiliriz.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 10)
y = x ** 2

plt.plot(x, y, marker="o")

plt.title("x ve x² İlişkisi")
plt.xlabel("x")
plt.ylabel("x²")

plt.show()
````

# 30. `linspace()` ile Grafik Verisi Oluşturmak

````python
x = np.linspace(0, 10, 50)
y = x ** 2

plt.plot(x, y)

plt.title("x² Fonksiyonu")
plt.xlabel("x")
plt.ylabel("y")

plt.show()
````

# 31. Sinüs Grafiği

NumPy'nin matematik fonksiyonlarını Matplotlib ile birlikte kullanabiliriz.

````python
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x, y)

plt.title("Sinüs Grafiği")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid()

plt.show()
````

# 32. Kosinüs Grafiği

````python
x = np.linspace(0, 2 * np.pi, 100)
y = np.cos(x)

plt.plot(x, y)

plt.title("Kosinüs Grafiği")
plt.xlabel("x")
plt.ylabel("cos(x)")
plt.grid()

plt.show()
````

# 33. Aynı Grafikte Sinüs ve Kosinüs

````python
x = np.linspace(0, 2 * np.pi, 100)

plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")

plt.title("Sinüs ve Kosinüs")
plt.xlabel("x")
plt.ylabel("Değer")
plt.legend()
plt.grid()

plt.show()
````

# 34. Pandas ile Matplotlib

Pandas DataFrame verilerini Matplotlib ile kolayca görselleştirebiliriz.

````python
import pandas as pd

ogrenciler = pd.DataFrame({
    "İsim": ["Ali", "Ayşe", "Mehmet", "Zeynep"],
    "Python": [90, 95, 70, 88],
    "Matematik": [80, 88, 75, 92]
})

print(ogrenciler)
````

Python puanlarını sütun grafikle gösterelim.

````python
plt.bar(
    ogrenciler["İsim"],
    ogrenciler["Python"]
)

plt.title("Python Puanları")
plt.xlabel("Öğrenciler")
plt.ylabel("Puan")
plt.ylim(0, 100)

plt.show()
````

# 35. DataFrame Ortalamasını Grafikle Göstermek

````python
ogrenciler["Ortalama"] = ogrenciler[
    ["Python", "Matematik"]
].mean(axis=1)

plt.bar(
    ogrenciler["İsim"],
    ogrenciler["Ortalama"]
)

plt.title("Öğrenci Ortalamaları")
plt.xlabel("Öğrenciler")
plt.ylabel("Ortalama")
plt.ylim(0, 100)

plt.show()
````

# 36. Pandas'ın `plot()` Metodu

````python
ogrenciler.plot(
    x="İsim",
    y="Ortalama",
    kind="bar",
    title="Öğrenci Ortalamaları"
)

plt.ylabel("Ortalama")
plt.show()
````

Pandas `plot()` hızlı veri incelemelerinde çok kullanışlıdır.

Ancak daha ayrıntılı kontrol gerektiğinde doğrudan Matplotlib kullanabiliriz.

# 38. Gerçek Problem: Öğrenci Başarı Analizi

Sekiz öğrencinin üç ders puanını inceleyelim.

````python
ogrenci_verisi = pd.DataFrame({
    "İsim": ["Ali", "Ayşe", "Mehmet", "Zeynep", "Can", "Elif", "Mert", "Ece"],
    "Python": [90, 95, 65, 88, 72, 98, 81, 76],
    "Matematik": [80, 90, 70, 92, 75, 95, 83, 88],
    "Fen": [85, 93, 68, 90, 79, 97, 86, 82]
})

ogrenci_verisi["Ortalama"] = ogrenci_verisi[
    ["Python", "Matematik", "Fen"]
].mean(axis=1)

print(ogrenci_verisi)
````

Öğrenci ortalamalarını grafikle gösterelim.

````python
plt.figure(figsize=(10, 5))

plt.bar(
    ogrenci_verisi["İsim"],
    ogrenci_verisi["Ortalama"]
)

plt.title("Öğrenci Genel Ortalamaları")
plt.xlabel("Öğrenciler")
plt.ylabel("Ortalama")
plt.ylim(0, 100)

plt.show()
````

# 39. Ders Ortalamalarını Karşılaştırmak

````python
dersler = ["Python", "Matematik", "Fen"]

ortalamalar = [
    ogrenci_verisi["Python"].mean(),
    ogrenci_verisi["Matematik"].mean(),
    ogrenci_verisi["Fen"].mean()
]

plt.bar(dersler, ortalamalar)

plt.title("Ders Ortalamaları")
plt.ylabel("Ortalama")
plt.ylim(0, 100)

plt.show()
````

# 40. En Yüksek ve En Düşük Ortalamayı Yorumlamak

````python
en_yuksek_index = ogrenci_verisi["Ortalama"].idxmax()
en_dusuk_index = ogrenci_verisi["Ortalama"].idxmin()

print("En yüksek ortalama:")
print(ogrenci_verisi.loc[en_yuksek_index, ["İsim", "Ortalama"]])

print("\nEn düşük ortalama:")
print(ogrenci_verisi.loc[en_dusuk_index, ["İsim", "Ortalama"]])
````

Grafik tek başına yeterli değildir.

Grafiğin yanında sayısal analiz de yapmalıyız.

Veri analizinde güçlü yaklaşım:

**Hesapla → Görselleştir → Yorumla**

# 42. Gerçek Problem: Sıcaklık ve Nem İlişkisi

# 43. Gerçek Problem: Stok Takibi

````python
stok = pd.DataFrame({
    "Gün": ["Pzt", "Sal", "Çar", "Per", "Cum", "Cmt", "Paz"],
    "Stok": [120, 110, 95, 82, 76, 60, 48]
})

plt.plot(
    stok["Gün"],
    stok["Stok"],
    marker="o"
)

plt.title("Haftalık Stok Değişimi")
plt.xlabel("Gün")
plt.ylabel("Stok")
plt.grid()

plt.show()
````

Stok miktarının sürekli düşmesi, yeni sipariş verilmesi gerektiğini gösterebilir.

# 44. Referans Çizgisi Eklemek

Örneğin stok için kritik seviyeyi grafik üzerinde gösterebiliriz.

````python
plt.plot(
    stok["Gün"],
    stok["Stok"],
    marker="o"
)

plt.axhline(
    y=60,
    label="Kritik Stok Seviyesi"
)

plt.title("Haftalık Stok Değişimi")
plt.xlabel("Gün")
plt.ylabel("Stok")
plt.legend()
plt.grid()

plt.show()
````

`axhline()` yatay referans çizgisi eklemek için kullanılabilir.

Benzer şekilde `axvline()` dikey referans çizgisi ekler.

# 54. NumPy + Pandas + Matplotlib

Şimdi üç önemli aracımızı aynı problemde birleştirelim.

NumPy ile veri üretelim:

````python
np.random.seed(7)

puanlar = np.random.randint(50, 101, 20)

print(puanlar)
````

Pandas DataFrame oluşturalım.

````python
veri = pd.DataFrame({
    "Öğrenci": [f"Öğrenci {i}" for i in range(1, 21)],
    "Puan": puanlar
})

veri["Durum"] = np.where(
    veri["Puan"] >= 70,
    "Başarılı",
    "Geliştirilmeli"
)

print(veri)
````

Analizi yapalım.

````python
print("Ortalama:", round(veri["Puan"].mean(), 2))
print("En yüksek:", veri["Puan"].max())
print("En düşük:", veri["Puan"].min())
print()
print(veri["Durum"].value_counts())
````

Son olarak grafikle gösterelim.

````python
plt.figure(figsize=(12, 5))

plt.bar(
    veri["Öğrenci"],
    veri["Puan"]
)

plt.title("Öğrenci Puanları")
plt.xlabel("Öğrenciler")
plt.ylabel("Puan")
plt.xticks(rotation=60)
plt.ylim(0, 100)

plt.show()
````

Bu örnek veri analizi sürecimizi gösteriyor:

**NumPy → Veri üretimi**

**Pandas → Veri düzenleme ve analiz**

**Matplotlib → Görselleştirme**
