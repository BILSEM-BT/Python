# 11. Eksik Veri Kontrolü

Gerçek veri setlerinde eksik kayıt bulunabilir.

Önce kontrol edelim.


````python
print(df.isna().sum())
````

Şu an veri setimizde eksik veri yok.

Ancak veri temizliği pratiği yapmak için örnek olarak birkaç eksik değer ekleyelim.


````python
df_ornek = df.copy()

df_ornek.loc[2, "Python"] = np.nan
df_ornek.loc[7, "Proje"] = np.nan

print(df_ornek.isna().sum())
````

# 12. Eksik Veriyi Ortalama ile Doldurmak

````python
python_ortalama = df_ornek["Python"].mean()
proje_ortalama = df_ornek["Proje"].mean()

df_ornek["Python"] = df_ornek["Python"].fillna(python_ortalama)
df_ornek["Proje"] = df_ornek["Proje"].fillna(proje_ortalama)

print(df_ornek.isna().sum())
````

Burada eksik değerleri ilgili sütunun ortalamasıyla doldurduk.

Gerçek projelerde her eksik veriyi ortalamayla doldurmak doğru olmayabilir. Kullanılacak yöntem verinin anlamına göre seçilmelidir.

# 14. Hatalı veya Uç Değerleri Kontrol Etmek

Örneğin sınav puanlarının 0-100 aralığında olması gerektiğini biliyoruz.

Bu kurala uymayan kayıt var mı?


````python
puan_sutunlari = ["Python", "Matematik", "Fen", "Proje"]

for sutun in puan_sutunlari:
    hatali = df[(df[sutun] < 0) | (df[sutun] > 100)]
    print(sutun, "hatalı kayıt sayısı:", len(hatali))
````

Bu kontrol, veri temizliğinin yalnızca eksik veriden ibaret olmadığını gösterir.


# 15. Yeni Sütun: Ders Ortalaması

Python, Matematik ve Fen puanlarının ortalamasını hesaplayalım.


````python
df["DersOrtalamasi"] = df[
    ["Python", "Matematik", "Fen"]
].mean(axis=1)

df[["Isim", "Python", "Matematik", "Fen", "DersOrtalamasi"]]
````

# 16. Genel Başarı Puanı

Ders ortalaması ve proje puanını birlikte değerlendirelim.

Örnek formül:

- ders ortalaması %70
- proje puanı %30


````python
df["GenelBasari"] = (
    df["DersOrtalamasi"] * 0.70 +
    df["Proje"] * 0.30
)

df[["Isim", "DersOrtalamasi", "Proje", "GenelBasari"]]
````

# 17. Başarı Durumu Oluşturmak

Genel başarı puanına göre bir kategori oluşturalım.


````python
df["BasariDurumu"] = np.where(
    df["GenelBasari"] >= 80,
    "Başarılı",
    "Geliştirilmeli"
)

df[["Isim", "GenelBasari", "BasariDurumu"]]
````

# 18. Daha Ayrıntılı Başarı Kategorileri

Tek bir eşik yerine birden fazla kategori oluşturabiliriz.


````python
def basari_kategorisi(puan):
    if puan >= 90:
        return "Çok İyi"
    elif puan >= 80:
        return "İyi"
    elif puan >= 70:
        return "Orta"
    else:
        return "Geliştirilmeli"

df["Kategori"] = df["GenelBasari"].apply(basari_kategorisi)

df[["Isim", "GenelBasari", "Kategori"]]
````

# 51. Otomatik Özet Üreten Fonksiyon

Şimdi önceki fonksiyon bilgilerimizi kullanarak veri setinden otomatik özet çıkaralım.


````python
def analiz_ozeti(veri):
    print("Öğrenci sayısı:", len(veri))
    print("Genel ortalama:", round(veri["GenelBasari"].mean(), 2))

    en_yuksek = veri.loc[veri["GenelBasari"].idxmax()]
    en_dusuk = veri.loc[veri["GenelBasari"].idxmin()]

    print("En yüksek başarı:", en_yuksek["Isim"], "-", round(en_yuksek["GenelBasari"], 2))
    print("En düşük başarı:", en_dusuk["Isim"], "-", round(en_dusuk["GenelBasari"], 2))
    print("80 ve üzeri öğrenci:", (veri["GenelBasari"] >= 80).sum())
    print("90 ve üzeri öğrenci:", (veri["GenelBasari"] >= 90).sum())

analiz_ozeti(df)
````

# 52. Raporlama Mantığı

Bir veri analizi raporu genellikle üç tür bilgiyi ayırmalıdır.

### Gözlem

Veride doğrudan görülen bilgi.

Örnek:

```text
Genel başarı ortalaması 84.6'dır.
```

### Yorum

Veriden çıkarılan anlam.

Örnek:

```text
Öğrencilerin büyük bölümü 80 ve üzeri başarı göstermektedir.
```

### Öneri

Analiz sonucunda alınabilecek aksiyon.

Örnek:

```text
Düşük başarı grubundaki öğrenciler için ek çalışma planı hazırlanabilir.
```

Bu üç kavramı birbirinden ayırmak önemlidir.

# 58. Proje Görevi

Kendi veri analizi projenizi hazırlayın.

Bir konu seçin:

- öğrenci başarıları,
- sıcaklık ölçümleri,
- spor istatistikleri,
- kitap okuma verileri,
- günlük harcamalar,
- ürün satışları,
- stok miktarları,
- sensör ölçümleri,
- anket sonuçları.

Projenizde en az:

- 20 satırlık veri,
- 5 sütun,
- 2 yeni hesaplanan sütun,
- 3 filtreleme,
- 2 gruplama,
- 5 istatistiksel sonuç,
- 3 farklı grafik türü,
- kısa sonuç raporu

bulunsun.

Amaç sadece kod yazmak değil, **veriden anlamlı sonuç üretmektir.**
