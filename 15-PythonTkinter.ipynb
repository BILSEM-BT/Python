# 9. Butona Fonksiyon Bağlamak

Daha önce öğrendiğimiz fonksiyonlar GUI programlamada çok önemlidir.


````python
import tkinter as tk

def mesaj_yaz():
    print("Butona tıklandı.")

pencere = tk.Tk()
pencere.geometry("400x200")

buton = tk.Button(
    pencere,
    text="Tıkla",
    command=mesaj_yaz
)

buton.pack(pady=40)

pencere.mainloop()
````

Dikkat:

```python
command=mesaj_yaz
```

yazıyoruz.

Fonksiyonu burada:

```python
command=mesaj_yaz()
```

şeklinde çağırmıyoruz.

Butona tıklandığında Tkinter fonksiyonu çağırır.


# 10. Label Metnini Butonla Değiştirmek


````python
import tkinter as tk

def degistir():
    etiket.config(text="Butona tıklandı.")

pencere = tk.Tk()
pencere.geometry("400x220")

etiket = tk.Label(
    pencere,
    text="Henüz butona tıklanmadı."
)
etiket.pack(pady=30)

buton = tk.Button(
    pencere,
    text="Metni Değiştir",
    command=degistir
)
buton.pack()

pencere.mainloop()
````

`config()` ile daha önce oluşturulmuş bir widget'ın özelliklerini değiştirebiliriz.


# 11. Entry — Kullanıcıdan Veri Almak

Tek satırlık metin girişi için `Entry` kullanılır.


````python
import tkinter as tk

pencere = tk.Tk()
pencere.geometry("400x220")

etiket = tk.Label(pencere, text="Adınızı girin:")
etiket.pack(pady=10)

giris = tk.Entry(pencere)
giris.pack()

pencere.mainloop()
````

# 12. Entry'deki Veriyi Okumak

`get()` ile Entry içindeki metni okuyabiliriz.


````python
import tkinter as tk

def adi_goster():
    isim = giris.get()
    sonuc.config(text=f"Merhaba {isim}")

pencere = tk.Tk()
pencere.geometry("400x250")

tk.Label(pencere, text="Adınız:").pack(pady=10)

giris = tk.Entry(pencere)
giris.pack()

tk.Button(
    pencere,
    text="Göster",
    command=adi_goster
).pack(pady=15)

sonuc = tk.Label(pencere, text="")
sonuc.pack()

pencere.mainloop()
````

# 13. Entry'yi Temizlemek

`delete()` ile giriş kutusundaki metni silebiliriz.


````python
import tkinter as tk

def temizle():
    giris.delete(0, tk.END)

pencere = tk.Tk()
pencere.geometry("400x220")

giris = tk.Entry(pencere)
giris.pack(pady=30)

tk.Button(
    pencere,
    text="Temizle",
    command=temizle
).pack()

pencere.mainloop()
````

# 14. Basit Toplama Uygulaması

Şimdi Entry, Button, Label ve fonksiyonları birlikte kullanalım.


````python
import tkinter as tk

def topla():
    sayi1 = float(giris1.get())
    sayi2 = float(giris2.get())

    toplam = sayi1 + sayi2

    sonuc.config(text=f"Sonuç: {toplam}")

pencere = tk.Tk()
pencere.title("Toplama Uygulaması")
pencere.geometry("400x300")

tk.Label(pencere, text="1. sayı").pack()
giris1 = tk.Entry(pencere)
giris1.pack()

tk.Label(pencere, text="2. sayı").pack()
giris2 = tk.Entry(pencere)
giris2.pack()

tk.Button(
    pencere,
    text="Topla",
    command=topla
).pack(pady=20)

sonuc = tk.Label(pencere, text="Sonuç:")
sonuc.pack()

pencere.mainloop()
````

# 15. GUI Uygulamalarında Hata Yönetimi

Bir önceki örnekte kullanıcı sayı yerine metin girerse program hata verebilir.

Bunu `try-except` ile yönetelim.


````python
import tkinter as tk

def topla():
    try:
        sayi1 = float(giris1.get())
        sayi2 = float(giris2.get())

        toplam = sayi1 + sayi2

        sonuc.config(text=f"Sonuç: {toplam}")

    except ValueError:
        sonuc.config(text="Lütfen geçerli sayı girin.")

pencere = tk.Tk()
pencere.geometry("400x300")

tk.Label(pencere, text="1. sayı").pack()
giris1 = tk.Entry(pencere)
giris1.pack()

tk.Label(pencere, text="2. sayı").pack()
giris2 = tk.Entry(pencere)
giris2.pack()

tk.Button(
    pencere,
    text="Topla",
    command=topla
).pack(pady=20)

sonuc = tk.Label(pencere)
sonuc.pack()

pencere.mainloop()
````

Bu örnek, daha önce öğrendiğimiz **hata yönetimi** konusunun masaüstü uygulamalarında da önemli olduğunu gösterir.

# 18. `grid()` Yerleşim Sistemi

Formlar için çoğunlukla `grid()` daha uygundur.

Satır ve sütun mantığıyla çalışır.


````python
import tkinter as tk

pencere = tk.Tk()
pencere.geometry("400x220")

tk.Label(pencere, text="Ad:").grid(row=0, column=0, padx=10, pady=10)
tk.Entry(pencere).grid(row=0, column=1)

tk.Label(pencere, text="Soyad:").grid(row=1, column=0, padx=10, pady=10)
tk.Entry(pencere).grid(row=1, column=1)

pencere.mainloop()
````

# 19. `grid()` ile Form Oluşturmak


````python
import tkinter as tk

pencere = tk.Tk()
pencere.title("Öğrenci Formu")
pencere.geometry("450x300")

tk.Label(pencere, text="Öğrenci Adı").grid(row=0, column=0, padx=10, pady=10)
ad = tk.Entry(pencere)
ad.grid(row=0, column=1)

tk.Label(pencere, text="Sınıf").grid(row=1, column=0, padx=10, pady=10)
sinif = tk.Entry(pencere)
sinif.grid(row=1, column=1)

tk.Label(pencere, text="Python Puanı").grid(row=2, column=0, padx=10, pady=10)
puan = tk.Entry(pencere)
puan.grid(row=2, column=1)

tk.Button(pencere, text="Kaydet").grid(
    row=3,
    column=0,
    columnspan=2,
    pady=20
)

pencere.mainloop()
````

# 20. `place()` Yerleşim Sistemi

`place()` bileşenleri kesin koordinatlara yerleştirir.


````python
import tkinter as tk

pencere = tk.Tk()
pencere.geometry("400x250")

tk.Label(
    pencere,
    text="Python"
).place(x=50, y=50)

tk.Button(
    pencere,
    text="Tıkla"
).place(x=150, y=100)

pencere.mainloop()
````

# 21. Frame — Arayüzü Bölümlere Ayırmak

# 22. StringVar

Tkinter'da bazı widget değerlerini değişkenlerle bağlamak için `StringVar()` kullanılabilir.


````python
import tkinter as tk

pencere = tk.Tk()
pencere.geometry("400x220")

isim = tk.StringVar()

giris = tk.Entry(
    pencere,
    textvariable=isim
)
giris.pack(pady=20)

def goster():
    print(isim.get())

tk.Button(
    pencere,
    text="Göster",
    command=goster
).pack()

pencere.mainloop()
````

# 23. IntVar

Sayısal veya seçim değerleri için `IntVar()` kullanılabilir.


````python
import tkinter as tk

pencere = tk.Tk()
pencere.geometry("400x220")

deger = tk.IntVar(value=10)

def goster():
    print(deger.get())

tk.Button(
    pencere,
    text="Değeri Göster",
    command=goster
).pack(pady=40)

pencere.mainloop()
````

# 24. Checkbutton

Kullanıcının bir seçeneği işaretlemesi için `Checkbutton` kullanılır.


````python
import tkinter as tk

pencere = tk.Tk()
pencere.geometry("400x250")

python_secili = tk.BooleanVar()
robotik_secili = tk.BooleanVar()

tk.Checkbutton(
    pencere,
    text="Python",
    variable=python_secili
).pack()

tk.Checkbutton(
    pencere,
    text="Robotik",
    variable=robotik_secili
).pack()

def goster():
    print("Python:", python_secili.get())
    print("Robotik:", robotik_secili.get())

tk.Button(
    pencere,
    text="Seçimleri Göster",
    command=goster
).pack(pady=20)

pencere.mainloop()
````

# 25. Radiobutton

Birden fazla seçenek arasından tek seçim yapmak için `Radiobutton` kullanılabilir.


````python
import tkinter as tk

pencere = tk.Tk()
pencere.geometry("400x260")

sinif = tk.IntVar(value=7)

tk.Radiobutton(
    pencere,
    text="7. Sınıf",
    variable=sinif,
    value=7
).pack()

tk.Radiobutton(
    pencere,
    text="8. Sınıf",
    variable=sinif,
    value=8
).pack()

def goster():
    print("Seçilen sınıf:", sinif.get())

tk.Button(
    pencere,
    text="Göster",
    command=goster
).pack(pady=20)

pencere.mainloop()
````

# 26. Listbox

Bir liste içinden seçim yapmak için `Listbox` kullanılabilir.


````python
import tkinter as tk

pencere = tk.Tk()
pencere.geometry("400x300")

liste = tk.Listbox(pencere)

dersler = [
    "Python",
    "Matematik",
    "Fen",
    "Robotik"
]

for ders in dersler:
    liste.insert(tk.END, ders)

liste.pack(pady=20)

pencere.mainloop()
````

# 27. Listbox'tan Seçilen Değeri Almak


````python
import tkinter as tk

def secimi_goster():
    secim = liste.curselection()

    if secim:
        index = secim[0]
        print(liste.get(index))

pencere = tk.Tk()
pencere.geometry("400x300")

liste = tk.Listbox(pencere)

for ders in ["Python", "Matematik", "Fen", "Robotik"]:
    liste.insert(tk.END, ders)

liste.pack(pady=20)

tk.Button(
    pencere,
    text="Seçimi Göster",
    command=secimi_goster
).pack()

pencere.mainloop()
````

# 34. Kullanıcıdan Onay Almak

`askyesno()` ile evet/hayır sorusu sorabiliriz.


````python
import tkinter as tk
from tkinter import messagebox

def cikis_sor():
    cevap = messagebox.askyesno(
        "Çıkış",
        "Programdan çıkmak istiyor musunuz?"
    )

    if cevap:
        pencere.destroy()

pencere = tk.Tk()
pencere.geometry("400x220")

tk.Button(
    pencere,
    text="Çıkış",
    command=cikis_sor
).pack(pady=60)

pencere.mainloop()
````

# 37. Klavye Olayı Bağlamak

`bind()` ile klavye veya fare olayları yakalanabilir.


````python
import tkinter as tk

def enter_basildi(event):
    sonuc.config(text="Enter tuşuna basıldı.")

pencere = tk.Tk()
pencere.geometry("400x220")

sonuc = tk.Label(
    pencere,
    text="Enter tuşuna basın."
)
sonuc.pack(pady=50)

pencere.bind(
    "<Return>",
    enter_basildi
)

pencere.mainloop()
````

# 39. Basit Form Doğrulama

Kullanıcı formu boş bırakırsa kayıt yapmayalım.


````python
import tkinter as tk
from tkinter import messagebox

def kaydet():
    isim = ad.get().strip()
    puan = puan_giris.get().strip()

    if isim == "" or puan == "":
        messagebox.showwarning(
            "Eksik Bilgi",
            "Tüm alanları doldurun."
        )
        return

    try:
        puan = float(puan)

        if puan < 0 or puan > 100:
            messagebox.showwarning(
                "Geçersiz Puan",
                "Puan 0-100 arasında olmalıdır."
            )
            return

    except ValueError:
        messagebox.showerror(
            "Hata",
            "Puan sayısal olmalıdır."
        )
        return

    messagebox.showinfo(
        "Başarılı",
        "Kayıt doğrulandı."
    )

pencere = tk.Tk()
pencere.geometry("450x280")

tk.Label(pencere, text="Öğrenci Adı").grid(
    row=0, column=0, padx=10, pady=10
)
ad = tk.Entry(pencere)
ad.grid(row=0, column=1)

tk.Label(pencere, text="Puan").grid(
    row=1, column=0, padx=10, pady=10
)
puan_giris = tk.Entry(pencere)
puan_giris.grid(row=1, column=1)

tk.Button(
    pencere,
    text="Kaydet",
    command=kaydet
).grid(
    row=2,
    column=0,
    columnspan=2,
    pady=20
)

pencere.mainloop()
````

# 40. Tkinter + Dosya İşlemleri

Önceki derslerde öğrendiğimiz dosya işlemlerini GUI uygulamasına ekleyebiliriz.


````python
import tkinter as tk
from tkinter import messagebox

def kaydet():
    isim = ad.get().strip()

    if isim == "":
        messagebox.showwarning(
            "Eksik Bilgi",
            "Öğrenci adını girin."
        )
        return

    with open(
        "ogrenciler.txt",
        "a",
        encoding="utf-8"
    ) as dosya:
        dosya.write(isim + "\n")

    messagebox.showinfo(
        "Başarılı",
        "Öğrenci kaydedildi."
    )

    ad.delete(0, tk.END)

pencere = tk.Tk()
pencere.geometry("400x220")

tk.Label(
    pencere,
    text="Öğrenci Adı"
).pack(pady=10)

ad = tk.Entry(pencere)
ad.pack()

tk.Button(
    pencere,
    text="Dosyaya Kaydet",
    command=kaydet
).pack(pady=20)

pencere.mainloop()
````

# 41. Tkinter + JSON

Öğrenci kayıtlarını JSON dosyasında saklayabiliriz.


````python
import tkinter as tk
from tkinter import messagebox
import json
import os

DOSYA = "ogrenciler.json"

def verileri_oku():
    if not os.path.exists(DOSYA):
        return []

    try:
        with open(
            DOSYA,
            "r",
            encoding="utf-8"
        ) as dosya:
            return json.load(dosya)

    except json.JSONDecodeError:
        return []

def kaydet():
    isim = ad.get().strip()
    puan_text = puan_giris.get().strip()

    if isim == "" or puan_text == "":
        messagebox.showwarning(
            "Eksik Bilgi",
            "Tüm alanları doldurun."
        )
        return

    try:
        puan = float(puan_text)
    except ValueError:
        messagebox.showerror(
            "Hata",
            "Puan sayısal olmalıdır."
        )
        return

    ogrenciler = verileri_oku()

    ogrenciler.append({
        "isim": isim,
        "puan": puan
    })

    with open(
        DOSYA,
        "w",
        encoding="utf-8"
    ) as dosya:
        json.dump(
            ogrenciler,
            dosya,
            ensure_ascii=False,
            indent=4
        )

    messagebox.showinfo(
        "Başarılı",
        "Kayıt tamamlandı."
    )

    ad.delete(0, tk.END)
    puan_giris.delete(0, tk.END)

pencere = tk.Tk()
pencere.title("Öğrenci Kayıt Sistemi")
pencere.geometry("450x300")

tk.Label(pencere, text="İsim").grid(
    row=0, column=0, padx=10, pady=10
)
ad = tk.Entry(pencere)
ad.grid(row=0, column=1)

tk.Label(pencere, text="Puan").grid(
    row=1, column=0, padx=10, pady=10
)
puan_giris = tk.Entry(pencere)
puan_giris.grid(row=1, column=1)

tk.Button(
    pencere,
    text="Kaydet",
    command=kaydet
).grid(
    row=2,
    column=0,
    columnspan=2,
    pady=20
)

pencere.mainloop()
````

# 42. Mini Uygulama: Dört İşlem Hesap Makinesi


````python
import tkinter as tk
from tkinter import messagebox

def hesapla(islem):
    try:
        a = float(giris1.get())
        b = float(giris2.get())

        if islem == "+":
            sonuc_degeri = a + b
        elif islem == "-":
            sonuc_degeri = a - b
        elif islem == "*":
            sonuc_degeri = a * b
        elif islem == "/":
            if b == 0:
                messagebox.showerror(
                    "Hata",
                    "Sıfıra bölme yapılamaz."
                )
                return

            sonuc_degeri = a / b

        sonuc.config(
            text=f"Sonuç: {sonuc_degeri}"
        )

    except ValueError:
        messagebox.showerror(
            "Hata",
            "Geçerli sayılar girin."
        )

pencere = tk.Tk()
pencere.title("Hesap Makinesi")
pencere.geometry("400x330")

tk.Label(pencere, text="1. sayı").pack()
giris1 = tk.Entry(pencere)
giris1.pack()

tk.Label(pencere, text="2. sayı").pack()
giris2 = tk.Entry(pencere)
giris2.pack()

buton_frame = tk.Frame(pencere)
buton_frame.pack(pady=20)

tk.Button(
    buton_frame,
    text="+",
    width=5,
    command=lambda: hesapla("+")
).grid(row=0, column=0, padx=5)

tk.Button(
    buton_frame,
    text="-",
    width=5,
    command=lambda: hesapla("-")
).grid(row=0, column=1, padx=5)

tk.Button(
    buton_frame,
    text="*",
    width=5,
    command=lambda: hesapla("*")
).grid(row=0, column=2, padx=5)

tk.Button(
    buton_frame,
    text="/",
    width=5,
    command=lambda: hesapla("/")
).grid(row=0, column=3, padx=5)

sonuc = tk.Label(
    pencere,
    text="Sonuç:",
    font=("Arial", 14)
)
sonuc.pack()

pencere.mainloop()
````

# 44. Mini Uygulama: Not Hesaplama


````python
import tkinter as tk
from tkinter import messagebox

def hesapla():
    try:
        vize = float(vize_giris.get())
        final = float(final_giris.get())

        ortalama = vize * 0.40 + final * 0.60

        if ortalama >= 50:
            durum = "Başarılı"
        else:
            durum = "Başarısız"

        sonuc.config(
            text=f"Ortalama: {ortalama:.2f}\nDurum: {durum}"
        )

    except ValueError:
        messagebox.showerror(
            "Hata",
            "Puanları sayısal girin."
        )

pencere = tk.Tk()
pencere.title("Not Hesaplama")
pencere.geometry("400x300")

tk.Label(pencere, text="Vize").pack()
vize_giris = tk.Entry(pencere)
vize_giris.pack()

tk.Label(pencere, text="Final").pack()
final_giris = tk.Entry(pencere)
final_giris.pack()

tk.Button(
    pencere,
    text="Hesapla",
    command=hesapla
).pack(pady=20)

sonuc = tk.Label(pencere, text="")
sonuc.pack()

pencere.mainloop()
````

# 45. Mini Uygulama: Günlük Not Defteri


````python
import tkinter as tk
from tkinter import messagebox
from datetime import datetime

DOSYA = "gunluk.txt"

def kaydet():
    icerik = metin.get("1.0", tk.END).strip()

    if icerik == "":
        messagebox.showwarning(
            "Uyarı",
            "Kaydedilecek metin yok."
        )
        return

    tarih = datetime.now().strftime(
        "%d.%m.%Y %H:%M"
    )

    with open(
        DOSYA,
        "a",
        encoding="utf-8"
    ) as dosya:
        dosya.write(
            f"[{tarih}]\n{icerik}\n\n"
        )

    messagebox.showinfo(
        "Başarılı",
        "Not kaydedildi."
    )

    metin.delete("1.0", tk.END)

pencere = tk.Tk()
pencere.title("Günlük Not Defteri")
pencere.geometry("550x400")

metin = tk.Text(
    pencere,
    width=55,
    height=15
)
metin.pack(pady=20)

tk.Button(
    pencere,
    text="Kaydet",
    command=kaydet
).pack()

pencere.mainloop()
````

# 46. Mini Uygulama: Öğrenci Listesi


````python
import tkinter as tk

def ekle():
    isim = giris.get().strip()

    if isim:
        liste.insert(tk.END, isim)
        giris.delete(0, tk.END)

def sil():
    secim = liste.curselection()

    if secim:
        liste.delete(secim[0])

pencere = tk.Tk()
pencere.title("Öğrenci Listesi")
pencere.geometry("400x400")

giris = tk.Entry(pencere)
giris.pack(pady=10)

tk.Button(
    pencere,
    text="Ekle",
    command=ekle
).pack()

liste = tk.Listbox(
    pencere,
    width=30,
    height=12
)
liste.pack(pady=20)

tk.Button(
    pencere,
    text="Seçileni Sil",
    command=sil
).pack()

pencere.mainloop()
````
