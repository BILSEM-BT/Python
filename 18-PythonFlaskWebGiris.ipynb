# 19. `render_template()`

Python tarafında HTML dosyasını göstermek için:


````python
from flask import Flask, render_template

app = Flask(__name__)

@app.route("/")
def ana_sayfa():
    return render_template("index.html")
````

# 38. GET ve POST Kullanan Route

Bir route hem formu gösterebilir hem form gönderildiğinde veriyi işleyebilir.


````python
from flask import Flask, render_template, request

app = Flask(__name__)

@app.route("/kayit", methods=["GET", "POST"])
def kayit():
    if request.method == "POST":
        isim = request.form.get("isim")

        return f"Gönderilen isim: {isim}"

    return render_template("kayit.html")
````

Burada:

### GET
Kullanıcı `/kayit` sayfasını açar.

### POST
Kullanıcı formu gönderir.

Aynı route iki farklı HTTP davranışını yönetebilir.

# 40. Sunucu Tarafında Form Doğrulama

Tarayıcıdaki kontroller tek başına yeterli değildir.

Flask tarafında:


````python
@app.route("/kayit", methods=["GET", "POST"])
def kayit():
    mesaj = None

    if request.method == "POST":
        isim = request.form.get("isim", "").strip()

        if not isim:
            mesaj = "Öğrenci adı boş bırakılamaz."
        else:
            mesaj = f"{isim} kayda hazır."

    return render_template(
        "kayit.html",
        mesaj=mesaj
    )
````

HTML:

```html
    <p>{{ mesaj }}</p>
{% endif %}
```


# 41. Sayısal Form Verisini Kontrol Etmek

HTML'den gelen form değerleri metin olarak gelir.

Puanı sayıya dönüştürürken hata yönetimi yapabiliriz.


````python
def puan_kontrol(puan_text):
    try:
        puan = int(puan_text)

        if puan < 0 or puan > 100:
            return None

        return puan

    except (TypeError, ValueError):
        return None
````

Bu fonksiyonu Flask route'unda kullanabiliriz.


# 42. Tam Form Doğrulama Örneği

````python
@app.route("/ogrenci-ekle", methods=["GET", "POST"])
def ogrenci_ekle():
    mesaj = None

    if request.method == "POST":
        isim = request.form.get("isim", "").strip()
        puan_text = request.form.get("puan", "").strip()

        if not isim:
            mesaj = "İsim boş bırakılamaz."

        else:
            puan = puan_kontrol(puan_text)

            if puan is None:
                mesaj = "Puan 0-100 arasında bir tam sayı olmalıdır."
            else:
                mesaj = f"{isim} - {puan} bilgisi doğrulandı."

    return render_template(
        "ogrenci_ekle.html",
        mesaj=mesaj
    )
````

# 43. POST Sonrası Neden Redirect Kullanılır?

Form başarıyla işlendiğinde aynı POST isteğine doğrudan sayfa döndürmek yerine başka route'a yönlendirme yapmak yaygın bir yaklaşımdır.

# 44. `redirect()` ve `url_for()`

````python
from flask import redirect, url_for

@app.route("/basarili")
def basarili():
    return "Kayıt tamamlandı."

@app.route("/ornek-kayit", methods=["POST"])
def ornek_kayit():
    # Kayıt işlemleri burada yapılır.

    return redirect(
        url_for("basarili")
    )
````

# 45. Query String

Buradaki `q` değerini:


````python
@app.route("/ara")
def ara():
    kelime = request.args.get("q", "")

    return f"Aranan kelime: {kelime}"
````

Bu yapı bir sonraki CRUD dersimizde kayıt arama özelliği için kullanılabilir.

# 47. Form Verisini Doğrudan HTML'e Yazarken Güvenlik

Kullanıcıdan gelen veriler güvenilir kabul edilmemelidir.

```html

```

şeklinde normal Jinja değişkeni kullanmak, HTML template'lerinde otomatik escaping mekanizmasından yararlanır.

Bu nedenle kullanıcı girdisini HTML koduyla string birleştirme yoluyla üretmek yerine template sistemini kullanmak daha doğru bir yaklaşımdır.


# 48. `safe` Filtresine Dikkat

Jinja'da:

```html
{{ veri | safe }}
```

kullanımı escaping mekanizmasını devre dışı bırakabilir.

Kullanıcıdan gelen güvenilmeyen verilerde gelişigüzel kullanılmamalıdır.

Bu derste kullanıcı verilerinde normal:

```html
{{ veri }}
```

kullanımını tercih edeceğiz.


# 49. Flash Mesajları

Web uygulamalarında kullanıcıya:

- kayıt başarılı,
- hata oluştu,
- kayıt silindi

gibi geçici mesajlar göstermek isteyebiliriz.

Flask'ta bunun için `flash()` kullanılabilir.

Flash mesajları için uygulamada bir `secret_key` tanımlanması gerekir.


# 50. Basit Flash Örneği

`app.py`:


````python
from flask import Flask, flash, redirect, url_for

app = Flask(__name__)
app.secret_key = "gelistirme-icin-ornek-anahtar"

@app.route("/mesaj")
def mesaj():
    flash("İşlem başarıyla tamamlandı.")

    return redirect(
        url_for("ana_sayfa")
    )
````

Gerçek projelerde secret key kod içine sabit ve tahmin edilebilir şekilde yazılmamalıdır. Bu örnek yalnızca mekanizmayı göstermek içindir.


# 51. Flash Mesajını Template'te Göstermek

`base.html` içinde:

```html
{% with mesajlar = get_flashed_messages() %}

    {% if mesajlar %}

        {% for mesaj in mesajlar %}
            <p>{{ mesaj }}</p>
        {% endfor %}

    {% endif %}

{% endwith %}
```

Böylece ortak template üzerinden mesajlar bütün sayfalarda gösterilebilir.

# 56. Mini Proje: Öğrenci Ekleme Formu

Şimdilik veritabanına yazmadan listeye ekleyelim.

Bu yalnızca form ve POST mantığını anlamak içindir.


````python
from flask import request, redirect, url_for

@app.route("/ogrenci-ekle", methods=["GET", "POST"])
def ogrenci_ekle():
    if request.method == "POST":
        isim = request.form.get("isim", "").strip()
        puan_text = request.form.get("puan", "").strip()

        puan = puan_kontrol(puan_text)

        if isim and puan is not None:
            yeni_id = len(ogrenciler) + 1

            ogrenciler.append({
                "id": yeni_id,
                "isim": isim,
                "puan": puan
            })

            return redirect(
                url_for("ogrenci_listesi")
            )

    return render_template(
        "ogrenci_ekle.html"
    )
````

`templates/ogrenci_ekle.html`:

```html
            name="isim"
            required
        >
    </p>

    <p>
        <label>Puan</label>
        <input
            type="number"
            name="puan"
            min="0"
            max="100"
            required
        >
    </p>

    </button>

{% endblock %}
```
