# 21. Eğitim ve Test Verisi Neden Ayrılır?

Modeli bütün veriyle eğitirsek, aynı veriler üzerinde başarı ölçmek yanıltıcı olabilir.

Bu nedenle veri kümesini iki parçaya ayırırız:

### Eğitim Verisi

Modelin öğrenmesi için kullanılır.

### Test Verisi

Modelin daha önce görmediği örnekler üzerinde ne kadar başarılı olduğunu ölçmek için kullanılır.

Temel akış:

```text
Tüm Veri
│
├── Eğitim Verisi → Model Öğrenir
│
└── Test Verisi → Model Değerlendirilir
```


# 22. `train_test_split`

Veriyi eğitim ve test gruplarına ayırmak için:


````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Eğitim özellikleri:", X_train.shape)
print("Test özellikleri:", X_test.shape)
````

Burada:

- `test_size=0.20` → verinin %20'sini test için ayırır,
- `random_state=42` → bölmenin tekrar üretilebilir olmasına yardımcı olur,
- `stratify=y` → sınıf oranlarının eğitim ve test bölümlerinde korunmasına yardımcı olur.


# 23. Eğitim Sınıf Dağılımı

````python
print(y_train.value_counts().sort_index())
````

# 24. Test Sınıf Dağılımı

````python
print(y_test.value_counts().sort_index())
````

# 25. İlk Algoritmamız: K-En Yakın Komşu

K-Nearest Neighbors veya KNN, yeni bir örneği çevresindeki benzer eğitim örneklerine bakarak sınıflandıran bir yöntemdir.

Basitleştirilmiş düşünce:

```text
Yeni Örnek
↓
En Yakın K Komşuyu Bul
↓
Komşuların Sınıflarına Bak
↓
Tahmin Üret
```


# 26. K Değeri Nedir?

`K`, kaç komşunun dikkate alınacağını belirtir.

Örneğin:

```text
K = 3
```

ise model yeni örneğe en yakın üç eğitim örneğini inceler.

K değeri model davranışını etkileyen bir **hiperparametre**dir.


# 27. Ölçeklendirme Neden Önemlidir?

KNN uzaklık hesabı kullandığı için farklı ölçeklerdeki özellikler sonucu etkileyebilir.

Bu nedenle özellikleri ölçeklendirmek yaygın bir yaklaşımdır.

Scikit-learn'de bunun için `StandardScaler` kullanabiliriz.


````python
from sklearn.preprocessing import StandardScaler
````

# 28. Pipeline Nedir?

Ön işleme ve modeli tek bir zincir halinde birleştirebiliriz.

Örneğin:

```text
Veri
↓
StandardScaler
↓
KNN
↓
Tahmin
```

Scikit-learn `Pipeline` yapısı bu işlemleri birlikte yönetmemizi sağlar.


````python
from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsClassifier
````

# 29. İlk Modelimizi Oluşturmak

````python
model = Pipeline([
    ("scaler", StandardScaler()),
    ("knn", KNeighborsClassifier(n_neighbors=5))
])

print(model)
````

Bu noktada model henüz öğrenmedi.

Sadece kullanılacak işlem zincirini oluşturduk.


# 30. `fit()`: Modeli Eğitmek

Modelin eğitim verisinden öğrenmesini sağlamak için:


````python
model.fit(
    X_train,
    y_train
)

print("Model eğitildi.")
````

Makine öğrenmesinde çok sık göreceğimiz temel yapı:

```python
model.fit(X_train, y_train)
```

olacaktır.


# 31. `predict()`: Tahmin Üretmek

Test verisi için tahmin yapalım.


````python
y_pred = model.predict(X_test)

print(y_pred)
````

Model her test örneği için:

```text
0
1
2
```

sınıflarından birini tahmin etti.


# 32. Gerçek ve Tahmin Edilen Değerleri Karşılaştırmak

````python
karsilastirma = pd.DataFrame({
    "Gercek": y_test.to_numpy(),
    "Tahmin": y_pred
})

karsilastirma.head(10)
````

# 33. Sınıf İsimlerini de Eklemek

````python
karsilastirma["GercekTur"] = [
    iris.target_names[i]
    for i in karsilastirma["Gercek"]
]

karsilastirma["TahminTur"] = [
    iris.target_names[i]
    for i in karsilastirma["Tahmin"]
]

karsilastirma.head(10)
````

# 34. Model Başarısı: Accuracy

Sınıflandırmada ilk öğreneceğimiz ölçülerden biri accuracy'dir.

Accuracy:

```text
Doğru Tahmin Sayısı
-------------------
Toplam Tahmin Sayısı
```

mantığıyla düşünülebilir.


````python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("Accuracy:", accuracy)
````

Yüzde biçiminde gösterebiliriz.


````python
print(
    f"Doğruluk oranı: %{accuracy * 100:.2f}"
)
````

# 35. Modelin Kendi `score()` Metodu

Sınıflandırıcılar test verisi üzerinde doğrudan skor da hesaplayabilir.


````python
print(
    "Model skoru:",
    model.score(X_test, y_test)
)
````

# 36. Confusion Matrix

Sadece toplam doğruluk oranına bakmak her zaman yeterli değildir.

Hangi sınıfların hangi sınıflarla karıştırıldığını görmek için confusion matrix kullanabiliriz.


````python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    y_test,
    y_pred
)

print(cm)
````

Satırlar gerçek sınıfları, sütunlar tahmin edilen sınıfları temsil eder.

Ana köşegen üzerindeki değerler doğru tahminleri gösterir.


# 37. Confusion Matrix'i Görselleştirmek

````python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=iris.target_names
).plot()

plt.title("Confusion Matrix")
plt.show()
````

# 38. Classification Report

Precision, recall ve F1-score gibi sınıflandırma ölçülerini tek tabloda görebiliriz.


````python
from sklearn.metrics import classification_report

print(
    classification_report(
        y_test,
        y_pred,
        target_names=iris.target_names
    )
)
````

Bu kavramları ilerleyen derslerde ayrıntılı inceleyeceğiz.

İlk aşamada önemli olan şudur:

**Tek bir accuracy değeri modelin bütün davranışını anlatmayabilir.**

# 41. Tahmin Olasılıkları

KNN sınıflandırıcımız sınıflar için tahmin olasılıkları da üretebilir.


````python
olasiliklar = model.predict_proba(yeni_cicek)[0]

for tur, olasilik in zip(
    iris.target_names,
    olasiliklar
):
    print(
        tur,
        "->",
        round(olasilik, 3)
    )
````

Olasılık değeri, modelin belirli bir örnek için sınıflara nasıl ağırlık verdiğini incelememize yardımcı olur.

# 42. Tahmin Fonksiyonu Yazmak

Model kullanımını bir Python fonksiyonuna dönüştürelim.


````python
def iris_tahmin(
    sepal_length,
    sepal_width,
    petal_length,
    petal_width
):
    veri = pd.DataFrame(
        [[
            sepal_length,
            sepal_width,
            petal_length,
            petal_width
        ]],
        columns=iris.feature_names
    )

    sinif = model.predict(veri)[0]

    return iris.target_names[sinif]
````

# 47. Farklı K Değerlerini Denemek

KNN modelinde `n_neighbors` değerini değiştirdiğimizde sonuç değişebilir.

Farklı değerleri karşılaştıralım.


````python
sonuclar = []

for k in range(1, 16):
    deneme_modeli = Pipeline([
        ("scaler", StandardScaler()),
        ("knn", KNeighborsClassifier(
            n_neighbors=k
        ))
    ])

    deneme_modeli.fit(
        X_train,
        y_train
    )

    skor = deneme_modeli.score(
        X_test,
        y_test
    )

    sonuclar.append({
        "K": k,
        "Accuracy": skor
    })

sonuc_df = pd.DataFrame(sonuclar)

sonuc_df
````

# 48. K Değerlerine Göre Başarı Grafiği

````python
plt.plot(
    sonuc_df["K"],
    sonuc_df["Accuracy"],
    marker="o"
)

plt.xlabel("K")
plt.ylabel("Accuracy")
plt.title("K Değerine Göre Model Başarısı")
plt.grid()
plt.show()
````

Bu deney bize **hiperparametre** kavramını gösterir.

Modelin eğitim verisinden öğrendiği değerlerle, bizim eğitim öncesi belirlediğimiz ayarlar aynı şey değildir.


# 49. En Başarılı K Değeri

Bu tek test bölmesi için en yüksek skoru veren değeri inceleyelim.


````python
en_iyi_satir = sonuc_df.loc[
    sonuc_df["Accuracy"].idxmax()
]

print(en_iyi_satir)
````

Burada önemli bir uyarı vardır:

Test verisine tekrar tekrar bakarak hiperparametre seçmek doğru değerlendirmeyi bozabilir.

İleride:

- validation,
- cross-validation,
- GridSearchCV

kullanarak model seçimini daha doğru biçimde yapacağız.


# 50. Başka Bir Model: Decision Tree

Aynı veri üzerinde farklı bir algoritma da deneyebiliriz.

Bu bölümün amacı algoritmaları ayrıntılı öğrenmek değil, aynı problemi farklı modellerin çözebildiğini görmektir.


````python
from sklearn.tree import DecisionTreeClassifier

agac_modeli = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)

agac_modeli.fit(
    X_train,
    y_train
)

agac_tahmin = agac_modeli.predict(
    X_test
)

print(
    "Decision Tree Accuracy:",
    accuracy_score(
        y_test,
        agac_tahmin
    )
)
````

# 51. Modelleri Karşılaştırmak

KNN ve Decision Tree sonuçlarını aynı tabloda görelim.


````python
model_karsilastirma = pd.DataFrame({
    "Model": [
        "KNN",
        "Decision Tree"
    ],
    "Accuracy": [
        accuracy_score(
            y_test,
            y_pred
        ),
        accuracy_score(
            y_test,
            agac_tahmin
        )
    ]
})

model_karsilastirma
````

# 55. Overfitting Kavramına Giriş

Model eğitim verisine aşırı uyum sağlarsa:

- eğitim başarısı çok yüksek,
- yeni veride başarı düşük

İlerleyen derslerde eğitim ve doğrulama sonuçlarını birlikte inceleyerek overfitting konusunu ayrıntılı ele alacağız.


# 56. Underfitting Kavramına Giriş

Model problem için fazla basit kalırsa:

- eğitim verisini bile iyi öğrenemez,
- test verisinde de zayıf olur.

Bu duruma **underfitting** denir.

# 66. Mini Uygulama: Tek Örnek Tahmin Raporu

Yeni örnek için sınıf ve olasılıkları birlikte döndüren fonksiyon yazalım.


````python
def iris_tahmin_raporu(degerler):
    veri = pd.DataFrame(
        [degerler],
        columns=iris.feature_names
    )

    tahmin = model.predict(veri)[0]
    olasilik = model.predict_proba(veri)[0]

    sonuc = {
        "tahmin": iris.target_names[tahmin],
        "olasiliklar": {}
    }

    for tur, deger in zip(
        iris.target_names,
        olasilik
    ):
        sonuc["olasiliklar"][tur] = round(
            float(deger),
            3
        )

    return sonuc
````

````python
rapor = iris_tahmin_raporu(
    [6.3, 3.3, 6.0, 2.5]
)

print(rapor)
````

Bu fonksiyon ileride JSON cevap döndüren bir Flask API'sine dönüştürülebilir.
