# 9. `re` Modülü ile Basit Temizleme

````python
import re

metin = "Python, yapay zeka için çok güçlü! Değil mi?"

temiz = re.sub(
    r"[^a-zA-ZçğıöşüÇĞİÖŞÜ0-9\s]",
    "",
    metin
)

print(
    temiz
)
````

Bu düzenli ifade Türkçe harfleri, sayıları ve boşlukları koruyup diğer karakterleri kaldırmaktadır.


# 10. Temizleme Fonksiyonu Yazmak

````python
def metin_temizle(metin):
    metin = metin.lower()

    metin = re.sub(
        r"[^a-zçğıöşü0-9\s]",
        " ",
        metin
    )

    metin = " ".join(
        metin.split()
    )

    return metin
````

````python
print(
    metin_temizle(
        "Python, Yapay Zeka için ÇOK güzel!"
    )
)
````

# 16. Kelime Frekansı

Bir kelimenin metinde kaç kez geçtiğini hesaplayabiliriz.


````python
metin = "python veri python yapay zeka veri python"

kelimeler = metin.split()

frekans = {}

for kelime in kelimeler:
    frekans[kelime] = (
        frekans.get(
            kelime,
            0
        )
        + 1
    )

print(
    frekans
)
````

# 17. `Counter` ile Kelime Frekansı

# 18. En Sık Geçen Kelimeler

# 20. Stop Word Nedir?

Metinlerde çok sık geçen ancak bazı görevlerde ayırt edici bilgi taşımayan kelimelere stop word denir.

Türkçe örnekler:

```text
ve
ile
bir
bu
için
çok
```

Ancak stop word listesi problemden probleme değişebilir.

Örneğin duygu analizinde:

```text
değil
```

kelimesini kaldırmak anlamı tamamen değiştirebilir.

Bu nedenle stop word temizleme otomatik ve düşünmeden yapılmamalıdır.


# 21. Basit Türkçe Stop Word Listesi

````python
turkce_stop_words = {
    "ve",
    "ile",
    "bir",
    "bu",
    "için",
    "çok",
    "olarak",
    "da",
    "de"
}

print(
    turkce_stop_words
)
````

# 22. Stop Word Temizlemek

````python
metin = "python ile veri analizi ve yapay zeka için çok kullanışlı"

kelimeler = (
    metin_temizle(
        metin
    )
    .split()
)

filtreli = [
    kelime
    for kelime in kelimeler
    if kelime not in turkce_stop_words
]

print(
    filtreli
)
````

# 23. Stop Word Temizleme Fonksiyonu

````python
def stop_word_temizle(
    tokenlar,
    stop_words
):
    return [
        token
        for token in tokenlar
        if token not in stop_words
    ]
````

````python
print(
    stop_word_temizle(
        kelimeler,
        turkce_stop_words
    )
)
````

# 25. Bag of Words Nedir?

Bag of Words yaklaşımında:

1. bütün belgelerde geçen kelimelerden bir sözlük oluşturulur,
2. her belge için kelimelerin kaç kez geçtiği sayılır.

Örneğin iki metin:

```text

```

Sözlük:

```text
python

veri
```

Sayısal temsil:

```text
Metin 1 → [1, 1, 1, 0]
Metin 2 → [1, 0, 0, 1]
```

# 48. TF-IDF Nedir?

Bag of Words yalnızca kelime sayısını kullanır.

TF-IDF ise bir kelimenin:

- belirli belgede ne kadar önemli olduğu,
- bütün belgelerde ne kadar yaygın olduğu

Açılımı:

```text
Term Frequency
×
Inverse Document Frequency
```


# 49. TF — Term Frequency

Örneğin:


# 50. IDF — Inverse Document Frequency

Örneğin her belgede:

```text

```

gibi kelimeler varsa bunların ayırt ediciliği az olabilir.

IDF çok yaygın kelimelerin ağırlığını azaltmaya yardımcı olur.

# 56. Metin Benzerliği

Metinleri sayısal vektörlere dönüştürdüğümüzde iki metnin vektörleri arasındaki benzerliği hesaplayabiliriz.

Yaygın yöntemlerden biri **cosine similarity**dir.


# 57. Cosine Similarity Nedir?

İki vektörün yönlerinin ne kadar benzer olduğunu ölçer.

Genel olarak:

```text
1'e yakın → daha benzer
0'a yakın → daha farklı
```

olarak yorumlanabilir.

Metin benzerliğinde özellikle TF-IDF vektörleriyle sık kullanılır.


# 58. Benzerlik Örneği

````python
benzerlik_metinleri = [
    "python ile yapay zeka uygulaması geliştirme",
    "python ile makine öğrenmesi projesi geliştirme",
    "bahçede domates yetiştirme yöntemleri"
]

benzerlik_vectorizer = (
    TfidfVectorizer()
)

benzerlik_matrix = (
    benzerlik_vectorizer.fit_transform(
        benzerlik_metinleri
    )
)

print(
    benzerlik_matrix.shape
)
````

# 59. Cosine Similarity Hesaplamak

````python
from sklearn.metrics.pairwise import (
    cosine_similarity
)

benzerlikler = cosine_similarity(
    benzerlik_matrix
)

print(
    benzerlikler
)
````

# 60. Benzerlik Matrisi DataFrame

````python
benzerlik_df = pd.DataFrame(
    benzerlikler,
    index=[
        "Metin 1",
        "Metin 2",
        "Metin 3"
    ],
    columns=[
        "Metin 1",
        "Metin 2",
        "Metin 3"
    ]
)

benzerlik_df.round(3)
````

# 61. En Benzer Metni Bulmak

Bir sorgu metnini dokümanlarla karşılaştıralım.


````python
dokumanlar = [
    "python ile veri analizi",
    "makine öğrenmesi model geliştirme",
    "robotik sensör uygulaması",
    "doğal dil işleme ile metin analizi"
]

sorgu = "python metin analizi"

tum = dokumanlar + [
    sorgu
]

arama_vectorizer = TfidfVectorizer()

arama_matrix = (
    arama_vectorizer.fit_transform(
        tum
    )
)

sorgu_vector = (
    arama_matrix[-1]
)

dokuman_matrix = (
    arama_matrix[:-1]
)

skorlar = cosine_similarity(
    sorgu_vector,
    dokuman_matrix
)[0]

print(
    skorlar
)
````

# 62. En Benzer Doküman

````python
en_iyi_index = int(
    np.argmax(
        skorlar
    )
)

print(
    "En benzer doküman:",
    dokumanlar[
        en_iyi_index
    ]
)

print(
    "Skor:",
    round(
        float(
            skorlar[
                en_iyi_index
            ]
        ),
        3
    )
)
````

Bu yaklaşım küçük bir metin arama sisteminin temelidir.


# 63. Basit Arama Fonksiyonu

````python
def metin_ara(
    dokumanlar,
    sorgu
):
    tum = dokumanlar + [
        sorgu
    ]

    vectorizer = TfidfVectorizer()

    matris = (
        vectorizer.fit_transform(
            tum
        )
    )

    skorlar = cosine_similarity(
        matris[-1],
        matris[:-1]
    )[0]

    siralama = np.argsort(
        skorlar
    )[::-1]

    sonuc = []

    for index in siralama:
        sonuc.append({
            "metin":
                dokumanlar[
                    index
                ],
            "skor":
                round(
                    float(
                        skorlar[
                            index
                        ]
                    ),
                    3
                )
        })

    return sonuc
````

````python
arama_sonucu = metin_ara(
    dokumanlar,
    "makine öğrenmesi python"
)

for sonuc in arama_sonucu:
    print(
        sonuc
    )
````

````python
analiz_df[
    "KelimeSayisi"
] = analiz_df[
    "Metin"
].apply(
    lambda x: len(
        metin_temizle(
            x
        ).split()
    )
)

analiz_df
````

# 68. Benzersiz Kelime Sayısı

````python
analiz_df[
    "BenzersizKelime"
] = analiz_df[
    "Metin"
].apply(
    lambda x: len(
        set(
            metin_temizle(
                x
            ).split()
        )
    )
)

analiz_df
````

# 69. Ortalama Kelime Uzunluğu

````python
def ortalama_kelime_uzunlugu(
    metin
):
    kelimeler = (
        metin_temizle(
            metin
        )
        .split()
    )

    if not kelimeler:
        return 0

    return sum(
        len(kelime)
        for kelime in kelimeler
    ) / len(kelimeler)
````

````python
analiz_df[
    "OrtKelimeUzunlugu"
] = analiz_df[
    "Metin"
].apply(
    ortalama_kelime_uzunlugu
)

analiz_df
````

# 71. Küçük Etiketli Metin Veri Kümesi

````python
etiketli_metinler = pd.DataFrame({
    "Metin": [
        "Python dersi çok eğlenceli",
        "Bu uygulamayı çok beğendim",
        "Proje başarılı şekilde tamamlandı",
        "Bu sonuç hiç iyi değil",
        "Uygulama sürekli hata veriyor",
        "Bu deneyimden memnun kalmadım"
    ],
    "Etiket": [
        "olumlu",
        "olumlu",
        "olumlu",
        "olumsuz",
        "olumsuz",
        "olumsuz"
    ]
})

etiketli_metinler
````

# 72. Metinleri TF-IDF'e Dönüştürmek

````python
mini_tfidf = TfidfVectorizer(
    ngram_range=(
        1,
        2
    )
)

X_metin = (
    mini_tfidf.fit_transform(
        etiketli_metinler[
            "Metin"
        ]
    )
)

print(
    X_metin.shape
)
````

# 73. Hedef Etiketler

````python
y_metin = (
    etiketli_metinler[
        "Etiket"
    ]
)

print(
    y_metin.to_list()
)
````

# 74. İlk Metin Sınıflandırma Modeli

Bu bölüm yalnızca bir sonraki derse hazırlık amaçlı küçük bir örnektir.

Veri kümesi çok küçük olduğu için gerçek model performansı olarak yorumlanmamalıdır.


````python
from sklearn.linear_model import (
    LogisticRegression
)

mini_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

mini_model.fit(
    X_metin,
    y_metin
)

print(
    "Model eğitildi."
)
````

# 75. Yeni Metin Tahmini

````python
yeni = [
    "Bu proje gerçekten çok güzel"
]

yeni_tfidf = (
    mini_tfidf.transform(
        yeni
    )
)

print(
    mini_model.predict(
        yeni_tfidf
    )
)
````

Bu küçük örnek bize bir sonraki dersin temel zincirini göstermektedir:


# 76. Pipeline ile Metin Sınıflandırma

Vectorizer ve sınıflandırıcıyı Pipeline içinde birleştirebiliriz.


````python
from sklearn.pipeline import Pipeline

metin_pipeline = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            ngram_range=(
                1,
                2
            )
        )
    ),
    (
        "model",
        LogisticRegression(
            max_iter=1000,
            random_state=42
        )
    )
])

metin_pipeline.fit(
    etiketli_metinler[
        "Metin"
    ],
    etiketli_metinler[
        "Etiket"
    ]
)

print(
    metin_pipeline.predict([
        "Bu uygulamayı çok sevdim"
    ])
)
````

Pipeline sayesinde yeni metni manuel olarak önce TF-IDF'e çevirmek zorunda kalmadık.

Akış:

```text
Ham Metin
↓
TfidfVectorizer
↓
Model
```

olarak tek nesne içinde yönetildi.


# 77. Metin Pipeline'ını Kaydetmek

Metin uygulamalarında yalnızca modeli değil, vectorizer'ı da aynı şekilde korumamız gerekir.

Pipeline bu nedenle çok kullanışlıdır.


````python
import joblib

joblib.dump(
    metin_pipeline,
    "27-metin-pipeline.joblib"
)

print(
    "Pipeline kaydedildi."
)
````
