# 28. Oranı Koruyarak Boyutlandırmak

Görüntüyü belirli genişliğe göre oranı bozmadan küçültebiliriz.


````python
def genislige_gore_boyutlandir(
    goruntu,
    yeni_genislik
):
    h, w = goruntu.shape[:2]

    oran = (
        yeni_genislik
        /
        w
    )

    yeni_yukseklik = int(
        h * oran
    )

    return cv2.resize(
        goruntu,
        (
            yeni_genislik,
            yeni_yukseklik
        )
    )
````

````python
oranli = (
    genislige_gore_boyutlandir(
        okunan,
        250
    )
)

print(
    oranli.shape
)
````

# 33. Serbest Açıyla Döndürme

Serbest açı için affine dönüşüm matrisi oluşturabiliriz.


````python
h, w = okunan.shape[:2]

merkez = (
    w // 2,
    h // 2
)

donus_matrisi = (
    cv2.getRotationMatrix2D(
        merkez,
        25,
        1.0
    )
)

donmus = cv2.warpAffine(
    okunan,
    donus_matrisi,
    (
        w,
        h
    )
)

plt.imshow(
    cv2.cvtColor(
        donmus,
        cv2.COLOR_BGR2RGB
    )
)

plt.axis("off")
plt.title("25 Derece Döndürme")
plt.show()
````

# 37. Yapay Gürültülü Görüntü Oluşturmak

Filtreleme işlemlerini görmek için kontrollü bir görüntü oluşturalım.


````python
temiz_goruntu = np.zeros(
    (
        250,
        350
    ),
    dtype=np.uint8
)

cv2.rectangle(
    temiz_goruntu,
    (60, 60),
    (160, 180),
    220,
    -1
)

cv2.circle(
    temiz_goruntu,
    (250, 130),
    55,
    150,
    -1
)

rng = np.random.default_rng(
    42
)

gurultu = rng.normal(
    0,
    25,
    temiz_goruntu.shape
)

gurultulu = np.clip(
    temiz_goruntu.astype(
        np.float32
    )
    +
    gurultu,
    0,
    255
).astype(
    np.uint8
)

plt.imshow(
    gurultulu,
    cmap="gray"
)

plt.axis("off")
plt.title("Gürültülü Görüntü")
plt.show()
````

# 38. Ortalama Blur

`cv2.blur()` komşu piksellerin ortalamasını alarak yumuşatma uygular.


````python
ortalama_blur = cv2.blur(
    gurultulu,
    (
        5,
        5
    )
)

plt.imshow(
    ortalama_blur,
    cmap="gray"
)

plt.axis("off")
plt.title("Ortalama Blur")
plt.show()
````

# 39. Gaussian Blur

Gaussian Blur komşu piksellere Gaussian ağırlıklar uygulayarak yumuşatma yapar.


````python
gaussian = cv2.GaussianBlur(
    gurultulu,
    (
        5,
        5
    ),
    0
)

plt.imshow(
    gaussian,
    cmap="gray"
)

plt.axis("off")
plt.title("Gaussian Blur")
plt.show()
````

# 42. Threshold Nedir?

Threshold, gri görüntüyü belirli bir eşik değerine göre iki gruba ayırabilir.

Örneğin:

```text
piksel > 127 → 255
piksel <= 127 → 0
```

Bu işlem binary görüntü oluşturabilir.


# 43. Sabit Threshold Uygulamak

````python
_, binary = cv2.threshold(
    gaussian,
    100,
    255,
    cv2.THRESH_BINARY
)

plt.imshow(
    binary,
    cmap="gray"
)

plt.axis("off")
plt.title("Binary Threshold")
plt.show()
````

`cv2.threshold()` iki değer döndürür:

```text
kullanılan threshold
çıktı görüntüsü
```


# 44. Threshold Değerini Görmek

````python
esik, binary_2 = cv2.threshold(
    gaussian,
    120,
    255,
    cv2.THRESH_BINARY
)

print(
    "Threshold:",
    esik
)
````

# 45. Ters Binary Threshold

````python
_, ters_binary = cv2.threshold(
    gaussian,
    100,
    255,
    cv2.THRESH_BINARY_INV
)

plt.imshow(
    ters_binary,
    cmap="gray"
)

plt.axis("off")
plt.title("Ters Binary")
plt.show()
````

# 46. Otsu Threshold

Her görüntü için threshold değerini elle belirlemek istemeyebiliriz.

Otsu yöntemi histogramdan otomatik bir eşik değeri belirlemeye çalışır.


````python
otsu_degeri, otsu = cv2.threshold(
    gaussian,
    0,
    255,
    cv2.THRESH_BINARY
    +
    cv2.THRESH_OTSU
)

print(
    "Otsu threshold:",
    otsu_degeri
)

plt.imshow(
    otsu,
    cmap="gray"
)

plt.axis("off")
plt.title("Otsu Threshold")
plt.show()
````

# 47. Adaptive Threshold Kavramı

Görüntünün farklı bölgelerinde aydınlatma değişiyorsa tek bir global threshold yeterli olmayabilir.

Adaptive threshold her bölgenin çevresine göre farklı eşik değerleri kullanabilir.


````python
adaptive = cv2.adaptiveThreshold(
    gaussian,
    255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    21,
    3
)

plt.imshow(
    adaptive,
    cmap="gray"
)

plt.axis("off")
plt.title("Adaptive Threshold")
plt.show()
````

# 50. Blur Sonrası Kenar Tespiti Neden Yapılır?

Gürültülü görüntülerde küçük piksel değişimleri gereksiz kenarlar oluşturabilir.

Bu nedenle sık kullanılan akışlardan biri:

```text

```

şeklindedir.


# 51. Morfolojik İşlemler

Binary görüntüler üzerinde şekilleri düzenlemek için:

- erosion,
- dilation,
- opening,
- closing

gibi morfolojik işlemler kullanılabilir.


# 52. Morfolojik Kernel

````python
kernel = np.ones(
    (
        5,
        5
    ),
    dtype=np.uint8
)

print(
    kernel
)
````

# 53. Erosion

Erosion ön plandaki beyaz bölgeleri küçültme eğilimindedir.


````python
erozyon = cv2.erode(
    otsu,
    kernel,
    iterations=1
)

plt.imshow(
    erozyon,
    cmap="gray"
)

plt.axis("off")
plt.title("Erosion")
plt.show()
````

# 54. Dilation

Dilation beyaz bölgeleri büyütme eğilimindedir.


````python
genisleme = cv2.dilate(
    otsu,
    kernel,
    iterations=1
)

plt.imshow(
    genisleme,
    cmap="gray"
)

plt.axis("off")
plt.title("Dilation")
plt.show()
````

# 55. Opening

Opening:

```text
Erosion
↓
Dilation
```

işlemidir.

Küçük beyaz gürültüleri temizlemede yararlı olabilir.


````python
opening = cv2.morphologyEx(
    otsu,
    cv2.MORPH_OPEN,
    kernel
)

plt.imshow(
    opening,
    cmap="gray"
)

plt.axis("off")
plt.title("Opening")
plt.show()
````

# 56. Closing

Closing:

```text
Dilation
↓
Erosion
```

işlemidir.

Küçük siyah boşlukları kapatmada yararlı olabilir.


````python
closing = cv2.morphologyEx(
    otsu,
    cv2.MORPH_CLOSE,
    kernel
)

plt.imshow(
    closing,
    cmap="gray"
)

plt.axis("off")
plt.title("Closing")
plt.show()
````

# 57. Contour Nedir?

Contour, binary görüntüde bir nesnenin sınır noktalarını temsil eder.

Contour analiziyle:

- nesne sayısı,
- alan,
- çevre,
- bounding box,
- merkez

gibi bilgiler çıkarılabilir.


# 58. Contour Bulmak

Binary görüntü üzerinden contour'ları bulalım.


````python
contours, hierarchy = cv2.findContours(
    otsu,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

print(
    "Contour sayısı:",
    len(contours)
)
````

# 59. Contour Alanları

````python
alanlar = [
    cv2.contourArea(
        contour
    )
    for contour in contours
]

print(
    alanlar
)
````

# 60. Küçük Contour'ları Filtrelemek

Gürültü kaynaklı çok küçük contour'ları elemek için minimum alan sınırı kullanabiliriz.


````python
buyuk_contours = [
    contour
    for contour in contours
    if cv2.contourArea(
        contour
    ) > 500
]

print(
    "Büyük contour sayısı:",
    len(buyuk_contours)
)
````

# 61. Bounding Rectangle

Her contour etrafına dikdörtgen çizebiliriz.


````python
contour_goruntu = cv2.cvtColor(
    gurultulu,
    cv2.COLOR_GRAY2BGR
)

for contour in buyuk_contours:
    x, y, w, h = (
        cv2.boundingRect(
            contour
        )
    )

    cv2.rectangle(
        contour_goruntu,
        (x, y),
        (
            x + w,
            y + h
        ),
        (
            255,
            255,
            255
        ),
        2
    )

plt.imshow(
    cv2.cvtColor(
        contour_goruntu,
        cv2.COLOR_BGR2RGB
    )
)

plt.axis("off")
plt.title("Bounding Rectangle")
plt.show()
````

# 62. Contour Çizmek

````python
cizim = cv2.cvtColor(
    gurultulu,
    cv2.COLOR_GRAY2BGR
)

cv2.drawContours(
    cizim,
    buyuk_contours,
    -1,
    (
        255,
        255,
        255
    ),
    2
)

plt.imshow(
    cv2.cvtColor(
        cizim,
        cv2.COLOR_BGR2RGB
    )
)

plt.axis("off")
plt.title("Contour Çizimi")
plt.show()
````

# 63. Contour Çevresi

Bir contour'un çevresini `arcLength()` ile hesaplayabiliriz.


````python
for i, contour in enumerate(
    buyuk_contours,
    start=1
):
    alan = cv2.contourArea(
        contour
    )

    cevre = cv2.arcLength(
        contour,
        True
    )

    print(
        i,
        "Alan:",
        round(
            alan,
            2
        ),
        "Çevre:",
        round(
            cevre,
            2
        )
    )
````

# 64. Contour Merkezi

Moments kullanarak contour'un yaklaşık merkezini hesaplayabiliriz.


````python
merkezler = []

for contour in buyuk_contours:
    M = cv2.moments(
        contour
    )

    if M["m00"] != 0:
        cx = int(
            M["m10"]
            /
            M["m00"]
        )

        cy = int(
            M["m01"]
            /
            M["m00"]
        )

        merkezler.append(
            (
                cx,
                cy
            )
        )

print(
    merkezler
)
````

# 65. Merkezleri Görüntüde Göstermek

````python
merkez_goruntu = (
    contour_goruntu.copy()
)

for cx, cy in merkezler:
    cv2.circle(
        merkez_goruntu,
        (
            cx,
            cy
        ),
        5,
        (
            255,
            255,
            255
        ),
        -1
    )

plt.imshow(
    cv2.cvtColor(
        merkez_goruntu,
        cv2.COLOR_BGR2RGB
    )
)

plt.axis("off")
plt.title("Contour Merkezleri")
plt.show()
````

# 80. Yapay Zeka İçin Ön İşleme Fonksiyonu

Basit bir gri görüntü sınıflandırma modeli için ön işleme fonksiyonu oluşturalım.


````python
def ai_on_isleme(
    goruntu,
    boyut=28
):
    if goruntu.ndim == 3:
        goruntu = cv2.cvtColor(
            goruntu,
            cv2.COLOR_BGR2GRAY
        )

    goruntu = cv2.resize(
        goruntu,
        (
            boyut,
            boyut
        )
    )

    goruntu = (
        goruntu.astype(
            np.float32
        )
        /
        255.0
    )

    return goruntu
````

````python
hazir = ai_on_isleme(
    okunan,
    boyut=28
)

print(
    hazir.shape
)

print(
    hazir.dtype
)

print(
    hazir.min(),
    hazir.max()
)
````

# 87. Basit Şekil Sınıflandırma İçin Özellik Çıkarma

Contour üzerinden:

- alan,
- çevre,
- genişlik,
- yükseklik

gibi sayısal özellikler çıkarabiliriz.

Bunlar klasik makine öğrenmesi modeline girdi olabilir.


````python
sekil_ozellikleri = []

for contour in buyuk_contours:
    alan = cv2.contourArea(
        contour
    )

    cevre = cv2.arcLength(
        contour,
        True
    )

    x, y, w, h = (
        cv2.boundingRect(
            contour
        )
    )

    sekil_ozellikleri.append({
        "Alan": alan,
        "Cevre": cevre,
        "Genislik": w,
        "Yukseklik": h
    })

sekil_df = pd.DataFrame(
    sekil_ozellikleri
)

sekil_df
````

Bu yaklaşımda yapay zeka modeli ham pikseller yerine bizim çıkardığımız şekil özelliklerini kullanır.

# 88. En-Boy Oranı

Şeklin genişliği ve yüksekliğinden:


````python
if not sekil_df.empty:
    sekil_df[
        "EnBoyOrani"
    ] = (
        sekil_df[
            "Genislik"
        ]
        /
        sekil_df[
            "Yukseklik"
        ]
    )

sekil_df
````

# 89. Dairesellik Özelliği

Şekil analizi için kullanılan örnek bir özellik:

```text
4π × Alan
---------
Çevre²
```

değeridir.

Daireye benzeyen şekillerde değer 1'e yaklaşabilir.


````python
if not sekil_df.empty:
    sekil_df[
        "Dairesellik"
    ] = (
        4
        *
        np.pi
        *
        sekil_df["Alan"]
        /
        (
            sekil_df["Cevre"]
            ** 2
        )
    )

sekil_df
````
