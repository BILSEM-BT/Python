# 1. Yapay Sinir Ağı Nedir?

Yapay sinir ağı, giriş verilerini katmanlar boyunca işleyerek bir çıktı üretmeye çalışan makine öğrenmesi modelidir.

Basitleştirilmiş yapı:

```text
Girdiler
↓
Gizli Katman
↓
Gizli Katman
↓
Çıkış Katmanı
↓
Tahmin
```

Sinir ağları özellikle:

- görüntü,
- ses,
- metin,
- zaman serisi,
- karmaşık örüntüler

gibi alanlarda güçlü modellerin temelini oluşturur.


# 2. Yapay Nöron

En küçük temel yapı taşlarından biri yapay nörondur.

Bir nöron:

```text
Girdiler
↓
Ağırlıklı Toplam
↓
Bias
↓
Aktivasyon Fonksiyonu
↓
Çıktı
```

üretir.


# 3. Matematiksel Nöron Modeli

İki girişli basit nöron:

```text
z = x1*w1 + x2*w2 + b
```

Ardından:

```text
çıktı = aktivasyon(z)
```

hesaplanır.

Burada:

- `x` → giriş,
- `w` → ağırlık,
- `b` → bias,
- `z` → ağırlıklı toplamdır.


# 4. İlk Nöronu Python ile Hesaplayalım

````python
x1 = 0.8
x2 = 0.4

w1 = 0.6
w2 = -0.2

bias = 0.1

z = (
    x1 * w1
    +
    x2 * w2
    +
    bias
)

print(
    "Ağırlıklı toplam:",
    z
)
````

# 5. NumPy ile Aynı Hesap

````python
import numpy as np

x = np.array([
    0.8,
    0.4
])

w = np.array([
    0.6,
    -0.2
])

bias = 0.1

z = (
    np.dot(
        x,
        w
    )
    +
    bias
)

print(z)
````

# 6. Ağırlık Nedir?

Ağırlık, giriş bilgisinin nöron çıktısını ne kadar ve hangi yönde etkilediğini belirleyen öğrenilebilir değerdir.

Örneğin:

```text
pozitif ağırlık → girdiyi artırıcı yönde kullanabilir
negatif ağırlık → girdiyi azaltıcı yönde kullanabilir
```

Sinir ağı eğitim sırasında uygun ağırlıkları bulmaya çalışır.


# 7. Bias Nedir?

Bias nöronun karar sınırını yalnızca girişlerin sıfır noktasına bağlı kalmadan kaydırmasına yardımcı olan öğrenilebilir değerdir.

Denklem:

```text
z = x*w + b
```

şeklinde düşünülür.


# 8. Aktivasyon Fonksiyonu Neden Gerekir?

Katmanlar arasında yalnızca doğrusal işlemler kullanırsak çok sayıda katman eklemek modele gerçek anlamda karmaşık doğrusal olmayan örüntüler öğrenme gücü kazandırmaz.

Aktivasyon fonksiyonları modele **doğrusal olmayan davranış** kazandırır.


# 9. Sigmoid Fonksiyonu

Sigmoid:

```text
0 ile 1
```

arasında çıktı üretir.

Formül:

```text
sigmoid(x) = 1 / (1 + e^-x)
```

Özellikle ikili sınıflandırmanın çıkış katmanında kullanılabilir.


````python
def sigmoid(x):
    return (
        1
        /
        (
            1
            +
            np.exp(-x)
        )
    )

for deger in [
    -5,
    -2,
    0,
    2,
    5
]:
    print(
        deger,
        "->",
        sigmoid(deger)
    )
````

# 10. Sigmoid Grafiği

````python
import matplotlib.pyplot as plt

x_values = np.linspace(
    -8,
    8,
    300
)

y_values = sigmoid(
    x_values
)

plt.plot(
    x_values,
    y_values
)

plt.xlabel("x")
plt.ylabel("sigmoid(x)")
plt.title("Sigmoid Aktivasyon Fonksiyonu")
plt.grid()
plt.show()
````

# 11. ReLU Fonksiyonu

ReLU:

```text
ReLU(x) = max(0, x)
```

şeklindedir.

Gizli katmanlarda yaygın kullanılan aktivasyonlardan biridir.


````python
def relu(x):
    return np.maximum(
        0,
        x
    )

print(
    relu(
        np.array([
            -3,
            -1,
            0,
            2,
            5
        ])
    )
)
````

# 12. ReLU Grafiği

````python
relu_y = relu(
    x_values
)

plt.plot(
    x_values,
    relu_y
)

plt.xlabel("x")
plt.ylabel("ReLU(x)")
plt.title("ReLU Aktivasyon Fonksiyonu")
plt.grid()
plt.show()
````

# 14. Softmax Nedir?

Çok sınıflı sınıflandırmada çıkış katmanında her sınıf için skor üretilebilir.

Softmax bu skorları toplamı 1 olan değerlere dönüştürür.

Örnek:

```text
Kedi   → 0.10
Köpek  → 0.75
Kuş    → 0.15
```

En yüksek değer modelin seçtiği sınıf olabilir.


````python
def softmax(x):
    x = np.array(
        x,
        dtype=float
    )

    exp_x = np.exp(
        x
        -
        np.max(x)
    )

    return (
        exp_x
        /
        exp_x.sum()
    )

print(
    softmax([
        1.2,
        3.0,
        0.8
    ])
)
````

Softmax hesaplamasında `np.max(x)` çıkarılması sayısal kararlılığı artırmaya yardımcı olur.


# 15. Tek Nöron + Sigmoid

````python
x = np.array([
    0.8,
    0.4
])

w = np.array([
    0.6,
    -0.2
])

b = 0.1

z = np.dot(
    x,
    w
) + b

cikti = sigmoid(
    z
)

print(
    "z:",
    z
)

print(
    "çıktı:",
    cikti
)
````

# 16. Bir Katmanda Birden Fazla Nöron

Bir katmanda birden fazla nöron olabilir.

Örneğin:

```text
3 giriş
↓
4 nöron
↓
4 çıktı
```

Bu durumda ağırlıklar matris halinde tutulabilir.


````python
girdi = np.array([
    0.2,
    0.7,
    0.4
])

agirliklar = np.array([
    [0.4, 0.2, -0.1],
    [-0.5, 0.8, 0.3],
    [0.6, -0.2, 0.7],
    [0.1, 0.5, 0.9]
])

biaslar = np.array([
    0.1,
    0.0,
    -0.2,
    0.3
])

z = (
    agirliklar
    @
    girdi
    +
    biaslar
)

cikti = relu(
    z
)

print(
    "z:",
    z
)

print(
    "ReLU:",
    cikti
)
````

# 21. Loss Nedir?

Model tahmini ile gerçek hedef arasındaki hatayı ölçen fonksiyona **loss** denir.

# 22. Basit Mean Squared Error Örneği

````python
gercek = np.array([
    1.0,
    0.0,
    1.0
])

tahmin = np.array([
    0.8,
    0.3,
    0.6
])

mse = np.mean(
    (
        gercek
        -
        tahmin
    )
    ** 2
)

print(
    "MSE:",
    mse
)
````

Sınıflandırma sinir ağlarında çoğunlukla probleme uygun cross-entropy loss fonksiyonları kullanılır.


# 23. Gradient Descent Kavramı

Modelin ağırlıkları başlangıçta genellikle uygun değildir.

Eğitim sırasında:

```text
Tahmin
↓
Loss
↓
Gradient
↓
Ağırlıkları Güncelle
↓
Yeni Tahmin
```

döngüsü tekrar edilir.

Gradient, loss değerinin ağırlıklara göre hangi yönde değiştiği hakkında bilgi verir.


# 24. Basit Gradient Descent Örneği

Tek parametreli çok basit bir örnek düşünelim:

```text
y = w * x
```

Hedefimiz `w` değerini öğrenmek olsun.


````python
x = 2.0
gercek_y = 8.0

w = 1.0
learning_rate = 0.1

for adim in range(10):
    tahmin_y = (
        w
        *
        x
    )

    hata = (
        tahmin_y
        -
        gercek_y
    )

    loss = (
        hata
        ** 2
    )

    gradient = (
        2
        *
        hata
        *
        x
    )

    w = (
        w
        -
        learning_rate
        *
        gradient
    )

    print(
        adim + 1,
        "w:",
        round(w, 4),
        "loss:",
        round(loss, 4)
    )
````

Bu örnek gerçek sinir ağlarından çok daha basittir ancak ağırlık güncelleme mantığını gösterir.

Sinir ağı eğitiminde başlangıç ağırlıkları ve batch sıraları gibi işlemler rastgelelik içerebilir.

# 31. İlk Gerçek Problem: İki Sınıflı Veri

İki iç içe geçmiş yay biçimindeki bir veri kümesi oluşturalım.


````python
from sklearn.datasets import make_moons

X, y = make_moons(
    n_samples=1200,
    noise=0.20,
    random_state=42
)

print(
    X.shape,
    y.shape
)
````

# 32. Veriyi Görselleştirmek

````python
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=y
)

plt.xlabel("Özellik 1")
plt.ylabel("Özellik 2")
plt.title("İki Sınıflı Veri Kümesi")
plt.show()
````

Bu problem doğrusal bir çizgiyle kolayca ayrılmayabilir.

Sinir ağı gizli katmanlar sayesinde doğrusal olmayan karar sınırları öğrenebilir.


# 33. Train-Test Ayrımı

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print(
    X_train.shape,
    X_test.shape
)
````

# 34. StandardScaler

Sinir ağı eğitiminde özellikleri benzer ölçeklere getirmek eğitim sürecini kolaylaştırabilir.


````python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
)

X_test_scaled = scaler.transform(
    X_test
)

print(
    X_train_scaled.mean(
        axis=0
    )
)
````

Scaler yalnızca eğitim verisinde `fit()` edilmiştir.

Test verisinde yalnızca `transform()` uygulanmıştır.


# 35. İlk Keras Sequential Modeli

Katmanlarımız:

```text
2 özellik
↓
Dense(16, ReLU)
↓
Dense(8, ReLU)
↓
Dense(1, Sigmoid)
```


````python
model = keras.Sequential([
    keras.Input(
        shape=(2,)
    ),
    layers.Dense(
        16,
        activation="relu"
    ),
    layers.Dense(
        8,
        activation="relu"
    ),
    layers.Dense(
        1,
        activation="sigmoid"
    )
])

model.summary()
````

# 36. `keras.Input`

```text

```

içerdiğini modele açıkça belirttik.

Bu sayede model katmanlarının şekilleri baştan oluşturulabilir.

Modeli eğitmeden önce:

````python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=[
        "accuracy"
    ]
)
````

# 40. Binary Crossentropy

# 41. Modeli Eğitmek

`fit()` eğitim sürecini başlatır.


````python
history = model.fit(
    X_train_scaled,
    y_train,
    validation_split=0.20,
    epochs=30,
    batch_size=32,
    verbose=0
)

print(
    "Eğitim tamamlandı."
)
````

# 42. History Nedir?

`fit()` eğitim sürecindeki metrikleri `History` nesnesinde saklar.


````python
print(
    history.history.keys()
)
````

Genellikle:

- loss,
- accuracy,
- val_loss,
- val_accuracy

gibi bilgiler bulunur.


# 43. Eğitim Loss Grafiği

````python
plt.plot(
    history.history[
        "loss"
    ],
    label="Train Loss"
)

plt.plot(
    history.history[
        "val_loss"
    ],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Eğitim ve Validation Loss")
plt.legend()
plt.show()
````

# 44. Eğitim Accuracy Grafiği

````python
plt.plot(
    history.history[
        "accuracy"
    ],
    label="Train Accuracy"
)

plt.plot(
    history.history[
        "val_accuracy"
    ],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Eğitim ve Validation Accuracy")
plt.legend()
plt.show()
````

# 45. Learning Curve Nasıl Yorumlanır?

Eğer:

```text
Train loss düşüyor
Validation loss da düşüyor
```

model öğreniyor olabilir.

Eğer:

```text
Train loss düşmeye devam ediyor
Validation loss yükseliyor
```

overfitting ihtimali olabilir.


# 46. Test Verisinde Evaluate

````python
test_loss, test_accuracy = (
    model.evaluate(
        X_test_scaled,
        y_test,
        verbose=0
    )
)

print(
    "Test Loss:",
    test_loss
)

print(
    "Test Accuracy:",
    test_accuracy
)
````

# 47. `predict()` ile Tahmin

Sinir ağı doğrudan sınıf etiketi yerine sigmoid çıktısı üretir.


````python
olasliklar = model.predict(
    X_test_scaled[:10],
    verbose=0
)

print(
    olasliklar.reshape(-1)
)
````

# 48. Olasılığı Sınıfa Çevirmek

Varsayılan basit eşik:

```text
>= 0.5 → 1
< 0.5 → 0
```

şeklinde kullanılabilir.


````python
tahmin_sinif = (
    olasliklar.reshape(-1)
    >= 0.5
).astype(int)

print(
    tahmin_sinif
)

print(
    y_test[:10]
)
````

# 49. Bütün Test Tahminleri

````python
test_olasilik = (
    model.predict(
        X_test_scaled,
        verbose=0
    )
    .reshape(-1)
)

y_pred = (
    test_olasilik
    >= 0.5
).astype(int)
````

# 50. Scikit-learn ile Ek Değerlendirme

````python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

print(
    "Accuracy:",
    accuracy_score(
        y_test,
        y_pred
    )
)

print()

print(
    classification_report(
        y_test,
        y_pred
    )
)
````

# 51. Confusion Matrix

````python
cm = confusion_matrix(
    y_test,
    y_pred
)

print(
    cm
)
````

# 52. Yeni Bir Nokta Tahmini

````python
yeni = np.array([
    [0.5, 0.2]
])

yeni_scaled = scaler.transform(
    yeni
)

p = model.predict(
    yeni_scaled,
    verbose=0
)[0, 0]

sinif = int(
    p >= 0.5
)

print(
    "Olasılık:",
    p
)

print(
    "Sınıf:",
    sinif
)
````

# 53. Karar Sınırını Görselleştirmek

Modelin iki boyutlu uzayda nasıl sınıflandırma yaptığını görelim.

Validation performansı artık iyileşmiyorsa eğitimi otomatik olarak durdurabiliriz.

Bu amaçla `EarlyStopping` callback'i kullanılabilir.


````python
early_stop = keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

print(
    early_stop
)
````

# 58. EarlyStopping ile Yeni Model Eğitmek

````python
keras.utils.set_random_seed(
    42
)

erken_model = keras.Sequential([
    keras.Input(
        shape=(2,)
    ),
    layers.Dense(
        32,
        activation="relu"
    ),
    layers.Dense(
        16,
        activation="relu"
    ),
    layers.Dense(
        1,
        activation="sigmoid"
    )
])

erken_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

erken_history = erken_model.fit(
    X_train_scaled,
    y_train,
    validation_split=0.20,
    epochs=100,
    batch_size=32,
    callbacks=[
        early_stop
    ],
    verbose=0
)

print(
    "Gerçekleşen epoch:",
    len(
        erken_history.history[
            "loss"
        ]
    )
)
````

# 63. Digits Veri Kümesi

````python
from sklearn.datasets import load_digits

digits = load_digits()

X_digits = (
    digits.data.astype(
        np.float32
    )
)

y_digits = (
    digits.target
)

print(
    X_digits.shape,
    y_digits.shape
)
````

# 65. Piksel Normalizasyonu

Digits piksel değerleri 0-16 arasındadır.

Bunları:

```text

```

````python
X_digits = (
    X_digits
    /
    16.0
)

print(
    X_digits.min(),
    X_digits.max()
)
````

Yapı:

```text

↓
Dense(128, ReLU)
↓

↓
```

# 82. Tahmin Fonksiyonu

````python
def rakam_tahmin(
    model,
    goruntu_8x8
):
    goruntu = np.array(
        goruntu_8x8,
        dtype=np.float32
    )

    if goruntu.shape != (
        8,
        8
    ):
        raise ValueError(
            "Görüntü 8x8 olmalıdır."
        )

    if goruntu.max() > 1:
        goruntu = (
            goruntu
            /
            16.0
        )

    girdi = goruntu.reshape(
        1,
        64
    )

    olasilik = model.predict(
        girdi,
        verbose=0
    )[0]

    sinif = int(
        np.argmax(
            olasilik
        )
    )

    return {
        "tahmin": sinif,
        "skorlar": olasilik
    }
````

````python
sonuc = rakam_tahmin(
    rakam_modeli,
    digits.images[5]
)

print(
    "Gerçek:",
    digits.target[5]
)

print(
    "Tahmin:",
    sonuc["tahmin"]
)
````

# 90. Backpropagation Kavramı

Sinir ağı eğitiminde loss değeri çıktı katmanından başlayarak önceki katmanlardaki parametrelerin etkisini hesaplamak için geriye doğru kullanılır.

Bu işleme **backpropagation / geri yayılım** denir.

Basitleştirilmiş zincir:

```text
Forward Pass
↓
Tahmin
↓
Loss
↓
Backpropagation
↓
Gradient
↓
Optimizer
↓
Ağırlık Güncelleme
```

TensorFlow/Keras bu gradient hesaplarını otomatik olarak yapar.


# 91. Automatic Differentiation

TensorFlow, hesaplama işlemlerinin türevlerini otomatik olarak hesaplayabilir.

Bu sayede yüz binlerce veya milyonlarca parametre içeren modellerin gradientlerini elle yazmak zorunda kalmayız.

İleri seviye derslerde `GradientTape` kullanarak bu işlemin daha alt seviyesine inebiliriz.
