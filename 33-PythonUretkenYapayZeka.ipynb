# 6. Basit Eğitim Tokenizer'ı

Gerçek LLM tokenizer'ı yerine kavramı anlamak için basit bir regex tokenizer yazalım.

Bu tokenizer yalnızca eğitim amaçlıdır.


````python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def basit_tokenize(metin):
    return re.findall(
        r"\w+|[^\w\s]",
        metin.lower(),
        flags=re.UNICODE
    )

ornek = (
    "Python ile yapay zeka uygulaması geliştiriyoruz!"
)

print(
    basit_tokenize(
        ornek
    )
)
````

# 17. Basit Context Bütçesi Simülasyonu

Öğretim amacıyla en fazla 12 token saklayan basit bir sistem yazalım.


````python
def context_sinirla(
    tokenlar,
    maksimum=12
):
    if len(
        tokenlar
    ) <= maksimum:
        return tokenlar

    return tokenlar[
        -maksimum:
    ]

uzun_metin = (
    "python ile yapay zeka öğreniyoruz "
    "sonra veri analizi yapıyoruz "
    "ardından web uygulaması geliştiriyoruz"
)

uzun_tokenlar = basit_tokenize(
    uzun_metin
)

print(
    "Tüm token:",
    uzun_tokenlar
)

print()

print(
    "Context:",
    context_sinirla(
        uzun_tokenlar,
        maksimum=12
    )
)
````

# 20. Eğitim Tokenları

````python
egitim_tokenlari = basit_tokenize(
    egitim_metni
)

print(
    egitim_tokenlari[:40]
)
````

# 21. Bigram Dil Modeli

Bir token'dan sonra gelen tokenları sayalım.


````python
from collections import defaultdict, Counter

gecisler = defaultdict(
    Counter
)

for onceki, sonraki in zip(
    egitim_tokenlari[:-1],
    egitim_tokenlari[1:]
):
    gecisler[
        onceki
    ][
        sonraki
    ] += 1

print(
    gecisler[
        "python"
    ]
)
````

# 22. Olasılık Dağılımı

Sayım sonuçlarını olasılığa dönüştürelim.


````python
def sonraki_token_olasiliklari(
    token
):
    sayac = gecisler.get(
        token,
        Counter()
    )

    toplam = sum(
        sayac.values()
    )

    if toplam == 0:
        return {}

    return {
        aday:
            adet / toplam
        for aday, adet
        in sayac.items()
    }

print(
    sonraki_token_olasiliklari(
        "ile"
    )
)
````

# 23. En Olası Tokenı Seçmek

Greedy decoding fikri:


````python
def en_olasi_token(
    token
):
    olasiliklar = (
        sonraki_token_olasiliklari(
            token
        )
    )

    if not olasiliklar:
        return None

    return max(
        olasiliklar,
        key=olasiliklar.get
    )

print(
    en_olasi_token(
        "python"
    )
)
````

# 24. Greedy Metin Üretimi

Her adımda en olası tokenı seçelim.


````python
def greedy_uret(
    baslangic,
    adim=10
):
    sonuc = [
        baslangic
    ]

    mevcut = baslangic

    for _ in range(
        adim
    ):
        yeni = en_olasi_token(
            mevcut
        )

        if yeni is None:
            break

        sonuc.append(
            yeni
        )

        mevcut = yeni

    return " ".join(
        sonuc
    )

print(
    greedy_uret(
        "python",
        adim=12
    )
)
````

Greedy üretim deterministik ve tekrar eden çıktılara eğilimli olabilir.

Gerçek LLM'ler daha gelişmiş sampling ve decoding yöntemleri kullanabilir.


# 25. Sampling

En yüksek olasılığı zorunlu seçmek yerine olasılık dağılımından örnek seçebiliriz.


````python
def rastgele_sonraki_token(
    token,
    rng
):
    olasiliklar = (
        sonraki_token_olasiliklari(
            token
        )
    )

    if not olasiliklar:
        return None

    adaylar = list(
        olasiliklar.keys()
    )

    p = np.array(
        list(
            olasiliklar.values()
        ),
        dtype=float
    )

    p = p / p.sum()

    return rng.choice(
        adaylar,
        p=p
    )
````

# 26. Sampling ile Metin Üretmek

````python
def sampling_uret(
    baslangic,
    adim=10,
    seed=42
):
    yerel_rng = (
        np.random.default_rng(
            seed
        )
    )

    sonuc = [
        baslangic
    ]

    mevcut = baslangic

    for _ in range(
        adim
    ):
        yeni = rastgele_sonraki_token(
            mevcut,
            yerel_rng
        )

        if yeni is None:
            break

        sonuc.append(
            str(
                yeni
            )
        )

        mevcut = str(
            yeni
        )

    return " ".join(
        sonuc
    )

for seed in [
    1,
    2,
    3
]:
    print(
        sampling_uret(
            "python",
            adim=12,
            seed=seed
        )
    )
````

# 27. Temperature Nedir?

Temperature, sampling sırasında olasılık dağılımının ne kadar keskin veya yaygın kullanılacağını etkileyen kavramlardan biridir.

Genel sezgi:

```text
Düşük temperature
→ güçlü adaylara daha fazla ağırlık
→ daha tutarlı / daha az çeşitli
```

```text
Yüksek temperature
→ alternatif adaylara daha fazla şans
→ daha çeşitli / daha rastlantısal
```

API ve model desteği modele göre değişebilir. Bu bölümde kavramı kendi mini modelimiz üzerinde uygulayacağız.


# 28. Temperature Matematiği

Olasılıkları doğrudan değiştirmek yerine logit benzeri değerler üzerinde sıcaklık uygulayabiliriz.

Eğitim amaçlı basitleştirilmiş fonksiyon:


````python
def temperature_uygula(
    olasiliklar,
    temperature
):
    p = np.array(
        olasiliklar,
        dtype=float
    )

    p = p / p.sum()

    if temperature <= 0:
        sonuc = np.zeros_like(
            p
        )

        sonuc[
            np.argmax(
                p
            )
        ] = 1.0

        return sonuc

    log_p = np.log(
        p
        +
        1e-12
    )

    scaled = (
        log_p
        /
        temperature
    )

    scaled = scaled - np.max(
        scaled
    )

    exp_values = np.exp(
        scaled
    )

    return (
        exp_values
        /
        exp_values.sum()
    )
````

# 29. Temperature Karşılaştırması

````python
orijinal = np.array([
    0.60,
    0.25,
    0.10,
    0.05
])

for t in [
    0.3,
    0.7,
    1.0,
    1.5
]:
    print(
        "T =",
        t,
        "->",
        np.round(
            temperature_uygula(
                orijinal,
                t
            ),
            3
        )
    )
````

# 30. Temperature Grafiği

````python
adaylar = [
    "A",
    "B",
    "C",
    "D"
]

for t in [
    0.3,
    1.0,
    1.5
]:
    plt.figure()

    plt.bar(
        adaylar,
        temperature_uygula(
            orijinal,
            t
        )
    )

    plt.ylim(
        0,
        1
    )

    plt.title(
        f"Temperature = {t}"
    )

    plt.ylabel(
        "Olasılık"
    )

    plt.show()
````

# 31. Temperature'lı Mini Üretici

````python
def temperature_ile_sonraki(
    token,
    rng,
    temperature=1.0
):
    olasiliklar = (
        sonraki_token_olasiliklari(
            token
        )
    )

    if not olasiliklar:
        return None

    adaylar = list(
        olasiliklar.keys()
    )

    p = np.array(
        list(
            olasiliklar.values()
        )
    )

    yeni_p = temperature_uygula(
        p,
        temperature
    )

    return str(
        rng.choice(
            adaylar,
            p=yeni_p
        )
    )
````

# 32. Farklı Temperature ile Üretim

````python
def temperature_ile_uret(
    baslangic,
    temperature,
    adim=12,
    seed=42
):
    yerel_rng = (
        np.random.default_rng(
            seed
        )
    )

    sonuc = [
        baslangic
    ]

    mevcut = baslangic

    for _ in range(
        adim
    ):
        yeni = temperature_ile_sonraki(
            mevcut,
            yerel_rng,
            temperature
        )

        if yeni is None:
            break

        sonuc.append(
            yeni
        )

        mevcut = yeni

    return " ".join(
        sonuc
    )

for t in [
    0.3,
    1.0,
    1.8
]:
    print()
    print(
        "T =",
        t
    )

    print(
        temperature_ile_uret(
            "python",
            t,
            seed=4
        )
    )
````

# 42. Prompt Şablonu

Genel amaçlı bir prompt şablonu:

```text

...

Kısıtlar:
...
```

````python
def prompt_olustur(
    gorev,
    baglam,
    kisitlar,
    cikti_bicimi
):
    return f'''
Görev:
{gorev}

Bağlam:
{baglam}

Kısıtlar:
{kisitlar}

Çıktı biçimi:
{cikti_bicimi}
'''.strip()

print(
    prompt_olustur(
        gorev=(
            "Python listelerini açıkla"
        ),
        baglam=(
            "Yeni başlayan BİLSEM öğrencisi"
        ),
        kisitlar=(
            "Teknik terimleri tanımla"
        ),
        cikti_bicimi=(
            "Tanım + 2 örnek + 3 soru"
        )
    )
)
````

# 69. LLM Uygulama Fonksiyonu Taslağı

API kapalıyken ağ isteği yapmadan güvenli biçimde çalışır.


````python
def llm_cevap_uret(
    user_input,
    instructions=(
        "Kısa ve açık cevap ver."
    )
):
    if not API_CAGRISI_AKTIF:
        return {
            "durum": "simulasyon",
            "mesaj": (
                "API_CAGRISI_AKTIF=False. "
                "Gerçek istek gönderilmedi."
            ),
            "instructions":
                instructions,
            "input":
                user_input
        }

    if not OPENAI_SDK_VAR:
        return {
            "durum": "hata",
            "mesaj":
                "OpenAI SDK kurulu değil."
        }

    if not OPENAI_KEY_VAR:
        return {
            "durum": "hata",
            "mesaj":
                "OPENAI_API_KEY tanımlı değil."
        }

    from openai import OpenAI

    client = OpenAI()

    response = (
        client.responses.create(
            model=MODEL_ADI,
            instructions=instructions,
            input=user_input
        )
    )

    return {
        "durum": "basarili",
        "metin":
            response.output_text
    }
````

````python
print(
    llm_cevap_uret(
        "Python'da tuple nedir?",
        instructions=(
            "Yeni başlayan öğrenciye "
            "uygun açıkla."
        )
    )
)
````

# 106. LLM Client Sınıfı Taslağı

Varsayılan durumda API çağrısı yapmaz.


````python
class LLMClient:
    def __init__(
        self,
        model=MODEL_ADI
    ):
        self.model = model

    def hazir_mi(
        self
    ):
        return (
            API_CAGRISI_AKTIF
            and
            OPENAI_SDK_VAR
            and
            OPENAI_KEY_VAR
        )

    def cevapla(
        self,
        instructions,
        user_input
    ):
        if not self.hazir_mi():
            return {
                "durum":
                    "api_pasif",
                "model":
                    self.model,
                "instructions":
                    instructions,
                "input":
                    user_input
            }

        from openai import OpenAI

        client = OpenAI()

        response = (
            client.responses.create(
                model=self.model,
                instructions=(
                    instructions
                ),
                input=(
                    user_input
                )
            )
        )

        return {
            "durum":
                "basarili",
            "model":
                self.model,
            "metin":
                response.output_text
        }
````

````python
llm_client = LLMClient()

print(
    llm_client.cevapla(
        "Kısa cevap ver.",
        "Python nedir?"
    )
)
````
