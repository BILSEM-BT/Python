# Ders 01 — Python'a Giriş

**Soru sayısı:** 1

Aşağıdaki yanıtlar bu dersin öğrenci çalışma soruları için örnek çözümlerdir.


## Soru (01-01)

**Soru:** Bir öğrencinin adı, soyadı, sınıfı, yaşı ve ders notlarını program içinde tutun; notları liste olarak saklayıp ortalamasını hesaplayın.

**Yanıt:**

```python
ogrenci = {
    "ad": "Ayşe",
    "soyad": "Yılmaz",
    "sinif": "7/A",
    "yas": 12,
    "notlar": [85, 92, 78]
}

ortalama = sum(ogrenci["notlar"]) / len(ogrenci["notlar"])

print("Öğrenci:", ogrenci["ad"], ogrenci["soyad"])
print("Sınıf:", ogrenci["sinif"])
print("Yaş:", ogrenci["yas"])
print("Notlar:", ogrenci["notlar"])
print("Ortalama:", round(ortalama, 2))
```

## Soru (03-01)

**Soru:** Kullanıcıdan iki sayı alın ve bölme işlemi yapın. Kullanıcının sıfıra bölme yapması durumunda uygun mesaj gösterin.

**Yanıt:**

Ders içeriğindeki **Hata Yönetimi ve Kullanıcı Girdileri** yapısı kullanılarak örnek çözüm:

```python
try:
    yas = int(input("Yaşınızı girin: "))

    if yas >= 18:
        print("Yetişkinsiniz.")
    else:
        print("18 yaşından küçüksünüz.")

except ValueError:
    print("Yaş bilgisi sayı olarak girilmelidir.")
```

## Soru (03-02)

**Soru:** Kullanıcıdan yaş alın. Sayı yerine metin girilirse programın kapanmasını engelleyin.

**Yanıt:**

Ders içeriğindeki **Hata Yönetimi ve Kullanıcı Girdileri** yapısı kullanılarak örnek çözüm:

```python
try:
    yas = int(input("Yaşınızı girin: "))

    if yas >= 18:
        print("Yetişkinsiniz.")
    else:
        print("18 yaşından küçüksünüz.")

except ValueError:
    print("Yaş bilgisi sayı olarak girilmelidir.")
```

## Soru (03-03)

## Soru (03-04)

**Soru:** Bir sözlük oluşturun. Kullanıcıdan anahtar alın ve ilgili değeri göstermeye çalışın. Olmayan anahtar için `KeyError` durumunu yönetin.

**Yanıt:**

Ders içeriğindeki **`KeyError` Örneği** yapısı kullanılarak örnek çözüm:

```python
ogrenci = {
    "ad": "Ali",
    "sinif": 8
}

try:
    print(ogrenci["yas"])
except KeyError:
    print("Sözlükte bu anahtar bulunmuyor.")
```

## Soru (03-05)

**Soru:** Basit bir hesap makinesi oluşturun. Sayı girişlerini, sıfıra bölmeyi ve geçersiz işlem seçimini kontrol edin.

**Yanıt:**

Ders içeriğindeki **Gerçek Uygulama Örneği: Basit Hesap Makinesi** yapısı kullanılarak örnek çözüm:

```python
try:
    sayi1 = float(input("Birinci sayıyı girin: "))
    sayi2 = float(input("İkinci sayıyı girin: "))
    islem = input("İşlem seçin (+, -, *, /): ")

    if islem == "+":
        sonuc = sayi1 + sayi2
    elif islem == "-":
        sonuc = sayi1 - sayi2
    elif islem == "*":
        sonuc = sayi1 * sayi2
    elif islem == "/":
        sonuc = sayi1 / sayi2
    else:
        print("Geçersiz işlem.")
        sonuc = None

    if sonuc is not None:
        print("Sonuç:", sonuc)

except ValueError:
    print("Lütfen sayıları doğru biçimde girin.")
except ZeroDivisionError:
    print("Bir sayı sıfıra bölünemez.")
finally:
    print("Hesaplama işlemi sona erdi.")
```

## Soru (03-06)

**Soru:** Kullanıcıdan doğru bir sayı girene kadar tekrar isteyen bir program yazın. Bu uygulamada `while`, `try`, `except` ve `break` yapılarını birlikte kullanın.

**Yanıt:**

Ders içeriğindeki **`try` ve `except`** yapısı kullanılarak örnek çözüm:

```python
try:
    sayi = int(input("Bir sayı girin: "))
    print("Girdiğiniz sayı:", sayi)
except:
    print("Geçerli bir sayı girmelisiniz.")
```

## Soru (04-05)

**Soru:** Bir listedeki sayıların toplamını ve ortalamasını bulun.

```python
notlar = [70, 85, 60, 90, 75]

toplam = 0

for notu in notlar:
    toplam += notu

ortalama = toplam / len(notlar)

print("Toplam:", toplam)
print("Ortalama:", ortalama)
```

## Soru (04-06)

## Soru (04-10)

**Soru:** Bir sayının asal olup olmadığını kontrol eden programı yazın.

**Yanıt:**

Ders içeriğindeki **Asal Sayı Kontrolü** yapısı kullanılarak örnek çözüm:

```python
sayi = 29
asal = True

if sayi < 2:
    asal = False
else:
    for i in range(2, sayi):
        if sayi % i == 0:
            asal = False
            break

if asal:
    print("Asal sayı")
else:
    print("Asal değil")
```

## Soru (05-06)

**Soru:** Kullanıcıdan 5 sayı alarak bir liste oluşturun. Hatalı girişleri `try-except` ile kontrol edin.

**Yanıt:**

Ders içeriğindeki **Kullanıcıdan Sayı Listesi Almak ve Hata Yönetimi** yapısı kullanılarak örnek çözüm:

```python
sayilar = []

for i in range(5):
    while True:
        try:
            sayi = int(input(f"{i + 1}. sayıyı girin: "))
            sayilar.append(sayi)
            break
        except ValueError:
            print("Lütfen geçerli bir tam sayı girin.")

print("Oluşturulan liste:", sayilar)
```

## Soru (07-02)

**Soru:** Bir öğrencinin ders ve notlarını dictionary içinde saklayın.

**Yanıt:**

Ders içeriğindeki **Dictionary'den En Yüksek Notu Bulmak** yapısı kullanılarak örnek çözüm:

```python
def en_yuksek_not(notlar):
    ders = None
    puan = 0

    for isim, not_degeri in notlar.items():
        if not_degeri > puan:
            ders = isim
            puan = not_degeri

    return ders, puan

notlar = {
    "Python": 90,
    "Matematik": 80,
    "Fizik": 75,
    "Biyoloji": 95
}

ders, puan = en_yuksek_not(notlar)

print("En yüksek ders:", ders)
print("Not:", puan)
```

## Soru (08-09)

**Soru:** Sıfıra bölme hatasını yöneten bölme fonksiyonu yazın.

**Yanıt:**

Ders içeriğindeki **Mini Proje: Hesap Makinesi** yapısı kullanılarak örnek çözüm:

```python
def topla(a, b):
    return a + b

def cikar(a, b):
    return a - b

def carp(a, b):
    return a * b

def bol(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Sıfıra bölme yapılamaz."

print("Toplam:", topla(10, 5))
print("Fark:", cikar(10, 5))
print("Çarpım:", carp(10, 5))
print("Bölüm:", bol(10, 5))
print("Hatalı bölüm:", bol(10, 0))
```

```python
def sesli_harf_sayisi(metin):
    sesliler = "aeıioöuü"
    return sum(
        1
        for harf in metin.lower()
        if harf in sesliler
    )

print(
    sesli_harf_sayisi(
        "Niyazi Sayın BİLSEM"
    )
)
```

## Soru (08-12)

**Soru:** Bir listedeki en küçük, en büyük, toplam ve ortalamayı birlikte döndüren fonksiyon yazın.

**Yanıt:**

Ders içeriğindeki **Liste Analizi Fonksiyonu** yapısı kullanılarak örnek çözüm:

```python
def liste_analizi(sayilar):
    en_kucuk = min(sayilar)
    en_buyuk = max(sayilar)
    toplam = sum(sayilar)
    ortalama = toplam / len(sayilar)

    return en_kucuk, en_buyuk, toplam, ortalama

veriler = [15, 20, 35, 40, 50]

kucuk, buyuk, toplam, ortalama = liste_analizi(veriler)

print("En küçük:", kucuk)
print("En büyük:", buyuk)
print("Toplam:", toplam)
print("Ortalama:", ortalama)
```

## Soru (08-13)

**Soru:** Bir ürün fiyat listesini alıp belirli fiyatın üzerindeki ürünleri döndüren fonksiyon yazın.

**Yanıt:**

```python
def pahali_urunler(fiyatlar, esik):
    return [
        fiyat
        for fiyat in fiyatlar
        if fiyat > esik
    ]

print(
    pahali_urunler(
        [45, 120, 80, 250, 99],
        100
    )
)
```

## Soru (08-14)

**Soru:** Öğrenci notlarını alıp ortalama ve başarı durumunu döndüren fonksiyon yazın.

**Yanıt:**

Ders içeriğindeki **Mini Proje: Öğrenci Not Analizi** yapısı kullanılarak örnek çözüm:

```python
def ortalama_hesapla(notlar):
    return sum(notlar) / len(notlar)

def basarili_sayisi(notlar):
    sayac = 0

    for notu in notlar:
        if notu >= 50:
            sayac += 1

    return sayac

def en_yuksek_not(notlar):
    return max(notlar)

notlar = [45, 70, 82, 35, 91, 64]

ortalama = ortalama_hesapla(notlar)
basarili = basarili_sayisi(notlar)
en_yuksek = en_yuksek_not(notlar)

print("Ortalama:", ortalama)
print("Başarılı sayısı:", basarili)
print("En yüksek not:", en_yuksek)
```

## Soru (09-08)

**Soru:** Fonksiyonlarla öğrenci kayıt sistemi oluşturun.

**Yanıt:**

Ders içeriğindeki **Mini Uygulama: JSON Tabanlı Öğrenci Kayıt Sistemi** yapısı kullanılarak örnek çözüm:

```python
import json

def ogrencileri_yukle():
    try:
        with open("ogrenciler.json", "r", encoding="utf-8") as dosya:
            return json.load(dosya)
    except FileNotFoundError:
        return []
    except json.JSONDecodeError:
        print("Veri dosyası bozuk.")
        return []


def ogrencileri_kaydet(ogrenciler):
    with open("ogrenciler.json", "w", encoding="utf-8") as dosya:
        json.dump(ogrenciler, dosya, ensure_ascii=False, indent=4)


ogrenciler = ogrencileri_yukle()

ogrenciler.append({
    "isim": "Can",
    "puan": 87
})

ogrencileri_kaydet(ogrenciler)

print(ogrenciler)
```

## Soru (10-15)

**Soru:** Modülleri ve dosya işlemlerini birleştirerek küçük bir kayıt sistemi geliştirin.

**Yanıt:**

Ders içeriğindeki **Modüller ile Dosya İşlemlerini Birleştirmek** yapısı kullanılarak örnek çözüm:

```python
def kaydet(veri):
    with open("veriler.txt", "a", encoding="utf-8") as dosya:
        dosya.write(veri + "\n")

def oku():
    try:
        with open("veriler.txt", "r", encoding="utf-8") as dosya:
            return dosya.readlines()
    except FileNotFoundError:
        return []
```

## Soru (11-15)

**Soru:** Bir haftalık sıcaklık verilerini analiz edin.

**Yanıt:**

Ders içeriğindeki **NumPy ile Fonksiyon Yazmak** yapısı kullanılarak örnek çözüm:

```python
def veri_ozeti(veriler):
    return {
        "toplam": np.sum(veriler),
        "ortalama": np.mean(veriler),
        "minimum": np.min(veriler),
        "maksimum": np.max(veriler),
        "standart_sapma": np.std(veriler)
    }

sayilar = np.array([10, 15, 20, 25, 30])

sonuc = veri_ozeti(sayilar)

print(sonuc)
```

## Soru (15-07)

**Soru:** Dört işlem hesap makinesi geliştirin.

**Yanıt:**

Ders içeriğindeki **Mini Uygulama: Dört İşlem Hesap Makinesi** yapısı kullanılarak örnek çözüm:

```python
import tkinter as tk
from tkinter import messagebox

def hesapla(islem):
    try:
        a = float(giris1.get())
        b = float(giris2.get())

        if islem == "+":
            sonuc_degeri = a + b
        elif islem == "-":
            sonuc_degeri = a - b
        elif islem == "*":
            sonuc_degeri = a * b
        elif islem == "/":
            if b == 0:
                messagebox.showerror(
                    "Hata",
                    "Sıfıra bölme yapılamaz."
                )
                return

            sonuc_degeri = a / b

        sonuc.config(
            text=f"Sonuç: {sonuc_degeri}"
        )

    except ValueError:
        messagebox.showerror(
            "Hata",
            "Geçerli sayılar girin."
        )

pencere = tk.Tk()
pencere.title("Hesap Makinesi")
pencere.geometry("400x330")

tk.Label(pencere, text="1. sayı").pack()
giris1 = tk.Entry(pencere)
giris1.pack()

tk.Label(pencere, text="2. sayı").pack()
giris2 = tk.Entry(pencere)
giris2.pack()

buton_frame = tk.Frame(pencere)
buton_frame.pack(pady=20)

tk.Button(
    buton_frame,
    text="+",
    width=5,
    command=lambda: hesapla("+")
).grid(row=0, column=0, padx=5)

tk.Button(
    buton_frame,
    text="-",
    width=5,
    command=lambda: hesapla("-")
).grid(row=0, column=1, padx=5)

tk.Button(
    buton_frame,
    text="*",
    width=5,
    command=lambda: hesapla("*")
).grid(row=0, column=2, padx=5)

tk.Button(
    buton_frame,
    text="/",
    width=5,
    command=lambda: hesapla("/")
).grid(row=0, column=3, padx=5)

sonuc = tk.Label(
    pencere,
    text="Sonuç:",
    font=("Arial", 14)
)
sonuc.pack()

pencere.mainloop()
```

## Soru (15-19)

**Soru:** Öğrenci verilerini JSON dosyasına kaydedin.

**Yanıt:**

Ders içeriğindeki **Tkinter + JSON** yapısı kullanılarak örnek çözüm:

```python
import tkinter as tk
from tkinter import messagebox
import json
import os

DOSYA = "ogrenciler.json"

def verileri_oku():
    if not os.path.exists(DOSYA):
        return []

    try:
        with open(
            DOSYA,
            "r",
            encoding="utf-8"
        ) as dosya:
            return json.load(dosya)

    except json.JSONDecodeError:
        return []

def kaydet():
    isim = ad.get().strip()
    puan_text = puan_giris.get().strip()

    if isim == "" or puan_text == "":
        messagebox.showwarning(
            "Eksik Bilgi",
            "Tüm alanları doldurun."
        )
        return

    try:
        puan = float(puan_text)
    except ValueError:
        messagebox.showerror(
            "Hata",
            "Puan sayısal olmalıdır."
        )
        return

    ogrenciler = verileri_oku()

    ogrenciler.append({
        "isim": isim,
        "puan": puan
    })

    with open(
        DOSYA,
        "w",
        encoding="utf-8"
    ) as dosya:
        json.dump(
            ogrenciler,
            dosya,
            ensure_ascii=False,
            indent=4
        )

    messagebox.showinfo(
        "Başarılı",
        "Kayıt tamamlandı."
    )

    ad.delete(0, tk.END)
    puan_giris.delete(0, tk.END)

pencere = tk.Tk()
pencere.title("Öğrenci Kayıt Sistemi")
pencere.geometry("450x300")

tk.Label(pencere, text="İsim").grid(
    row=0, column=0, padx=10, pady=10
)
ad = tk.Entry(pencere)
ad.grid(row=0, column=1)

tk.Label(pencere, text="Puan").grid(
    row=1, column=0, padx=10, pady=10
)
puan_giris = tk.Entry(pencere)
puan_giris.grid(row=1, column=1)

tk.Button(
    pencere,
    text="Kaydet",
    command=kaydet
).grid(
    row=2,
    column=0,
    columnspan=2,
    pady=20
)

pencere.mainloop()
```

## Soru (15-23)

**Soru:** Öğrenci puanını 0-100 arasında doğrulayın.

**Yanıt:**

Ders içeriğindeki **Basit Form Doğrulama** yapısı kullanılarak örnek çözüm:

```python
import tkinter as tk
from tkinter import messagebox

def kaydet():
    isim = ad.get().strip()
    puan = puan_giris.get().strip()

    if isim == "" or puan == "":
        messagebox.showwarning(
            "Eksik Bilgi",
            "Tüm alanları doldurun."
        )
        return

    try:
        puan = float(puan)

        if puan < 0 or puan > 100:
            messagebox.showwarning(
                "Geçersiz Puan",
                "Puan 0-100 arasında olmalıdır."
            )
            return

    except ValueError:
        messagebox.showerror(
            "Hata",
            "Puan sayısal olmalıdır."
        )
        return

    messagebox.showinfo(
        "Başarılı",
        "Kayıt doğrulandı."
    )

pencere = tk.Tk()
pencere.geometry("450x280")

tk.Label(pencere, text="Öğrenci Adı").grid(
    row=0, column=0, padx=10, pady=10
)
ad = tk.Entry(pencere)
ad.grid(row=0, column=1)

tk.Label(pencere, text="Puan").grid(
    row=1, column=0, padx=10, pady=10
)
puan_giris = tk.Entry(pencere)
puan_giris.grid(row=1, column=1)

tk.Button(
    pencere,
    text="Kaydet",
    command=kaydet
).grid(
    row=2,
    column=0,
    columnspan=2,
    pady=20
)

pencere.mainloop()
```

## Soru (16-20)

**Yanıt:**

```python
def ara(kelime):
    with sqlite3.connect(DB) as baglanti:
        cursor = baglanti.cursor()

        cursor.execute("""
        SELECT *
        FROM Ogrenciler
        WHERE Isim LIKE ?
        ORDER BY Isim
        """, (f"%{kelime}%",))

        return cursor.fetchall()

for kayit in ara("e"):
    print(kayit)
```

```python
def ogrenci_ekle(isim, sinif, python_puani, matematik_puani):
    with sqlite3.connect("ogrenciler.db") as baglanti:
        cursor = baglanti.cursor()

        cursor.execute("""
        INSERT INTO Ogrenciler (
            Isim, Sinif, PythonPuani, MatematikPuani
        )
        VALUES (?, ?, ?, ?)
        """, (
            isim,
            sinif,
            python_puani,
            matematik_puani
        ))

ogrenci_ekle("Duru", 7, 91, 89)

print("Fonksiyon ile kayıt eklendi.")
```

Ders içeriğindeki **Silme İşlemi** yapısı kullanılarak örnek çözüm:

```python
def kayit_sil():
    global secili_id

    if secili_id is None:
        messagebox.showwarning(
            "Kayıt Seçilmedi",
            "Silmek için önce bir kayıt seçin."
        )
        return

    cevap = messagebox.askyesno(
        "Silme Onayı",
        "Seçili öğrenci kaydı silinsin mi?"
    )

    if not cevap:
        return

    try:
        with sqlite3.connect(DB) as baglanti:
            cursor = baglanti.cursor()

            cursor.execute(
                "DELETE FROM Ogrenciler WHERE Id = ?",
                (secili_id,)
            )

        messagebox.showinfo(
            "Başarılı",
            "Kayıt silindi."
        )

        formu_temizle()
        kayitlari_listele()

    except sqlite3.Error as hata:
        messagebox.showerror(
            "Veritabanı Hatası",
            str(hata)
        )
```

Ders içeriğindeki **Create: Öğrenci Ekleme Route'u** yapısı kullanılarak örnek çözüm:

```python
@app.route("/ogrenci-ekle", methods=["GET", "POST"])
def ogrenci_ekle():
    if request.method == "POST":
        veri, hata = form_dogrula(request.form)

        if hata:
            flash(hata)

        else:
            db = get_db()

            db.execute("""
            INSERT INTO Ogrenciler (
                Isim,
                Sinif,
                PythonPuani,
                MatematikPuani
            )
            VALUES (?, ?, ?, ?)
            """, veri)

            db.commit()

            flash("Öğrenci başarıyla eklendi.")

            return redirect(
                url_for("ogrenci_listesi")
            )

    return render_template(
        "ogrenci_ekle.html"
    )
```

**Yanıt:**

Ders içeriğindeki **`login_required` Decorator** yapısı kullanılarak örnek çözüm:

```python
def login_required(view):
    @functools.wraps(view)
    def wrapped_view(**kwargs):

        if g.user is None:
            flash("Bu sayfaya erişmek için giriş yapmalısınız.")

            return redirect(
                url_for("giris")
            )

        return view(**kwargs)

    return wrapped_view
```

## Soru (20-17)

**Soru:** `admin_required` decorator yazın.

**Yanıt:**

Ders içeriğindeki **`admin_required` Decorator** yapısı kullanılarak örnek çözüm:

```python
def admin_required(view):
    @functools.wraps(view)
    def wrapped_view(**kwargs):

        if g.user is None:
            flash("Önce giriş yapmalısınız.")

            return redirect(
                url_for("giris")
            )

        if g.user["Rol"] != "yonetici":
            flash("Bu işlem için yetkiniz yok.")

            return redirect(
                url_for("ana_sayfa")
            )

        return view(**kwargs)

    return wrapped_view
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(
    df
)

print(
    X_scaled[:5]
)
```

## Soru (25-04)

**Soru:** K=2 ile K-Means oluşturun.

**Yanıt:**

Ders içeriğindeki **K=2 ile K=8 Arasında Silhouette Score** yapısı kullanılarak örnek çözüm:

```python
silhouette_sonuclari = []

for k in range(2, 9):
    model_k = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    etiket = model_k.fit_predict(
        X_scaled
    )

    skor = silhouette_score(
        X_scaled,
        etiket
    )

    silhouette_sonuclari.append({
        "K": k,
        "Silhouette": skor
    })

silhouette_df = pd.DataFrame(
    silhouette_sonuclari
)

silhouette_df
```

## Soru (25-05)

**Soru:** K=3 ile K-Means oluşturun.

**Yanıt:**

Ders içeriğindeki **İlk K-Means Modeli** yapısı kullanılarak örnek çözüm:

```python
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

kmeans.fit(
    X_scaled
)

print(
    "Model eğitildi."
)
```

## Soru (25-06)

**Soru:** K=4 ile K-Means oluşturun.

**Yanıt:**

Ders içeriğindeki **K=4 Sonucunu İncelemek** yapısı kullanılarak örnek çözüm:

```python
model_k4 = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

etiket_k4 = model_k4.fit_predict(
    X_scaled
)

print(
    "K=4 Silhouette:",
    silhouette_score(
        X_scaled,
        etiket_k4
    )
)
```

## Soru (25-07)

**Soru:** Her modelin inertia değerini bulun.

**Yanıt:**

## Soru (25-08)

**Soru:** Her modelin silhouette score değerini bulun.

**Yanıt:**

Ders içeriğindeki **Silhouette Score Grafiği** yapısı kullanılarak örnek çözüm:

```python
plt.plot(
    silhouette_df["K"],
    silhouette_df["Silhouette"],
    marker="o"
)

plt.xlabel("K")
plt.ylabel("Silhouette Score")
plt.title("K Değerine Göre Silhouette Score")
plt.grid()
plt.show()
```

## Soru (25-09)

**Soru:** K=1 ile K=10 arasında elbow grafiği oluşturun.

**Yanıt:**

Ders içeriğindeki **K=1 ile K=9 Arasında Inertia Hesaplamak** yapısı kullanılarak örnek çözüm:

```python
inertia_sonuclari = []

for k in range(1, 10):
    model_k = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model_k.fit(
        X_scaled
    )

    inertia_sonuclari.append({
        "K": k,
        "Inertia": model_k.inertia_
    })

inertia_df = pd.DataFrame(
    inertia_sonuclari
)

inertia_df
```

## Soru (25-10)

**Soru:** K=2 ile K=9 arasında silhouette grafiği oluşturun.

**Yanıt:**

Ders içeriğindeki **K=2 ile K=8 Arasında Silhouette Score** yapısı kullanılarak örnek çözüm:

```python
silhouette_sonuclari = []

for k in range(2, 9):
    model_k = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    etiket = model_k.fit_predict(
        X_scaled
    )

    skor = silhouette_score(
        X_scaled,
        etiket
    )

    silhouette_sonuclari.append({
        "K": k,
        "Silhouette": skor
    })

silhouette_df = pd.DataFrame(
    silhouette_sonuclari
)

silhouette_df
```

## Soru (25-11)

**Soru:** En yüksek silhouette skorunu otomatik bulun.

**Yanıt:**

Ders içeriğindeki **En Yüksek Silhouette Değerini Bulmak** yapısı kullanılarak örnek çözüm:

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

lojistik = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),
    (
        "model",
        LogisticRegression(
            max_iter=3000,
            random_state=42
        )
    )
])

print(lojistik)
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),
    (
        "model",
        KNeighborsClassifier(
            n_neighbors=5
        )
    )
])

knn_cv = cross_validate(
    knn,
    X_gelistirme,
    y_gelistirme,
    cv=cv,
    scoring=skorlar,
    n_jobs=-1
)

print(
    "KNN Accuracy:",
    knn_cv[
        "test_accuracy"
    ].mean()
)

print(
    "KNN Macro F1:",
    knn_cv[
        "test_f1"
    ].mean()
)
```

## Soru (27-03)

**Soru:** Noktalama işaretlerini temizleyin.

**Yanıt:**

```python
import re

def noktalama_temizle(metin):
    return re.sub(
        r"[^a-zA-ZçğıöşüÇĞİÖŞÜ0-9\s]",
        " ",
        metin
    )

print(
    noktalama_temizle(
        "Merhaba, Python!"
    )
)
```

## Soru (27-04)

**Soru:** Kelimelere ayırın.

**Yanıt:**

Ders içeriğindeki **Ortalama Kelime Uzunluğu** yapısı kullanılarak örnek çözüm:

```python
def ortalama_kelime_uzunlugu(
    metin
):
    kelimeler = (
        metin_temizle(
            metin
        )
        .split()
    )

    if not kelimeler:
        return 0

    return sum(
        len(kelime)
        for kelime in kelimeler
    ) / len(kelimeler)
```

## Soru (28-05)

**Soru:** DummyClassifier baseline oluşturun.

**Yanıt:**

Ders içeriğindeki **Baseline Model** yapısı kullanılarak örnek çözüm:

```python
from sklearn.dummy import DummyClassifier
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer

baseline = Pipeline([
    (
        "tfidf",
        TfidfVectorizer()
    ),
    (
        "model",
        DummyClassifier(
            strategy="most_frequent"
        )
    )
])

baseline.fit(
    X_train,
    y_train
)

baseline_pred = baseline.predict(
    X_test
)

print(
    baseline.score(
        X_test,
        y_test
    )
)
```

## Soru (28-06)

**Soru:** MultinomialNB modeli oluşturun.

**Yanıt:**

Ders içeriğindeki **Model 1: Multinomial Naive Bayes** yapısı kullanılarak örnek çözüm:

```python
from sklearn.naive_bayes import MultinomialNB

nb_model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            ngram_range=(
                1,
                2
            )
        )
    ),
    (
        "model",
        MultinomialNB()
    )
])

nb_model.fit(
    X_train,
    y_train
)

nb_pred = nb_model.predict(
    X_test
)

print(
    "Model eğitildi."
)
```

## Soru (28-07)

**Soru:** Logistic Regression modeli oluşturun.

**Yanıt:**

Ders içeriğindeki **Model 2: Logistic Regression** yapısı kullanılarak örnek çözüm:

```python
from sklearn.linear_model import LogisticRegression

lojistik_model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            ngram_range=(
                1,
                2
            )
        )
    ),
    (
        "model",
        LogisticRegression(
            max_iter=1500,
            random_state=42
        )
    )
])

lojistik_model.fit(
    X_train,
    y_train
)

lojistik_pred = lojistik_model.predict(
    X_test
)
```

## Soru (28-08)

**Soru:** LinearSVC modeli oluşturun.

**Yanıt:**

Ders içeriğindeki **Model 3: Linear SVM** yapısı kullanılarak örnek çözüm:

```python
from sklearn.svm import LinearSVC

svm_model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            ngram_range=(
                1,
                2
            )
        )
    ),
    (
        "model",
        LinearSVC(
            random_state=42
        )
    )
])

svm_model.fit(
    X_train,
    y_train
)

svm_pred = svm_model.predict(
    X_test
)
```

## Soru (28-17)

**Soru:** Unigram ve bigram modellerini karşılaştırın.

**Yanıt:**

Ders içeriğindeki **Unigram ve Bigram Karşılaştırması** yapısı kullanılarak örnek çözüm:

```python
unigram_model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            ngram_range=(
                1,
                1
            )
        )
    ),
    (
        "model",
        LogisticRegression(
            max_iter=1500,
            random_state=42
        )
    )
])

unigram_model.fit(
    X_train,
    y_train
)

unigram_pred = unigram_model.predict(
    X_test
)

print(
    "Unigram F1:",
    f1_score(
        y_test,
        unigram_pred,
        pos_label="spam"
    )
)

print(
    "Unigram + Bigram F1:",
    f1_score(
        y_test,
        lojistik_pred,
        pos_label="spam"
    )
)
```

## Soru (28-18)

**Soru:** Character n-gram modeli oluşturun.

**Yanıt:**

Ders içeriğindeki **Character N-Gram Modeli** yapısı kullanılarak örnek çözüm:

```python
char_model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            analyzer="char_wb",
            ngram_range=(
                3,
                5
            ),
            min_df=2
        )
    ),
    (
        "model",
        LogisticRegression(
            max_iter=1500,
            random_state=42
        )
    )
])

char_model.fit(
    X_train,
    y_train
)

char_pred = char_model.predict(
    X_test
)

print(
    "Character N-Gram F1:",
    f1_score(
        y_test,
        char_pred,
        pos_label="spam"
    )
)
```

## Soru (28-19)

**Soru:** Cross Validation yapın.

**Yanıt:**

Ders içeriğindeki **Cross Validation ile Daha Sağlam Karşılaştırma** yapısı kullanılarak örnek çözüm:

```python
from sklearn.model_selection import (
    StratifiedKFold,
    cross_val_score
)

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_f1 = cross_val_score(
    lojistik_model,
    X,
    y,
    cv=cv,
    scoring="f1_macro"
)

print(
    cv_f1
)

print(
    "Ortalama:",
    cv_f1.mean()
)
```

## Soru (28-20)

**Soru:** GridSearchCV ile n-gram ayarı arayın.

**Yanıt:**

Ders içeriğindeki **GridSearchCV ile TF-IDF Ayarlarını Aramak** yapısı kullanılarak örnek çözüm:

```python
from sklearn.model_selection import GridSearchCV

arama_pipeline = Pipeline([
    (
        "tfidf",
        TfidfVectorizer()
    ),
    (
        "model",
        LogisticRegression(
            max_iter=1500,
            random_state=42
        )
    )
])

param_grid = {
    "tfidf__ngram_range": [
        (1, 1),
        (1, 2)
    ],
    "tfidf__min_df": [
        1,
        2
    ],
    "model__C": [
        0.5,
        1.0,
        2.0
    ]
}

grid = GridSearchCV(
    arama_pipeline,
    param_grid=param_grid,
    cv=cv,
    scoring="f1_macro",
    n_jobs=-1
)

grid.fit(
    X,
    y
)

print(
    grid.best_params_
)

print(
    grid.best_score_
)
```

## Soru (30-03)

**Soru:** Sigmoid fonksiyonu yazın.

**Yanıt:**

Ders içeriğindeki **Sigmoid Fonksiyonu** yapısı kullanılarak örnek çözüm:

```python
def sigmoid(x):
    return (
        1
        /
        (
            1
            +
            np.exp(-x)
        )
    )

for deger in [
    -5,
    -2,
    0,
    2,
    5
]:
    print(
        deger,
        "->",
        sigmoid(deger)
    )
```

## Soru (30-04)

**Soru:** ReLU fonksiyonu yazın.

**Yanıt:**

Ders içeriğindeki **ReLU Fonksiyonu** yapısı kullanılarak örnek çözüm:

```python
def relu(x):
    return np.maximum(
        0,
        x
    )

print(
    relu(
        np.array([
            -3,
            -1,
            0,
            2,
            5
        ])
    )
)
```

## Soru (30-11)

**Soru:** Basit gradient descent örneğini farklı learning rate ile deneyin.

**Yanıt:**

Ders içeriğindeki **Basit Gradient Descent Örneği** yapısı kullanılarak örnek çözüm:

```python
x = 2.0
gercek_y = 8.0

w = 1.0
learning_rate = 0.1

for adim in range(10):
    tahmin_y = (
        w
        *
        x
    )

    hata = (
        tahmin_y
        -
        gercek_y
    )

    loss = (
        hata
        ** 2
    )

    gradient = (
        2
        *
        hata
        *
        x
    )

    w = (
        w
        -
        learning_rate
        *
        gradient
    )

    print(
        adim + 1,
        "w:",
        round(w, 4),
        "loss:",
        round(loss, 4)
    )
```

## Soru (31-02)

**Soru:** Tek convolution işlemini elle hesaplayın.

**Yanıt:**

```python
goruntu_parcasi = np.array([
    [1, 1, 0],
    [1, 1, 0],
    [0, 0, 0]
], dtype=float)

kernel = np.array([
    [1, 0, -1],
    [1, 0, -1],
    [1, 0, -1]
], dtype=float)

sonuc = np.sum(
    goruntu_parcasi
    *
    kernel
)

print(
    "Convolution sonucu:",
    sonuc
)
```

## Soru (31-03)

**Soru:** Basit convolution fonksiyonu yazın.

**Yanıt:**

```python
def convolution2d(
    goruntu,
    kernel
):
    h, w = goruntu.shape
    kh, kw = kernel.shape

    cikti = np.zeros(
        (
            h - kh + 1,
            w - kw + 1
        )
    )

    for y in range(
        cikti.shape[0]
    ):
        for x in range(
            cikti.shape[1]
        ):
            bolge = goruntu[
                y:y + kh,
                x:x + kw
            ]

            cikti[y, x] = np.sum(
                bolge
                *
                kernel
            )

    return cikti
```

## Soru (32-24)

**Soru:** GlobalAveragePooling2D + Dense head ekleyin.

**Yanıt:**

Ders içeriğindeki **MobileNetV2 Transfer Modeli** yapısı kullanılarak örnek çözüm:

```python
inputs = keras.Input(
    shape=(
        96,
        96,
        3
    )
)

x = shape_augmentation(
    inputs
)

x = keras.applications.mobilenet_v2.preprocess_input(
    x
)

x = mobilenet_base(
    x,
    training=False
)

x = layers.GlobalAveragePooling2D()(
    x
)

x = layers.Dropout(
    0.25
)(
    x
)

outputs = layers.Dense(
    3,
    activation="softmax"
)(
    x
)

mobilenet_transfer = keras.Model(
    inputs,
    outputs,
    name="mobilenet_transfer"
)

mobilenet_transfer.summary()
```

**Soru:** Context uzunluğunu sınırlayan fonksiyon yazın.

**Yanıt:**

Ders içeriğindeki **Basit Context Bütçesi Simülasyonu** yapısı kullanılarak örnek çözüm:

```python
def context_sinirla(
    tokenlar,
    maksimum=12
):
    if len(
        tokenlar
    ) <= maksimum:
        return tokenlar

    return tokenlar[
        -maksimum:
    ]

uzun_metin = (
    "python ile yapay zeka öğreniyoruz "
    "sonra veri analizi yapıyoruz "
    "ardından web uygulaması geliştiriyoruz"
)

uzun_tokenlar = basit_tokenize(
    uzun_metin
)

print(
    "Tüm token:",
    uzun_tokenlar
)

print()

print(
    "Context:",
    context_sinirla(
        uzun_tokenlar,
        maksimum=12
    )
)
```

## Soru (33-08)

**Soru:** Sonraki token olasılıklarını hesaplayın.

**Yanıt:**

Ders içeriğindeki **En Olası Tokenı Seçmek** yapısı kullanılarak örnek çözüm:

```python
def en_olasi_token(
    token
):
    olasiliklar = (
        sonraki_token_olasiliklari(
            token
        )
    )

    if not olasiliklar:
        return None

    return max(
        olasiliklar,
        key=olasiliklar.get
    )

print(
    en_olasi_token(
        "python"
    )
)
```

## Soru (33-09)

**Soru:** Greedy metin üretici yazın.

**Yanıt:**

Ders içeriğindeki **Greedy Metin Üretimi** yapısı kullanılarak örnek çözüm:

```python
def greedy_uret(
    baslangic,
    adim=10
):
    sonuc = [
        baslangic
    ]

    mevcut = baslangic

    for _ in range(
        adim
    ):
        yeni = en_olasi_token(
            mevcut
        )

        if yeni is None:
            break

        sonuc.append(
            yeni
        )

        mevcut = yeni

    return " ".join(
        sonuc
    )

print(
    greedy_uret(
        "python",
        adim=12
    )
)
```

## Soru (33-10)

**Soru:** Sampling tabanlı üretici yazın.

**Yanıt:**

Ders içeriğindeki **Sampling ile Metin Üretmek** yapısı kullanılarak örnek çözüm:

```python
def sampling_uret(
    baslangic,
    adim=10,
    seed=42
):
    yerel_rng = (
        np.random.default_rng(
            seed
        )
    )

    sonuc = [
        baslangic
    ]

    mevcut = baslangic

    for _ in range(
        adim
    ):
        yeni = rastgele_sonraki_token(
            mevcut,
            yerel_rng
        )

        if yeni is None:
            break

        sonuc.append(
            str(
                yeni
            )
        )

        mevcut = str(
            yeni
        )

    return " ".join(
        sonuc
    )

for seed in [
    1,
    2,
    3
]:
    print(
        sampling_uret(
            "python",
            adim=12,
            seed=seed
        )
    )
```

## Soru (34-04)

**Soru:** `previous_response_id` tabanlı konuşma tasarlayın.

**Yanıt:**

Ders içeriğindeki **Çok Turlu Asistan Fonksiyonu** yapısı kullanılarak örnek çözüm:

```python
def konusmali_cevap(
    mesaj,
    durum,
    instructions=INSTRUCTIONS,
):
    if not api_hazir_mi()["hazir"]:
        return {
            "durum": "simulasyon",
            "previous_response_id":
                durum.previous_response_id,
            "input": mesaj,
        }

    from openai import OpenAI

    client = OpenAI()

    parametreler = {
        "model": MODEL_ADI,
        "instructions": instructions,
        "input": mesaj,
    }

    if (
        durum.previous_response_id
        is not None
    ):
        parametreler[
            "previous_response_id"
        ] = (
            durum.previous_response_id
        )

    response = (
        client.responses.create(
            **parametreler
        )
    )

    durum.guncelle(
        response.id
    )

    return {
        "durum": "basarili",
        "text": response.output_text,
        "response_id": response.id,
    }
```

**Soru:** Geçmiş mesajları sınırlayan fonksiyon yazın.

**Yanıt:**

Ders içeriğindeki **Context Bütçesi** yapısı kullanılarak örnek çözüm:

```python
def gecmisi_sinirla(
    mesajlar,
    maksimum_karakter=500,
):
    secilen = []
    toplam = 0

    for mesaj in reversed(
        mesajlar
    ):
        uzunluk = len(
            mesaj["content"]
        )

        if (
            toplam
            + uzunluk
            >
            maksimum_karakter
        ):
            break

        secilen.append(
            mesaj
        )

        toplam += uzunluk

    return list(
        reversed(
            secilen
        )
    )

print(
    gecmisi_sinirla(
        konusma_gecmisi,
        maksimum_karakter=100
    )
)
```

```python
def backoff_sureleri(
    deneme_sayisi=5,
    taban=1.0,
):
    return [
        taban
        *
        (
            2 ** i
        )
        for i in range(
            deneme_sayisi
        )
    ]

print(
    backoff_sureleri()
)
```

## Soru (35-01)

**Soru:** Üç farklı metin dokümanı oluşturun.

**Yanıt:**

Ders içeriğindeki **Basit Duplicate Kontrolü** yapısı kullanılarak örnek çözüm:

```python
def benzersiz_metinler(
    sonuclar,
):
    gorulen = set()
    kayitlar = []

    for _, satir in (
        sonuclar.iterrows()
    ):
        metin = satir[
            "metin"
        ].strip()

        if metin in gorulen:
            continue

        gorulen.add(
            metin
        )

        kayitlar.append(
            satir.to_dict()
        )

    return pd.DataFrame(
        kayitlar
    )
```

## Soru (35-02)

**Soru:** Kelime tabanlı chunking fonksiyonu yazın.

**Yanıt:**

Ders içeriğindeki **Kelime Tabanlı Chunking Fonksiyonu** yapısı kullanılarak örnek çözüm:

```python
def kelime_chunkla(
    metin,
    chunk_boyutu=30,
    overlap=8,
):
    if chunk_boyutu <= 0:
        raise ValueError(
            "chunk_boyutu pozitif olmalıdır."
        )

    if overlap < 0:
        raise ValueError(
            "overlap negatif olamaz."
        )

    if overlap >= chunk_boyutu:
        raise ValueError(
            "overlap chunk_boyutundan küçük olmalıdır."
        )

    kelimeler = metin.split()

    adim = (
        chunk_boyutu
        -
        overlap
    )

    chunklar = []

    for baslangic in range(
        0,
        len(kelimeler),
        adim,
    ):
        parca = kelimeler[
            baslangic:
            baslangic + chunk_boyutu
        ]

        if not parca:
            continue

        chunklar.append(
            " ".join(
                parca
            )
        )

        if (
            baslangic
            + chunk_boyutu
            >=
            len(kelimeler)
        ):
            break

    return chunklar
```

## Soru (35-07)

**Soru:** Cosine similarity ile sorgu arayın.

**Yanıt:**

Ders içeriğindeki **Cosine Similarity'yi NumPy ile Yazmak** yapısı kullanılarak örnek çözüm:

```python
def cosine_similarity_numpy(
    a,
    b,
):
    a = np.asarray(
        a,
        dtype=float,
    )

    b = np.asarray(
        b,
        dtype=float,
    )

    payda = (
        np.linalg.norm(a)
        *
        np.linalg.norm(b)
    )

    if payda == 0:
        return 0.0

    return float(
        np.dot(
            a,
            b
        )
        /
        payda
    )

print(
    cosine_similarity_numpy(
        [1, 0],
        [1, 0]
    )
)

print(
    cosine_similarity_numpy(
        [1, 0],
        [0, 1]
    )
)
```

## Soru (35-09)

**Soru:** Minimum retrieval threshold ekleyin.

**Yanıt:**

Ders içeriğindeki **Threshold Denemesi** yapısı kullanılarak örnek çözüm:

```python
def rejection_rate(
    retriever,
    sorular,
    threshold,
):
    reddedilen = 0

    for soru in sorular:
        sonuc = retriever.search(
            soru,
            top_k=1,
            min_score=threshold,
        )

        if sonuc.empty:
            reddedilen += 1

    return (
        reddedilen
        /
        len(
            sorular
        )
    )

for threshold in [
    0.05,
    0.10,
    0.15,
    0.20,
]:
    print(
        threshold,
        "->",
        rejection_rate(
            retriever,
            cevapsiz_sorular,
            threshold,
        )
    )
```

```python
sonuclar = []

for overlap in [
    2,
    8
]:
    df_temp = dokumanlari_chunkla(
        dokumanlar,
        chunk_boyutu=30,
        overlap=overlap
    )

    retriever_temp = (
        TfidfRetriever(
            df_temp
        )
    )

    sonuclar.append({
        "overlap":
            overlap,
        "recall_3":
            recall_at_k(
                retriever_temp,
                retrieval_eval,
                k=3
            )
    })

pd.DataFrame(
    sonuclar
)
```

## Soru (36-04)

**Yanıt:**

Ders içeriğindeki **Kullanıcı Girdisi Doğrulama** yapısı kullanılarak örnek çözüm:

```python
def soru_dogrula(soru, maksimum=1000):
    if not isinstance(soru, str):
        return False, "Soru metin olmalıdır."

    temiz = soru.strip()

    if not temiz:
        return False, "Soru boş olamaz."

    if len(temiz) > maksimum:
        return False, f"Soru en fazla {maksimum} karakter olabilir."

    return True, temiz

print(soru_dogrula("Python Kulübü nerede?"))
print(soru_dogrula("   "))
```

## Soru (36-05)

**Soru:** SQLite `documents` tablosu oluşturun.

**Yanıt:**

Ders içeriğindeki **Dokümanları Okumak** yapısı kullanılarak örnek çözüm:

```python
def dokumanlari_getir(db_yolu=DB_YOLU):
    with sqlite3.connect(db_yolu) as conn:
        conn.row_factory = sqlite3.Row
        rows = conn.execute(
            '''
            SELECT id, filename, title, content
            FROM documents
            ORDER BY id
            '''
        ).fetchall()

    return [dict(row) for row in rows]

db_dokumanlar = dokumanlari_getir()
pd.DataFrame(db_dokumanlar)
```

## Soru (36-09)

**Soru:** TF-IDF retriever oluşturun.

**Yanıt:**

Ders içeriğindeki **RAGService** yapısı kullanılarak örnek çözüm:

```python
class RAGService:
    def __init__(
        self,
        retriever,
        llm_service,
        top_k=3,
        min_score=0.10,
    ):
        self.retriever = retriever
        self.llm_service = llm_service
        self.top_k = top_k
        self.min_score = min_score

    def answer(self, question):
        results = self.retriever.search(
            question,
            top_k=self.top_k,
            min_score=self.min_score,
        )

        if results.empty:
            return {
                "answer": "Bu bilgi verilen dokümanlarda bulunmuyor.",
                "sources": [],
                "mode": "no_source",
            }

        context = context_olustur(results)

        llm_answer = self.llm_service.cevapla(
            question,
            context,
        )

        if llm_answer is None:
            answer = yerel_cevap(results)
            mode = "local_rag"
        else:
            answer = llm_answer
            mode = "llm_rag"

        return {
            "answer": answer,
            "sources": kaynak_listesi(results),
            "mode": mode,
        }

rag_service = RAGService(
    retriever,
    LLMService(enabled=False),
)

rag_service.answer(
    "Laboratuvarda içecek serbest mi?"
)
```

```text
(Ders No - Soru No)
```

biçiminde numaralandırılmıştır.

```text
(12-05)
```

1. Önce soruyu cevap anahtarına bakmadan çözün.
2. Kendi kodunuzu çalıştırın.
3. Sonucu örnek yanıtla karşılaştırın.
4. İki çözüm farklıysa neden farklı olduğunu inceleyin.
5. Daha kısa, daha okunabilir veya daha güvenli bir çözüm geliştirmeye çalışın.
